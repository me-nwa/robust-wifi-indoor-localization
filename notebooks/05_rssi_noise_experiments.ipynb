```python
# ============================================================
# DEN702 Research Paper
# Notebook 05 — RSSI Noise Robustness Experiments
#
# Research Topic:
# Robust Wi-Fi Fingerprint-Based Indoor Localization
# Under Access-Point Loss and RSSI Noise
#
# Research Question:
# How does RSSI measurement noise affect indoor localization
# performance?
#
# Purpose:
# 1. Load clean models from Notebook 03
# 2. Add Gaussian noise to detected RSSI values only
# 3. Evaluate classification robustness
# 4. Evaluate coordinate-localization robustness
# 5. Repeat experiments using multiple random seeds
# 6. Report mean ± standard deviation
# 7. Save all result tables and figures
#
# IMPORTANT:
# - Missing AP values remain at -110 dBm
# - Noise is applied only to originally detected APs
# - Models are not retrained
# - Final test set remains untouched
# ============================================================


# ------------------------------------------------------------
# Step 1 — Import Required Libraries
# ------------------------------------------------------------

from pathlib import Path
import time
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from IPython.display import display

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    mean_absolute_error,
    mean_squared_error
)


# ------------------------------------------------------------
# Step 2 — Reproducibility Configuration
# ------------------------------------------------------------

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


RSSI_NOISE_LEVELS = [
    0,
    2,
    5,
    10
]


EXPERIMENT_SEEDS = [
    42,
    43,
    44,
    45,
    46
]


MISSING_RSSI = -110


print("=" * 76)
print("DEN702 — RSSI NOISE ROBUSTNESS EXPERIMENT")
print("=" * 76)

print("RSSI noise levels (sigma dB):", RSSI_NOISE_LEVELS)
print("Experiment seeds:", EXPERIMENT_SEEDS)


# ------------------------------------------------------------
# Step 3 — Detect Project Root
# ------------------------------------------------------------

if Path("../data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path("..").resolve()

elif Path("data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path(".").resolve()

else:

    raise FileNotFoundError(
        "Could not locate uji_preprocessed.npz."
    )


PROCESSED_DATA_DIR = (
    PROJECT_ROOT / "data" / "processed"
)

MODEL_DIR = (
    PROJECT_ROOT / "models" / "03_clean_baselines"
)

RESULTS_CSV_DIR = (
    PROJECT_ROOT / "results" / "csv"
)

RESULTS_FIGURES_DIR = (
    PROJECT_ROOT / "results" / "figures"
)


RESULTS_CSV_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# Step 4 — Load Development Dataset
# ------------------------------------------------------------

data = np.load(
    PROCESSED_DATA_DIR / "uji_preprocessed.npz",
    allow_pickle=True
)


X_dev_raw = data["X_dev_raw"]

observed_mask_dev = data[
    "observed_mask_dev"
]

y_class_dev = data[
    "y_class_dev"
]

y_coord_dev = data[
    "y_coord_dev"
]


print("\nDevelopment shape:", X_dev_raw.shape)
print("Observed mask shape:", observed_mask_dev.shape)


# ------------------------------------------------------------
# Step 5 — Load Original RSSI Scaler
# ------------------------------------------------------------

rssi_scaler = joblib.load(
    PROCESSED_DATA_DIR
    / "uji_minmax_scaler.joblib"
)


# ------------------------------------------------------------
# Step 6 — Load Classification Models
# ------------------------------------------------------------

classification_models = {

    "KNN": joblib.load(
        MODEL_DIR
        / "knn_classifier.joblib"
    ),

    "Random Forest": joblib.load(
        MODEL_DIR
        / "random_forest_classifier.joblib"
    ),

    "MLP": joblib.load(
        MODEL_DIR
        / "mlp_classifier.joblib"
    )
}


# ------------------------------------------------------------
# Step 7 — Load Regression Models
# ------------------------------------------------------------

regression_models = {

    "KNN": joblib.load(
        MODEL_DIR
        / "knn_regressor.joblib"
    ),

    "Random Forest": joblib.load(
        MODEL_DIR
        / "random_forest_regressor.joblib"
    ),

    "MLP": joblib.load(
        MODEL_DIR
        / "mlp_regressor.joblib"
    )
}


coord_scaler = joblib.load(
    MODEL_DIR
    / "coordinate_scaler.joblib"
)


# ------------------------------------------------------------
# Step 8 — Define RSSI Noise Function
#
# Gaussian noise:
#
#     RSSI_noisy = RSSI + N(0, sigma^2)
#
# Noise is added ONLY to detected AP measurements.
#
# Missing APs remain -110 dBm.
#
# Noisy observed values are clipped to the valid observed
# RSSI range [-104, 0].
# ------------------------------------------------------------

def apply_rssi_noise(
    X_raw,
    observed_mask,
    sigma,
    seed
):

    if sigma < 0:

        raise ValueError(
            "Noise sigma must be non-negative."
        )


    if sigma == 0:

        return X_raw.copy()


    rng = np.random.default_rng(
        seed
    )


    X_noisy = (
        X_raw.copy().astype(float)
    )


    noise = rng.normal(
        loc=0.0,
        scale=sigma,
        size=X_noisy.shape
    )


    X_noisy[
        observed_mask
    ] = (

        X_noisy[
            observed_mask
        ]

        +

        noise[
            observed_mask
        ]
    )


    # Clip detected RSSI values only

    X_noisy[
        observed_mask
    ] = np.clip(

        X_noisy[
            observed_mask
        ],

        -104,

        0
    )


    # Explicitly preserve missing AP values

    X_noisy[
        ~observed_mask
    ] = MISSING_RSSI


    return X_noisy


# ------------------------------------------------------------
# Step 9 — Verify Noise Function
# ------------------------------------------------------------

example = apply_rssi_noise(
    X_dev_raw[:1],
    observed_mask_dev[:1],
    sigma=10,
    seed=RANDOM_STATE
)


print("\nNoise function verification:")

print(
    "Missing APs preserved:",
    np.all(
        example[
            ~observed_mask_dev[:1]
        ]
        == MISSING_RSSI
    )
)


# ------------------------------------------------------------
# Step 10 — Classification Evaluation Function
# ------------------------------------------------------------

def evaluate_classification(
    model,
    X,
    y_true
):

    predictions = model.predict(
        X
    )


    return {

        "Accuracy":
            accuracy_score(
                y_true,
                predictions
            ),

        "Precision_Macro":
            precision_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            ),

        "Recall_Macro":
            recall_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            ),

        "F1_Macro":
            f1_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            )
    }


# ------------------------------------------------------------
# Step 11 — Regression Evaluation Function
# ------------------------------------------------------------

def evaluate_regression(
    model_name,
    model,
    X,
    y_true
):

    predictions = model.predict(
        X
    )


    if model_name == "MLP":

        predictions = (
            coord_scaler.inverse_transform(
                predictions
            )
        )


    coordinate_mae = (
        mean_absolute_error(
            y_true,
            predictions
        )
    )


    coordinate_rmse = np.sqrt(
        mean_squared_error(
            y_true,
            predictions
        )
    )


    euclidean_errors = np.sqrt(
        np.sum(
            (
                predictions
                - y_true
            ) ** 2,
            axis=1
        )
    )


    return {

        "Coordinate_MAE":
            coordinate_mae,

        "Coordinate_RMSE":
            coordinate_rmse,

        "Mean_Localization_Error":
            np.mean(
                euclidean_errors
            ),

        "Median_Localization_Error":
            np.median(
                euclidean_errors
            ),

        "P90_Localization_Error":
            np.percentile(
                euclidean_errors,
                90
            )
    }


# ------------------------------------------------------------
# Step 12 — Run RSSI Noise Experiments
# ------------------------------------------------------------

classification_records = []
regression_records = []


print("\n" + "=" * 76)
print("STEP 12 — RUNNING RSSI NOISE EXPERIMENTS")
print("=" * 76)


start_time = time.time()


for sigma in RSSI_NOISE_LEVELS:

    print(
        f"\nRSSI noise sigma: {sigma} dB"
    )


    for seed in EXPERIMENT_SEEDS:


        X_noisy_raw = apply_rssi_noise(

            X_dev_raw,

            observed_mask_dev,

            sigma=sigma,

            seed=seed
        )


        X_noisy_scaled = (
            rssi_scaler.transform(
                X_noisy_raw
            )
        )


        # Classification

        for (
            model_name,
            model
        ) in classification_models.items():


            metrics = evaluate_classification(

                model,
                X_noisy_scaled,
                y_class_dev
            )


            classification_records.append(
                {

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    "Model":
                        model_name,

                    **metrics
                }
            )


        # Regression

        for (
            model_name,
            model
        ) in regression_models.items():


            metrics = evaluate_regression(

                model_name,
                model,
                X_noisy_scaled,
                y_coord_dev
            )


            regression_records.append(
                {

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    "Model":
                        model_name,

                    **metrics
                }
            )


total_time = (
    time.time()
    - start_time
)


print(
    "\nRSSI noise experiments completed in",
    round(total_time, 2),
    "seconds"
)


# ------------------------------------------------------------
# Step 13 — Create Detailed Tables
# ------------------------------------------------------------

classification_detailed = pd.DataFrame(
    classification_records
)

regression_detailed = pd.DataFrame(
    regression_records
)


display(
    classification_detailed.head(
        12
    ).round(4)
)


display(
    regression_detailed.head(
        12
    ).round(4)
)


# ------------------------------------------------------------
# Step 14 — Save Detailed Results
# ------------------------------------------------------------

CLASS_DETAILED_FILE = (
    RESULTS_CSV_DIR
    / "05_rssi_noise_classification_detailed.csv"
)


REG_DETAILED_FILE = (
    RESULTS_CSV_DIR
    / "05_rssi_noise_regression_detailed.csv"
)


classification_detailed.to_csv(
    CLASS_DETAILED_FILE,
    index=False
)


regression_detailed.to_csv(
    REG_DETAILED_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 15 — Summarize Mean ± Standard Deviation
# ------------------------------------------------------------

classification_summary = (

    classification_detailed

    .groupby(
        [
            "Model",
            "RSSI_Noise_Sigma_dB"
        ],
        as_index=False
    )

    .agg(

        Accuracy_Mean=(
            "Accuracy",
            "mean"
        ),

        Accuracy_SD=(
            "Accuracy",
            "std"
        ),

        F1_Mean=(
            "F1_Macro",
            "mean"
        ),

        F1_SD=(
            "F1_Macro",
            "std"
        ),

        Precision_Mean=(
            "Precision_Macro",
            "mean"
        ),

        Recall_Mean=(
            "Recall_Macro",
            "mean"
        )
    )
)


regression_summary = (

    regression_detailed

    .groupby(
        [
            "Model",
            "RSSI_Noise_Sigma_dB"
        ],
        as_index=False
    )

    .agg(

        Coordinate_MAE_Mean=(
            "Coordinate_MAE",
            "mean"
        ),

        Coordinate_MAE_SD=(
            "Coordinate_MAE",
            "std"
        ),

        Coordinate_RMSE_Mean=(
            "Coordinate_RMSE",
            "mean"
        ),

        Mean_Localization_Error_Mean=(
            "Mean_Localization_Error",
            "mean"
        ),

        Mean_Localization_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),

        Median_Localization_Error_Mean=(
            "Median_Localization_Error",
            "mean"
        ),

        P90_Localization_Error_Mean=(
            "P90_Localization_Error",
            "mean"
        )
    )
)


classification_summary = (
    classification_summary.fillna(0)
)

regression_summary = (
    regression_summary.fillna(0)
)


print("\nClassification summary:")

display(
    classification_summary.round(4)
)


print("\nRegression summary:")

display(
    regression_summary.round(4)
)


# ------------------------------------------------------------
# Step 16 — Save Summary Results
# ------------------------------------------------------------

CLASS_SUMMARY_FILE = (
    RESULTS_CSV_DIR
    / "05_rssi_noise_classification_summary.csv"
)


REG_SUMMARY_FILE = (
    RESULTS_CSV_DIR
    / "05_rssi_noise_regression_summary.csv"
)


classification_summary.to_csv(
    CLASS_SUMMARY_FILE,
    index=False
)


regression_summary.to_csv(
    REG_SUMMARY_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 17 — Calculate Classification Degradation
# ------------------------------------------------------------

classification_delta = (
    classification_summary.copy()
)


classification_delta[
    "Accuracy_Drop"
] = 0.0


classification_delta[
    "F1_Drop"
] = 0.0


for model_name in (
    classification_delta[
        "Model"
    ].unique()
):


    clean_row = (
        classification_delta[
            (
                classification_delta[
                    "Model"
                ]
                == model_name
            )
            &
            (
                classification_delta[
                    "RSSI_Noise_Sigma_dB"
                ]
                == 0
            )
        ]
    )


    clean_accuracy = (
        clean_row[
            "Accuracy_Mean"
        ].iloc[0]
    )


    clean_f1 = (
        clean_row[
            "F1_Mean"
        ].iloc[0]
    )


    mask = (
        classification_delta[
            "Model"
        ]
        == model_name
    )


    classification_delta.loc[
        mask,
        "Accuracy_Drop"
    ] = (

        clean_accuracy

        -

        classification_delta.loc[
            mask,
            "Accuracy_Mean"
        ]
    )


    classification_delta.loc[
        mask,
        "F1_Drop"
    ] = (

        clean_f1

        -

        classification_delta.loc[
            mask,
            "F1_Mean"
        ]
    )


# ------------------------------------------------------------
# Step 18 — Calculate Regression Degradation
# ------------------------------------------------------------

regression_delta = (
    regression_summary.copy()
)


regression_delta[
    "Localization_Error_Increase"
] = 0.0


for model_name in (
    regression_delta[
        "Model"
    ].unique()
):


    clean_row = (
        regression_delta[
            (
                regression_delta[
                    "Model"
                ]
                == model_name
            )
            &
            (
                regression_delta[
                    "RSSI_Noise_Sigma_dB"
                ]
                == 0
            )
        ]
    )


    clean_error = (
        clean_row[
            "Mean_Localization_Error_Mean"
        ].iloc[0]
    )


    mask = (
        regression_delta[
            "Model"
        ]
        == model_name
    )


    regression_delta.loc[
        mask,
        "Localization_Error_Increase"
    ] = (

        regression_delta.loc[
            mask,
            "Mean_Localization_Error_Mean"
        ]

        -

        clean_error
    )


# ------------------------------------------------------------
# Step 19 — Save Degradation Tables
# ------------------------------------------------------------

CLASS_DELTA_FILE = (
    RESULTS_CSV_DIR
    / "05_rssi_noise_classification_degradation.csv"
)


REG_DELTA_FILE = (
    RESULTS_CSV_DIR
    / "05_rssi_noise_regression_degradation.csv"
)


classification_delta.to_csv(
    CLASS_DELTA_FILE,
    index=False
)


regression_delta.to_csv(
    REG_DELTA_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 20 — Plot Accuracy vs RSSI Noise
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 6)
)


for model_name in (
    classification_summary[
        "Model"
    ].unique()
):


    model_data = (
        classification_summary[
            classification_summary[
                "Model"
            ]
            == model_name
        ]
        .sort_values(
            "RSSI_Noise_Sigma_dB"
        )
    )


    plt.errorbar(

        model_data[
            "RSSI_Noise_Sigma_dB"
        ],

        model_data[
            "Accuracy_Mean"
        ],

        yerr=model_data[
            "Accuracy_SD"
        ],

        marker="o",

        capsize=3,

        label=model_name
    )


plt.xlabel(
    "RSSI Gaussian Noise σ (dB)"
)

plt.ylabel(
    "Building-Floor Classification Accuracy"
)

plt.title(
    "Classification Robustness Under RSSI Noise"
)

plt.ylim(
    0,
    1.05
)

plt.grid(
    alpha=0.2
)

plt.legend()

plt.tight_layout()


ACCURACY_FIGURE_FILE = (
    RESULTS_FIGURES_DIR
    / "05_accuracy_vs_rssi_noise.png"
)


plt.savefig(
    ACCURACY_FIGURE_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 21 — Plot Macro F1 vs RSSI Noise
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 6)
)


for model_name in (
    classification_summary[
        "Model"
    ].unique()
):


    model_data = (
        classification_summary[
            classification_summary[
                "Model"
            ]
            == model_name
        ]
        .sort_values(
            "RSSI_Noise_Sigma_dB"
        )
    )


    plt.errorbar(

        model_data[
            "RSSI_Noise_Sigma_dB"
        ],

        model_data[
            "F1_Mean"
        ],

        yerr=model_data[
            "F1_SD"
        ],

        marker="o",

        capsize=3,

        label=model_name
    )


plt.xlabel(
    "RSSI Gaussian Noise σ (dB)"
)

plt.ylabel(
    "Macro F1 Score"
)

plt.title(
    "Macro F1 Robustness Under RSSI Noise"
)

plt.ylim(
    0,
    1.05
)

plt.grid(
    alpha=0.2
)

plt.legend()

plt.tight_layout()


F1_FIGURE_FILE = (
    RESULTS_FIGURES_DIR
    / "05_f1_vs_rssi_noise.png"
)


plt.savefig(
    F1_FIGURE_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 22 — Plot Localization Error vs RSSI Noise
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 6)
)


for model_name in (
    regression_summary[
        "Model"
    ].unique()
):


    model_data = (
        regression_summary[
            regression_summary[
                "Model"
            ]
            == model_name
        ]
        .sort_values(
            "RSSI_Noise_Sigma_dB"
        )
    )


    plt.errorbar(

        model_data[
            "RSSI_Noise_Sigma_dB"
        ],

        model_data[
            "Mean_Localization_Error_Mean"
        ],

        yerr=model_data[
            "Mean_Localization_Error_SD"
        ],

        marker="o",

        capsize=3,

        label=model_name
    )


plt.xlabel(
    "RSSI Gaussian Noise σ (dB)"
)

plt.ylabel(
    "Mean Localization Error "
    "(dataset coordinate units)"
)

plt.title(
    "Coordinate Localization Robustness "
    "Under RSSI Noise"
)

plt.grid(
    alpha=0.2
)

plt.legend()

plt.tight_layout()


LOCALIZATION_FIGURE_FILE = (
    RESULTS_FIGURES_DIR
    / "05_localization_error_vs_rssi_noise.png"
)


plt.savefig(
    LOCALIZATION_FIGURE_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 23 — Extract Severe-Noise Results (σ = 10 dB)
# ------------------------------------------------------------

class_sigma10 = (
    classification_delta[
        classification_delta[
            "RSSI_Noise_Sigma_dB"
        ]
        == 10
    ]
)


reg_sigma10 = (
    regression_delta[
        regression_delta[
            "RSSI_Noise_Sigma_dB"
        ]
        == 10
    ]
)


best_classifier_sigma10 = (
    class_sigma10.loc[
        class_sigma10[
            "Accuracy_Drop"
        ].idxmin()
    ]
)


best_regressor_sigma10 = (
    reg_sigma10.loc[
        reg_sigma10[
            "Mean_Localization_Error_Mean"
        ].idxmin()
    ]
)


print("\n" + "=" * 76)
print("RESULTS AT RSSI NOISE σ = 10 dB")
print("=" * 76)


print("\nClassification:")

display(
    class_sigma10[
        [
            "Model",
            "Accuracy_Mean",
            "Accuracy_SD",
            "F1_Mean",
            "F1_SD",
            "Accuracy_Drop",
            "F1_Drop"
        ]
    ].round(4)
)


print("\nCoordinate localization:")

display(
    reg_sigma10[
        [
            "Model",
            "Mean_Localization_Error_Mean",
            "Mean_Localization_Error_SD",
            "Median_Localization_Error_Mean",
            "P90_Localization_Error_Mean",
            "Localization_Error_Increase"
        ]
    ].round(4)
)


# ------------------------------------------------------------
# Step 24 — Final Summary
# ------------------------------------------------------------

print("\n" + "=" * 76)
print("NOTEBOOK 05 COMPLETED SUCCESSFULLY")
print("=" * 76)


print("\nExperiment design:")

print(
    "RSSI noise levels:",
    RSSI_NOISE_LEVELS
)

print(
    "Repeated seeds:",
    len(EXPERIMENT_SEEDS)
)

print(
    "Noise applied only to originally "
    "detected AP measurements."
)


print("\nMOST ROBUST CLASSIFIER AT σ = 10 dB")

print(
    "Model:",
    best_classifier_sigma10[
        "Model"
    ]
)

print(
    "Accuracy:",
    round(
        float(
            best_classifier_sigma10[
                "Accuracy_Mean"
            ]
        ),
        4
    )
)

print(
    "Accuracy SD:",
    round(
        float(
            best_classifier_sigma10[
                "Accuracy_SD"
            ]
        ),
        4
    )
)

print(
    "Accuracy drop from clean:",
    round(
        float(
            best_classifier_sigma10[
                "Accuracy_Drop"
            ]
        ),
        4
    )
)


print("\nLOWEST LOCALIZATION ERROR AT σ = 10 dB")

print(
    "Model:",
    best_regressor_sigma10[
        "Model"
    ]
)

print(
    "Mean localization error:",
    round(
        float(
            best_regressor_sigma10[
                "Mean_Localization_Error_Mean"
            ]
        ),
        4
    )
)

print(
    "Localization error SD:",
    round(
        float(
            best_regressor_sigma10[
                "Mean_Localization_Error_SD"
            ]
        ),
        4
    )
)

print(
    "Increase from clean:",
    round(
        float(
            best_regressor_sigma10[
                "Localization_Error_Increase"
            ]
        ),
        4
    )
)


print("\nSaved CSV files:")

print(
    "1.",
    CLASS_DETAILED_FILE
)

print(
    "2.",
    REG_DETAILED_FILE
)

print(
    "3.",
    CLASS_SUMMARY_FILE
)

print(
    "4.",
    REG_SUMMARY_FILE
)

print(
    "5.",
    CLASS_DELTA_FILE
)

print(
    "6.",
    REG_DELTA_FILE
)


print("\nSaved figures:")

print(
    "1.",
    ACCURACY_FIGURE_FILE
)

print(
    "2.",
    F1_FIGURE_FILE
)

print(
    "3.",
    LOCALIZATION_FIGURE_FILE
)


print("\nNotebook 05 is complete.")

print(
    "Next notebook: "
    "06_combined_degradation.ipynb"
)
```
