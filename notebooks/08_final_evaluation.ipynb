```python
# ============================================================
# DEN702 Research Paper
# Notebook 08 — Final Held-Out Evaluation
#
# Research Topic:
# Robust Wi-Fi Fingerprint-Based Indoor Localization
# Under Access-Point Loss and RSSI Noise
#
# PURPOSE:
# Perform the FINAL evaluation using the official UJIIndoorLoc
# validation dataset, which has remained untouched during
# model development and model selection.
#
# Model selection was completed using development data:
#
# Classification selected model:
#     MLP + AP-dropout augmentation
#
# Localization selected model:
#     MLP + combined AP-dropout + RSSI-noise augmentation
#
# These choices MUST NOT be changed based on final-test results.
#
# Final evaluation includes:
# 1. Clean test performance
# 2. 50% AP-loss condition
# 3. RSSI noise sigma = 10 dB
# 4. Combined 50% AP loss + sigma = 10 dB
# 5. Full corruption-grid evaluation
# 6. Mean ± SD over five corruption seeds
# 7. Final publication-ready result tables
#
# IMPORTANT:
# The final test set is used for reporting only.
# No retraining, tuning, threshold selection, or model selection
# may be performed using these results.
# ============================================================


# ------------------------------------------------------------
# Step 1 — Import Required Libraries
# ------------------------------------------------------------

from pathlib import Path

import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from IPython.display import display

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    mean_absolute_error,
    mean_squared_error
)


warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# Step 2 — Reproducibility Configuration
# ------------------------------------------------------------

RANDOM_STATE = 42

np.random.seed(
    RANDOM_STATE
)

random.seed(
    RANDOM_STATE
)


MISSING_RSSI = -110

OBSERVED_MIN_RSSI = -104
OBSERVED_MAX_RSSI = 0


EVALUATION_SEEDS = [
    42,
    43,
    44,
    45,
    46
]


AP_LOSS_LEVELS = [
    0.00,
    0.10,
    0.20,
    0.30,
    0.40,
    0.50
]


RSSI_NOISE_LEVELS = [
    0,
    2,
    5,
    10
]


print("=" * 82)
print("DEN702 — FINAL HELD-OUT TEST EVALUATION")
print("=" * 82)

print(
    "Evaluation seeds:",
    EVALUATION_SEEDS
)

print(
    "FINAL TEST DATA WILL BE USED "
    "FOR REPORTING ONLY."
)


# ------------------------------------------------------------
# Step 3 — Detect Project Root
# ------------------------------------------------------------

if Path(
    "../data/processed/uji_preprocessed.npz"
).exists():

    PROJECT_ROOT = (
        Path("..").resolve()
    )

elif Path(
    "data/processed/uji_preprocessed.npz"
).exists():

    PROJECT_ROOT = (
        Path(".").resolve()
    )

else:

    raise FileNotFoundError(
        "Could not locate "
        "uji_preprocessed.npz."
    )


PROCESSED_DATA_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)


BASELINE_MODEL_DIR = (
    PROJECT_ROOT
    / "models"
    / "03_clean_baselines"
)


ROBUST_MODEL_DIR = (
    PROJECT_ROOT
    / "models"
    / "07_robust_training"
)


RESULTS_CSV_DIR = (
    PROJECT_ROOT
    / "results"
    / "csv"
)


RESULTS_FIGURES_DIR = (
    PROJECT_ROOT
    / "results"
    / "figures"
)


RESULTS_CSV_DIR.mkdir(
    parents=True,
    exist_ok=True
)


RESULTS_FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print(
    "\nProject root:"
)

print(
    PROJECT_ROOT
)


# ------------------------------------------------------------
# Step 4 — Load Final Held-Out Test Dataset
#
# X_test_raw:
#     Original RSSI fingerprints after replacing value 100
#     with -110 dBm.
#
# observed_mask_test:
#     Identifies APs that were originally detected.
#
# X_test:
#     Clean scaled test fingerprints.
# ------------------------------------------------------------

data = np.load(

    PROCESSED_DATA_DIR
    / "uji_preprocessed.npz",

    allow_pickle=True
)


X_test_clean = (
    data["X_test"]
)


X_test_raw = (
    data["X_test_raw"]
)


observed_mask_test = (
    data["observed_mask_test"]
)


y_class_test = (
    data["y_class_test"]
)


y_coord_test = (
    data["y_coord_test"]
)


print("\n" + "=" * 82)
print("STEP 4 — FINAL TEST DATASET")
print("=" * 82)


print(
    "Final test samples:",
    X_test_clean.shape
)


print(
    "Observed-mask shape:",
    observed_mask_test.shape
)


print(
    "Classification target:",
    y_class_test.shape
)


print(
    "Coordinate target:",
    y_coord_test.shape
)


print(
    "Building-floor classes:",
    len(
        np.unique(
            y_class_test
        )
    )
)


# ------------------------------------------------------------
# Step 5 — Inspect Final-Test AP Visibility
# ------------------------------------------------------------

test_detected_counts = (
    observed_mask_test.sum(
        axis=1
    )
)


print("\nFinal-test AP visibility:")


print(
    "Mean detected APs:",
    round(
        float(
            test_detected_counts.mean()
        ),
        2
    )
)


print(
    "Median detected APs:",
    round(
        float(
            np.median(
                test_detected_counts
            )
        ),
        2
    )
)


print(
    "Minimum detected APs:",
    int(
        test_detected_counts.min()
    )
)


print(
    "Maximum detected APs:",
    int(
        test_detected_counts.max()
    )
)


# ------------------------------------------------------------
# Step 6 — Load Fixed Preprocessing Objects
# ------------------------------------------------------------

rssi_scaler = joblib.load(

    PROCESSED_DATA_DIR
    / "uji_minmax_scaler.joblib"
)


coord_scaler = joblib.load(

    BASELINE_MODEL_DIR
    / "coordinate_scaler.joblib"
)


# ------------------------------------------------------------
# Step 7 — Load Clean Baseline Models
# ------------------------------------------------------------

clean_classifier = joblib.load(

    BASELINE_MODEL_DIR
    / "mlp_classifier.joblib"
)


clean_regressor = joblib.load(

    BASELINE_MODEL_DIR
    / "mlp_regressor.joblib"
)


# ------------------------------------------------------------
# Step 8 — Load Models Selected on Development Data
#
# Classification:
# AP-dropout-trained MLP
#
# Regression:
# combined AP + RSSI trained MLP
# ------------------------------------------------------------

robust_classifier = joblib.load(

    ROBUST_MODEL_DIR
    / "mlp_ap_dropout_classifier.joblib"
)


robust_regressor = joblib.load(

    ROBUST_MODEL_DIR
    / "mlp_combined_regressor.joblib"
)


print("\nSelected final models:")


print(
    "Classification:",
    "MLP + AP Dropout"
)


print(
    "Localization:",
    "MLP + Combined Augmentation"
)


# ============================================================
# PART A — CORRUPTION FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# Step 9 — AP-Loss Function
#
# AP loss affects originally detected APs only.
# ------------------------------------------------------------

def apply_ap_loss(
    X_raw,
    observed_mask,
    loss_fraction,
    seed
):

    if not 0 <= loss_fraction <= 1:

        raise ValueError(
            "loss_fraction must be "
            "between 0 and 1."
        )


    if loss_fraction == 0:

        return (
            X_raw.copy()
        )


    rng = (
        np.random.default_rng(
            seed
        )
    )


    X_corrupted = (
        X_raw
        .copy()
        .astype(float)
    )


    for row_index in range(
        X_corrupted.shape[0]
    ):


        detected_indices = (
            np.flatnonzero(
                observed_mask[
                    row_index
                ]
            )
        )


        n_detected = (
            len(
                detected_indices
            )
        )


        if n_detected == 0:

            continue


        n_to_drop = int(
            np.floor(
                n_detected
                * loss_fraction
            )
        )


        if (
            loss_fraction > 0
            and n_to_drop == 0
        ):

            n_to_drop = 1


        n_to_drop = min(
            n_to_drop,
            n_detected
        )


        drop_indices = (
            rng.choice(

                detected_indices,

                size=n_to_drop,

                replace=False
            )
        )


        X_corrupted[
            row_index,
            drop_indices
        ] = MISSING_RSSI


    return X_corrupted


# ------------------------------------------------------------
# Step 10 — RSSI Noise Function
#
# Noise is applied only to APs that remain visible after
# AP dropout.
# ------------------------------------------------------------

def apply_noise_after_loss(
    X_after_loss,
    observed_mask,
    sigma,
    seed
):

    if sigma < 0:

        raise ValueError(
            "sigma must be non-negative."
        )


    if sigma == 0:

        return (
            X_after_loss.copy()
        )


    rng = (
        np.random.default_rng(
            seed
        )
    )


    X_noisy = (
        X_after_loss
        .copy()
        .astype(float)
    )


    surviving_mask = (

        observed_mask

        &

        (
            X_after_loss
            != MISSING_RSSI
        )
    )


    noise = rng.normal(

        loc=0.0,

        scale=sigma,

        size=X_noisy.shape
    )


    X_noisy[
        surviving_mask
    ] += noise[
        surviving_mask
    ]


    X_noisy[
        surviving_mask
    ] = np.clip(

        X_noisy[
            surviving_mask
        ],

        OBSERVED_MIN_RSSI,

        OBSERVED_MAX_RSSI
    )


    X_noisy[
        ~surviving_mask
    ] = MISSING_RSSI


    return X_noisy


# ------------------------------------------------------------
# Step 11 — Create Corrupted Test Dataset
# ------------------------------------------------------------

def create_corrupted_test(
    ap_loss,
    sigma,
    seed
):

    X_after_loss = (
        apply_ap_loss(

            X_test_raw,

            observed_mask_test,

            loss_fraction=ap_loss,

            seed=seed
        )
    )


    X_corrupted = (
        apply_noise_after_loss(

            X_after_loss,

            observed_mask_test,

            sigma=sigma,

            seed=seed + 10000
        )
    )


    X_scaled = (
        rssi_scaler.transform(
            X_corrupted
        )
    )


    return X_scaled


# ============================================================
# PART B — EVALUATION FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# Step 12 — Classification Evaluation
# ------------------------------------------------------------

def evaluate_classifier(
    model,
    X_eval
):

    predictions = (
        model.predict(
            X_eval
        )
    )


    return {

        "Accuracy":

            accuracy_score(
                y_class_test,
                predictions
            ),

        "Precision_Macro":

            precision_score(
                y_class_test,
                predictions,
                average="macro",
                zero_division=0
            ),

        "Recall_Macro":

            recall_score(
                y_class_test,
                predictions,
                average="macro",
                zero_division=0
            ),

        "F1_Macro":

            f1_score(
                y_class_test,
                predictions,
                average="macro",
                zero_division=0
            )
    }, predictions


# ------------------------------------------------------------
# Step 13 — Localization Evaluation
# ------------------------------------------------------------

def evaluate_regressor(
    model,
    X_eval
):

    scaled_predictions = (
        model.predict(
            X_eval
        )
    )


    predictions = (
        coord_scaler.inverse_transform(
            scaled_predictions
        )
    )


    coordinate_mae = (
        mean_absolute_error(
            y_coord_test,
            predictions
        )
    )


    coordinate_rmse = np.sqrt(

        mean_squared_error(
            y_coord_test,
            predictions
        )
    )


    euclidean_errors = np.sqrt(

        np.sum(

            (
                predictions
                - y_coord_test
            ) ** 2,

            axis=1
        )
    )


    return {

        "Coordinate_MAE":

            coordinate_mae,

        "Coordinate_RMSE":

            coordinate_rmse,

        "Mean_Localization_Error":

            np.mean(
                euclidean_errors
            ),

        "Median_Localization_Error":

            np.median(
                euclidean_errors
            ),

        "P90_Localization_Error":

            np.percentile(
                euclidean_errors,
                90
            )

    }, predictions, euclidean_errors


# ============================================================
# PART C — CLEAN FINAL-TEST PERFORMANCE
# ============================================================


# ------------------------------------------------------------
# Step 14 — Evaluate Clean Classification Performance
#
# Clean test data contains no simulated corruption.
# ------------------------------------------------------------

(
    clean_class_metrics,
    clean_class_predictions
) = evaluate_classifier(

    clean_classifier,

    X_test_clean
)


(
    robust_class_clean_metrics,
    robust_class_clean_predictions
) = evaluate_classifier(

    robust_classifier,

    X_test_clean
)


print("\n" + "=" * 82)
print("STEP 14 — CLEAN FINAL-TEST CLASSIFICATION")
print("=" * 82)


clean_class_table = pd.DataFrame(
    [
        {
            "Training_Regime":
                "Clean",

            **clean_class_metrics
        },

        {
            "Training_Regime":
                "AP Dropout",

            **robust_class_clean_metrics
        }
    ]
)


display(
    clean_class_table.round(4)
)


# ------------------------------------------------------------
# Step 15 — Evaluate Clean Localization Performance
# ------------------------------------------------------------

(
    clean_reg_metrics,
    clean_reg_predictions,
    clean_reg_errors
) = evaluate_regressor(

    clean_regressor,

    X_test_clean
)


(
    robust_reg_clean_metrics,
    robust_reg_clean_predictions,
    robust_reg_clean_errors
) = evaluate_regressor(

    robust_regressor,

    X_test_clean
)


print("\n" + "=" * 82)
print("STEP 15 — CLEAN FINAL-TEST LOCALIZATION")
print("=" * 82)


clean_reg_table = pd.DataFrame(
    [
        {
            "Training_Regime":
                "Clean",

            **clean_reg_metrics
        },

        {
            "Training_Regime":
                "Combined",

            **robust_reg_clean_metrics
        }
    ]
)


display(
    clean_reg_table.round(4)
)


# ============================================================
# PART D — KEY FINAL-TEST CONDITIONS
# ============================================================


# ------------------------------------------------------------
# Step 16 — Define Pre-Specified Key Conditions
# ------------------------------------------------------------

KEY_CONDITIONS = [

    {
        "Condition":
            "Clean",

        "AP_Loss":
            0.00,

        "Sigma":
            0
    },

    {
        "Condition":
            "50% AP Loss",

        "AP_Loss":
            0.50,

        "Sigma":
            0
    },

    {
        "Condition":
            "RSSI Noise σ=10",

        "AP_Loss":
            0.00,

        "Sigma":
            10
    },

    {
        "Condition":
            "50% AP Loss + σ=10",

        "AP_Loss":
            0.50,

        "Sigma":
            10
    }
]


# ------------------------------------------------------------
# Step 17 — Run Key Classification Evaluation
# ------------------------------------------------------------

classification_records = []


for condition in KEY_CONDITIONS:


    condition_name = (
        condition[
            "Condition"
        ]
    )


    ap_loss = (
        condition[
            "AP_Loss"
        ]
    )


    sigma = (
        condition[
            "Sigma"
        ]
    )


    print(
        "\nClassification condition:",
        condition_name
    )


    for seed in EVALUATION_SEEDS:


        X_eval = (
            create_corrupted_test(

                ap_loss,

                sigma,

                seed
            )
        )


        for (
            training_regime,
            model
        ) in {

            "Clean":
                clean_classifier,

            "AP Dropout":
                robust_classifier

        }.items():


            metrics, _ = (
                evaluate_classifier(
                    model,
                    X_eval
                )
            )


            classification_records.append(
                {

                    "Training_Regime":
                        training_regime,

                    "Condition":
                        condition_name,

                    "AP_Loss_Percent":
                        int(
                            ap_loss
                            * 100
                        ),

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    **metrics
                }
            )


# ------------------------------------------------------------
# Step 18 — Run Key Localization Evaluation
# ------------------------------------------------------------

regression_records = []


for condition in KEY_CONDITIONS:


    condition_name = (
        condition[
            "Condition"
        ]
    )


    ap_loss = (
        condition[
            "AP_Loss"
        ]
    )


    sigma = (
        condition[
            "Sigma"
        ]
    )


    print(
        "\nLocalization condition:",
        condition_name
    )


    for seed in EVALUATION_SEEDS:


        X_eval = (
            create_corrupted_test(

                ap_loss,

                sigma,

                seed
            )
        )


        for (
            training_regime,
            model
        ) in {

            "Clean":
                clean_regressor,

            "Combined":
                robust_regressor

        }.items():


            metrics, _, _ = (
                evaluate_regressor(
                    model,
                    X_eval
                )
            )


            regression_records.append(
                {

                    "Training_Regime":
                        training_regime,

                    "Condition":
                        condition_name,

                    "AP_Loss_Percent":
                        int(
                            ap_loss
                            * 100
                        ),

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    **metrics
                }
            )


# ------------------------------------------------------------
# Step 19 — Summarize Key Classification Results
# ------------------------------------------------------------

classification_detailed = pd.DataFrame(
    classification_records
)


classification_summary = (

    classification_detailed

    .groupby(
        [
            "Training_Regime",
            "Condition"
        ],
        as_index=False,
        sort=False
    )

    .agg(

        Accuracy_Mean=(
            "Accuracy",
            "mean"
        ),

        Accuracy_SD=(
            "Accuracy",
            "std"
        ),

        Precision_Mean=(
            "Precision_Macro",
            "mean"
        ),

        Recall_Mean=(
            "Recall_Macro",
            "mean"
        ),

        F1_Mean=(
            "F1_Macro",
            "mean"
        ),

        F1_SD=(
            "F1_Macro",
            "std"
        )
    )

    .fillna(0)
)


print("\n" + "=" * 82)
print("STEP 19 — FINAL CLASSIFICATION RESULTS")
print("=" * 82)


display(
    classification_summary.round(4)
)


# ------------------------------------------------------------
# Step 20 — Summarize Key Localization Results
# ------------------------------------------------------------

regression_detailed = pd.DataFrame(
    regression_records
)


regression_summary = (

    regression_detailed

    .groupby(
        [
            "Training_Regime",
            "Condition"
        ],
        as_index=False,
        sort=False
    )

    .agg(

        Coordinate_MAE_Mean=(
            "Coordinate_MAE",
            "mean"
        ),

        Coordinate_RMSE_Mean=(
            "Coordinate_RMSE",
            "mean"
        ),

        Mean_Error_Mean=(
            "Mean_Localization_Error",
            "mean"
        ),

        Mean_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),

        Median_Error_Mean=(
            "Median_Localization_Error",
            "mean"
        ),

        P90_Error_Mean=(
            "P90_Localization_Error",
            "mean"
        )
    )

    .fillna(0)
)


print("\n" + "=" * 82)
print("STEP 20 — FINAL LOCALIZATION RESULTS")
print("=" * 82)


display(
    regression_summary.round(4)
)


# ------------------------------------------------------------
# Step 21 — Save Key Final-Test Results
# ------------------------------------------------------------

CLASS_DETAILED_FILE = (

    RESULTS_CSV_DIR
    / "08_final_classification_detailed.csv"
)


CLASS_SUMMARY_FILE = (

    RESULTS_CSV_DIR
    / "08_final_classification_summary.csv"
)


REG_DETAILED_FILE = (

    RESULTS_CSV_DIR
    / "08_final_localization_detailed.csv"
)


REG_SUMMARY_FILE = (

    RESULTS_CSV_DIR
    / "08_final_localization_summary.csv"
)


classification_detailed.to_csv(
    CLASS_DETAILED_FILE,
    index=False
)


classification_summary.to_csv(
    CLASS_SUMMARY_FILE,
    index=False
)


regression_detailed.to_csv(
    REG_DETAILED_FILE,
    index=False
)


regression_summary.to_csv(
    REG_SUMMARY_FILE,
    index=False
)


# ============================================================
# PART E — SEVERE-CONDITION FINAL IMPROVEMENT
# ============================================================


# ------------------------------------------------------------
# Step 22 — Severe Classification Improvement
# ------------------------------------------------------------

severe_condition = (
    "50% AP Loss + σ=10"
)


severe_class_results = (

    classification_summary[
        classification_summary[
            "Condition"
        ]
        == severe_condition
    ]
    .copy()
)


clean_severe_accuracy = float(

    severe_class_results[
        severe_class_results[
            "Training_Regime"
        ]
        == "Clean"
    ][
        "Accuracy_Mean"
    ].iloc[0]
)


robust_severe_accuracy = float(

    severe_class_results[
        severe_class_results[
            "Training_Regime"
        ]
        == "AP Dropout"
    ][
        "Accuracy_Mean"
    ].iloc[0]
)


final_accuracy_improvement = (

    robust_severe_accuracy
    -
    clean_severe_accuracy
)


# ------------------------------------------------------------
# Step 23 — Severe Localization Improvement
# ------------------------------------------------------------

severe_reg_results = (

    regression_summary[
        regression_summary[
            "Condition"
        ]
        == severe_condition
    ]
    .copy()
)


clean_severe_error = float(

    severe_reg_results[
        severe_reg_results[
            "Training_Regime"
        ]
        == "Clean"
    ][
        "Mean_Error_Mean"
    ].iloc[0]
)


robust_severe_error = float(

    severe_reg_results[
        severe_reg_results[
            "Training_Regime"
        ]
        == "Combined"
    ][
        "Mean_Error_Mean"
    ].iloc[0]
)


final_error_reduction = (

    clean_severe_error
    -
    robust_severe_error
)


final_error_reduction_percent = (

    final_error_reduction
    /
    clean_severe_error
    *
    100
)


# ============================================================
# PART F — FULL FINAL ROBUSTNESS GRID
# ============================================================


# ------------------------------------------------------------
# Step 24 — Evaluate Selected Classifier Across Full Grid
# ------------------------------------------------------------

full_class_records = []


total_conditions = (

    len(
        AP_LOSS_LEVELS
    )

    *

    len(
        RSSI_NOISE_LEVELS
    )
)


condition_number = 0


print("\n" + "=" * 82)
print("STEP 24 — FINAL CLASSIFICATION ROBUSTNESS GRID")
print("=" * 82)


for ap_loss in AP_LOSS_LEVELS:


    for sigma in RSSI_NOISE_LEVELS:


        condition_number += 1


        print(

            f"Condition "
            f"{condition_number}/"
            f"{total_conditions}: "

            f"AP={int(ap_loss * 100)}%, "
            f"σ={sigma}"
        )


        for seed in EVALUATION_SEEDS:


            X_eval = (
                create_corrupted_test(

                    ap_loss,

                    sigma,

                    seed
                )
            )


            for (
                training_regime,
                model
            ) in {

                "Clean":
                    clean_classifier,

                "AP Dropout":
                    robust_classifier

            }.items():


                metrics, _ = (
                    evaluate_classifier(
                        model,
                        X_eval
                    )
                )


                full_class_records.append(
                    {

                        "Training_Regime":
                            training_regime,

                        "AP_Loss_Percent":
                            int(
                                ap_loss
                                * 100
                            ),

                        "RSSI_Noise_Sigma_dB":
                            sigma,

                        "Seed":
                            seed,

                        **metrics
                    }
                )


# ------------------------------------------------------------
# Step 25 — Evaluate Selected Regressor Across Full Grid
# ------------------------------------------------------------

full_reg_records = []


condition_number = 0


print("\n" + "=" * 82)
print("STEP 25 — FINAL LOCALIZATION ROBUSTNESS GRID")
print("=" * 82)


for ap_loss in AP_LOSS_LEVELS:


    for sigma in RSSI_NOISE_LEVELS:


        condition_number += 1


        print(

            f"Condition "
            f"{condition_number}/"
            f"{total_conditions}: "

            f"AP={int(ap_loss * 100)}%, "
            f"σ={sigma}"
        )


        for seed in EVALUATION_SEEDS:


            X_eval = (
                create_corrupted_test(

                    ap_loss,

                    sigma,

                    seed
                )
            )


            for (
                training_regime,
                model
            ) in {

                "Clean":
                    clean_regressor,

                "Combined":
                    robust_regressor

            }.items():


                metrics, _, _ = (
                    evaluate_regressor(
                        model,
                        X_eval
                    )
                )


                full_reg_records.append(
                    {

                        "Training_Regime":
                            training_regime,

                        "AP_Loss_Percent":
                            int(
                                ap_loss
                                * 100
                            ),

                        "RSSI_Noise_Sigma_dB":
                            sigma,

                        "Seed":
                            seed,

                        **metrics
                    }
                )


# ------------------------------------------------------------
# Step 26 — Summarize Full Final Grid
# ------------------------------------------------------------

full_class_df = pd.DataFrame(
    full_class_records
)


full_reg_df = pd.DataFrame(
    full_reg_records
)


full_class_summary = (

    full_class_df

    .groupby(
        [
            "Training_Regime",
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ],
        as_index=False
    )

    .agg(

        Accuracy_Mean=(
            "Accuracy",
            "mean"
        ),

        Accuracy_SD=(
            "Accuracy",
            "std"
        ),

        F1_Mean=(
            "F1_Macro",
            "mean"
        ),

        F1_SD=(
            "F1_Macro",
            "std"
        )
    )

    .fillna(0)
)


full_reg_summary = (

    full_reg_df

    .groupby(
        [
            "Training_Regime",
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ],
        as_index=False
    )

    .agg(

        Mean_Error_Mean=(
            "Mean_Localization_Error",
            "mean"
        ),

        Mean_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),

        Median_Error_Mean=(
            "Median_Localization_Error",
            "mean"
        ),

        P90_Error_Mean=(
            "P90_Localization_Error",
            "mean"
        )
    )

    .fillna(0)
)


FULL_CLASS_FILE = (

    RESULTS_CSV_DIR
    / "08_final_fullgrid_classification.csv"
)


FULL_REG_FILE = (

    RESULTS_CSV_DIR
    / "08_final_fullgrid_localization.csv"
)


full_class_summary.to_csv(
    FULL_CLASS_FILE,
    index=False
)


full_reg_summary.to_csv(
    FULL_REG_FILE,
    index=False
)


# ============================================================
# PART G — FINAL IMPROVEMENT GRIDS
# ============================================================


# ------------------------------------------------------------
# Step 27 — Classification Improvement Grid
# ------------------------------------------------------------

clean_class_grid = (

    full_class_summary[
        full_class_summary[
            "Training_Regime"
        ]
        == "Clean"
    ][
        [
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB",
            "Accuracy_Mean"
        ]
    ]

    .rename(
        columns={
            "Accuracy_Mean":
                "Clean_Accuracy"
        }
    )
)


robust_class_grid = (

    full_class_summary[
        full_class_summary[
            "Training_Regime"
        ]
        == "AP Dropout"
    ][
        [
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB",
            "Accuracy_Mean"
        ]
    ]

    .rename(
        columns={
            "Accuracy_Mean":
                "Robust_Accuracy"
        }
    )
)


classification_improvement_grid = (

    clean_class_grid.merge(

        robust_class_grid,

        on=[
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ]
    )
)


classification_improvement_grid[
    "Accuracy_Improvement"
] = (

    classification_improvement_grid[
        "Robust_Accuracy"
    ]

    -

    classification_improvement_grid[
        "Clean_Accuracy"
    ]
)


# ------------------------------------------------------------
# Step 28 — Localization Improvement Grid
# ------------------------------------------------------------

clean_reg_grid = (

    full_reg_summary[
        full_reg_summary[
            "Training_Regime"
        ]
        == "Clean"
    ][
        [
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB",
            "Mean_Error_Mean"
        ]
    ]

    .rename(
        columns={
            "Mean_Error_Mean":
                "Clean_Error"
        }
    )
)


robust_reg_grid = (

    full_reg_summary[
        full_reg_summary[
            "Training_Regime"
        ]
        == "Combined"
    ][
        [
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB",
            "Mean_Error_Mean"
        ]
    ]

    .rename(
        columns={
            "Mean_Error_Mean":
                "Robust_Error"
        }
    )
)


regression_improvement_grid = (

    clean_reg_grid.merge(

        robust_reg_grid,

        on=[
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ]
    )
)


regression_improvement_grid[
    "Error_Reduction"
] = (

    regression_improvement_grid[
        "Clean_Error"
    ]

    -

    regression_improvement_grid[
        "Robust_Error"
    ]
)


CLASS_IMPROVEMENT_FILE = (

    RESULTS_CSV_DIR
    / "08_final_classification_improvement_grid.csv"
)


REG_IMPROVEMENT_FILE = (

    RESULTS_CSV_DIR
    / "08_final_localization_improvement_grid.csv"
)


classification_improvement_grid.to_csv(
    CLASS_IMPROVEMENT_FILE,
    index=False
)


regression_improvement_grid.to_csv(
    REG_IMPROVEMENT_FILE,
    index=False
)


# ============================================================
# PART H — FINAL FIGURES
# ============================================================


# ------------------------------------------------------------
# Step 29 — Classification Improvement Heatmap
# ------------------------------------------------------------

class_heatmap = (

    classification_improvement_grid

    .pivot(

        index="RSSI_Noise_Sigma_dB",

        columns="AP_Loss_Percent",

        values="Accuracy_Improvement"
    )

    .sort_index()

    .sort_index(
        axis=1
    )
)


fig, ax = plt.subplots(
    figsize=(9, 5)
)


image = ax.imshow(

    class_heatmap.values,

    aspect="auto",

    origin="lower"
)


ax.set_xticks(
    range(
        len(
            class_heatmap.columns
        )
    )
)


ax.set_xticklabels(
    class_heatmap.columns
)


ax.set_yticks(
    range(
        len(
            class_heatmap.index
        )
    )
)


ax.set_yticklabels(
    class_heatmap.index
)


ax.set_xlabel(
    "Detected APs Removed (%)"
)


ax.set_ylabel(
    "RSSI Noise σ (dB)"
)


ax.set_title(
    "Final-Test Accuracy Improvement "
    "from AP-Dropout Training"
)


for row in range(
    class_heatmap.shape[0]
):


    for col in range(
        class_heatmap.shape[1]
    ):


        ax.text(

            col,

            row,

            f"{class_heatmap.iloc[row, col]:+.3f}",

            ha="center",

            va="center"
        )


fig.colorbar(

    image,

    ax=ax,

    label="Accuracy Improvement"
)


plt.tight_layout()


CLASS_HEATMAP_FILE = (

    RESULTS_FIGURES_DIR
    / "08_final_accuracy_improvement_heatmap.png"
)


plt.savefig(

    CLASS_HEATMAP_FILE,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ------------------------------------------------------------
# Step 30 — Localization Error-Reduction Heatmap
# ------------------------------------------------------------

reg_heatmap = (

    regression_improvement_grid

    .pivot(

        index="RSSI_Noise_Sigma_dB",

        columns="AP_Loss_Percent",

        values="Error_Reduction"
    )

    .sort_index()

    .sort_index(
        axis=1
    )
)


fig, ax = plt.subplots(
    figsize=(9, 5)
)


image = ax.imshow(

    reg_heatmap.values,

    aspect="auto",

    origin="lower"
)


ax.set_xticks(
    range(
        len(
            reg_heatmap.columns
        )
    )
)


ax.set_xticklabels(
    reg_heatmap.columns
)


ax.set_yticks(
    range(
        len(
            reg_heatmap.index
        )
    )
)


ax.set_yticklabels(
    reg_heatmap.index
)


ax.set_xlabel(
    "Detected APs Removed (%)"
)


ax.set_ylabel(
    "RSSI Noise σ (dB)"
)


ax.set_title(
    "Final-Test Localization Error Reduction "
    "from Combined Training"
)


for row in range(
    reg_heatmap.shape[0]
):


    for col in range(
        reg_heatmap.shape[1]
    ):


        ax.text(

            col,

            row,

            f"{reg_heatmap.iloc[row, col]:+.2f}",

            ha="center",

            va="center"
        )


fig.colorbar(

    image,

    ax=ax,

    label="Mean Localization Error Reduction"
)


plt.tight_layout()


REG_HEATMAP_FILE = (

    RESULTS_FIGURES_DIR
    / "08_final_localization_improvement_heatmap.png"
)


plt.savefig(

    REG_HEATMAP_FILE,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ------------------------------------------------------------
# Step 31 — Severe Classification Comparison Figure
# ------------------------------------------------------------

severe_class_results.set_index(
    "Training_Regime"
)[
    "Accuracy_Mean"
].plot(

    kind="bar",

    figsize=(8, 5)
)


plt.ylabel(
    "Classification Accuracy"
)


plt.xlabel(
    "Training Regime"
)


plt.title(
    "Final Test — 50% AP Loss + σ=10 dB"
)


plt.ylim(
    0,
    1.05
)


plt.xticks(
    rotation=0
)


plt.grid(
    axis="y",
    alpha=0.2
)


plt.tight_layout()


SEVERE_CLASS_FIGURE = (

    RESULTS_FIGURES_DIR
    / "08_final_severe_classification.png"
)


plt.savefig(

    SEVERE_CLASS_FIGURE,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ------------------------------------------------------------
# Step 32 — Severe Localization Comparison Figure
# ------------------------------------------------------------

severe_reg_results.set_index(
    "Training_Regime"
)[
    "Mean_Error_Mean"
].plot(

    kind="bar",

    figsize=(8, 5)
)


plt.ylabel(
    "Mean Localization Error "
    "(dataset coordinate units)"
)


plt.xlabel(
    "Training Regime"
)


plt.title(
    "Final Test — 50% AP Loss + σ=10 dB"
)


plt.xticks(
    rotation=0
)


plt.grid(
    axis="y",
    alpha=0.2
)


plt.tight_layout()


SEVERE_REG_FIGURE = (

    RESULTS_FIGURES_DIR
    / "08_final_severe_localization.png"
)


plt.savefig(

    SEVERE_REG_FIGURE,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ============================================================
# PART I — FINAL CONFUSION MATRICES
# ============================================================


# ------------------------------------------------------------
# Step 33 — Robust Classifier Confusion Matrix Under Clean Test
# ------------------------------------------------------------

labels = sorted(
    np.unique(
        y_class_test
    )
)


clean_cm = confusion_matrix(

    y_class_test,

    robust_class_clean_predictions,

    labels=labels
)


fig, ax = plt.subplots(
    figsize=(10, 8)
)


disp = ConfusionMatrixDisplay(

    confusion_matrix=clean_cm,

    display_labels=labels
)


disp.plot(

    ax=ax,

    cmap="Blues",

    xticks_rotation=45,

    colorbar=False
)


plt.title(
    "Final-Test Confusion Matrix — "
    "AP-Dropout MLP (Clean Input)"
)


plt.tight_layout()


CONFUSION_FILE = (

    RESULTS_FIGURES_DIR
    / "08_final_robust_classifier_confusion_matrix.png"
)


plt.savefig(

    CONFUSION_FILE,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ============================================================
# PART J — FINAL PAPER SUMMARY TABLE
# ============================================================


# ------------------------------------------------------------
# Step 34 — Create Compact Final Result Table
# ------------------------------------------------------------

paper_summary = pd.DataFrame(
    [
        {
            "Task":
                "Classification",

            "Condition":
                "Clean",

            "Clean_Training":
                clean_class_metrics[
                    "Accuracy"
                ],

            "Robust_Training":
                robust_class_clean_metrics[
                    "Accuracy"
                ],

            "Difference":
                (
                    robust_class_clean_metrics[
                        "Accuracy"
                    ]
                    -
                    clean_class_metrics[
                        "Accuracy"
                    ]
                ),

            "Metric":
                "Accuracy"
        },

        {
            "Task":
                "Classification",

            "Condition":
                "50% AP Loss + σ=10",

            "Clean_Training":
                clean_severe_accuracy,

            "Robust_Training":
                robust_severe_accuracy,

            "Difference":
                final_accuracy_improvement,

            "Metric":
                "Accuracy"
        },

        {
            "Task":
                "Localization",

            "Condition":
                "Clean",

            "Clean_Training":
                clean_reg_metrics[
                    "Mean_Localization_Error"
                ],

            "Robust_Training":
                robust_reg_clean_metrics[
                    "Mean_Localization_Error"
                ],

            "Difference":
                (
                    clean_reg_metrics[
                        "Mean_Localization_Error"
                    ]
                    -
                    robust_reg_clean_metrics[
                        "Mean_Localization_Error"
                    ]
                ),

            "Metric":
                "Mean Localization Error"
        },

        {
            "Task":
                "Localization",

            "Condition":
                "50% AP Loss + σ=10",

            "Clean_Training":
                clean_severe_error,

            "Robust_Training":
                robust_severe_error,

            "Difference":
                final_error_reduction,

            "Metric":
                "Mean Localization Error"
        }
    ]
)


FINAL_PAPER_TABLE_FILE = (

    RESULTS_CSV_DIR
    / "08_final_paper_summary.csv"
)


paper_summary.to_csv(

    FINAL_PAPER_TABLE_FILE,

    index=False
)


print("\n" + "=" * 82)
print("STEP 34 — FINAL PAPER SUMMARY")
print("=" * 82)


display(
    paper_summary.round(4)
)


# ============================================================
# PART K — FINAL VERIFICATION
# ============================================================


# ------------------------------------------------------------
# Step 35 — Final Results
# ------------------------------------------------------------

print("\n" + "=" * 82)
print("NOTEBOOK 08 COMPLETED SUCCESSFULLY")
print("=" * 82)


print(
    "\nFINAL HELD-OUT TEST SAMPLES:",
    len(
        X_test_clean
    )
)


print(
    "\nSELECTED CLASSIFICATION MODEL:"
)

print(
    "MLP + AP-Dropout Augmentation"
)


print(
    "\nCLEAN FINAL-TEST CLASSIFICATION"
)

print(
    "Clean-trained MLP accuracy:",
    round(
        float(
            clean_class_metrics[
                "Accuracy"
            ]
        ),
        4
    )
)

print(
    "AP-dropout MLP accuracy:",
    round(
        float(
            robust_class_clean_metrics[
                "Accuracy"
            ]
        ),
        4
    )
)


print(
    "\nSEVERE FINAL-TEST CLASSIFICATION"
)

print(
    "Condition: "
    "50% AP loss + σ=10 dB"
)

print(
    "Clean-trained MLP accuracy:",
    round(
        clean_severe_accuracy,
        4
    )
)

print(
    "AP-dropout MLP accuracy:",
    round(
        robust_severe_accuracy,
        4
    )
)

print(
    "Final accuracy improvement:",
    round(
        final_accuracy_improvement,
        4
    )
)

print(
    "Improvement in percentage points:",
    round(
        final_accuracy_improvement
        * 100,
        2
    )
)


print(
    "\nSELECTED LOCALIZATION MODEL:"
)

print(
    "MLP + Combined "
    "AP-Dropout/RSSI-Noise Augmentation"
)


print(
    "\nCLEAN FINAL-TEST LOCALIZATION"
)

print(
    "Clean-trained MLP mean error:",
    round(
        float(
            clean_reg_metrics[
                "Mean_Localization_Error"
            ]
        ),
        4
    )
)

print(
    "Combined-trained MLP mean error:",
    round(
        float(
            robust_reg_clean_metrics[
                "Mean_Localization_Error"
            ]
        ),
        4
    )
)


print(
    "\nSEVERE FINAL-TEST LOCALIZATION"
)

print(
    "Condition: "
    "50% AP loss + σ=10 dB"
)

print(
    "Clean-trained MLP mean error:",
    round(
        clean_severe_error,
        4
    )
)

print(
    "Combined-trained MLP mean error:",
    round(
        robust_severe_error,
        4
    )
)

print(
    "Final localization-error reduction:",
    round(
        final_error_reduction,
        4
    )
)

print(
    "Relative error reduction:",
    round(
        final_error_reduction_percent,
        2
    ),
    "%"
)


print("\nSaved final result files:")

print(
    "1.",
    CLASS_SUMMARY_FILE
)

print(
    "2.",
    REG_SUMMARY_FILE
)

print(
    "3.",
    FULL_CLASS_FILE
)

print(
    "4.",
    FULL_REG_FILE
)

print(
    "5.",
    CLASS_IMPROVEMENT_FILE
)

print(
    "6.",
    REG_IMPROVEMENT_FILE
)

print(
    "7.",
    FINAL_PAPER_TABLE_FILE
)


print("\nSaved final figures:")

print(
    "1.",
    CLASS_HEATMAP_FILE
)

print(
    "2.",
    REG_HEATMAP_FILE
)

print(
    "3.",
    SEVERE_CLASS_FIGURE
)

print(
    "4.",
    SEVERE_REG_FIGURE
)

print(
    "5.",
    CONFUSION_FILE
)


print("\nFINAL TEST EVALUATION IS NOW FROZEN.")

print(
    "Do not use the final test results "
    "for additional model tuning."
)

print(
    "\nNext notebook: "
    "09_statistical_analysis.ipynb"
)
```
