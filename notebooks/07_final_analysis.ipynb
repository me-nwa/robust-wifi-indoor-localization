```python
# ============================================================
# DEN702 Research Paper
# Notebook 07 — Robust Training and Augmentation Ablation
#
# Research Topic:
# Robust Wi-Fi Fingerprint-Based Indoor Localization
# Under Access-Point Loss and RSSI Noise
#
# Main Research Hypothesis:
# Models trained with simulated AP loss and RSSI perturbation
# will experience less performance degradation under corrupted
# Wi-Fi conditions than models trained only on clean data.
#
# Purpose:
# 1. Use MLP as the robustness-training model
# 2. Train three augmented variants:
#       A. AP-dropout augmentation
#       B. RSSI-noise augmentation
#       C. Combined augmentation
# 3. Compare them with the clean MLP baseline
# 4. Perform an ablation study
# 5. Evaluate key corruption conditions
# 6. Evaluate clean vs combined-robust MLP across full grid
# 7. Save models, CSV results, and robustness figures
#
# IMPORTANT:
# - Final test set is NOT used.
# - Training corruption affects detected APs only.
# - RSSI noise affects surviving APs only.
# - Original scaler from Notebook 02 is reused.
# - Model architecture is kept the same as Notebook 03.
# ============================================================


# ------------------------------------------------------------
# Step 1 — Import Required Libraries
# ------------------------------------------------------------

from pathlib import Path

import gc
import time
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from IPython.display import display

from sklearn.neural_network import (
    MLPClassifier,
    MLPRegressor
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    mean_absolute_error,
    mean_squared_error
)


warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# Step 2 — Reproducibility Configuration
# ------------------------------------------------------------

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


MISSING_RSSI = -110

OBSERVED_MIN_RSSI = -104
OBSERVED_MAX_RSSI = 0


# Training augmentation severities

TRAIN_AP_LEVELS = [
    0.10,
    0.20,
    0.30,
    0.40,
    0.50
]


TRAIN_NOISE_LEVELS = [
    2,
    5,
    10
]


# Two augmented copies + original clean data

AUGMENTATION_SEEDS = [
    142,
    242
]


# Evaluation corruption seeds

EVALUATION_SEEDS = [
    42,
    43,
    44,
    45,
    46
]


print("=" * 80)
print("DEN702 — ROBUST TRAINING AND AUGMENTATION ABLATION")
print("=" * 80)

print(
    "Training AP-loss levels:",
    TRAIN_AP_LEVELS
)

print(
    "Training noise levels:",
    TRAIN_NOISE_LEVELS
)

print(
    "Augmented copies per regime:",
    len(AUGMENTATION_SEEDS)
)

print(
    "Evaluation seeds:",
    EVALUATION_SEEDS
)


# ------------------------------------------------------------
# Step 3 — Detect Project Root
# ------------------------------------------------------------

if Path("../data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path("..").resolve()

elif Path("data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path(".").resolve()

else:

    raise FileNotFoundError(
        "Could not locate uji_preprocessed.npz."
    )


PROCESSED_DATA_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)


BASELINE_MODEL_DIR = (
    PROJECT_ROOT
    / "models"
    / "03_clean_baselines"
)


ROBUST_MODEL_DIR = (
    PROJECT_ROOT
    / "models"
    / "07_robust_training"
)


RESULTS_CSV_DIR = (
    PROJECT_ROOT
    / "results"
    / "csv"
)


RESULTS_FIGURES_DIR = (
    PROJECT_ROOT
    / "results"
    / "figures"
)


ROBUST_MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_CSV_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("\nProject root:")
print(PROJECT_ROOT)


# ------------------------------------------------------------
# Step 4 — Load Training and Development Data
# ------------------------------------------------------------

data = np.load(
    PROCESSED_DATA_DIR
    / "uji_preprocessed.npz",
    allow_pickle=True
)


# Clean scaled features

X_train_clean = (
    data["X_train"]
)

X_dev_clean = (
    data["X_dev"]
)


# Raw RSSI values

X_train_raw = (
    data["X_train_raw"]
)

X_dev_raw = (
    data["X_dev_raw"]
)


# Observed AP masks

observed_mask_train = (
    data["observed_mask_train"]
)

observed_mask_dev = (
    data["observed_mask_dev"]
)


# Classification targets

y_class_train = (
    data["y_class_train"]
)

y_class_dev = (
    data["y_class_dev"]
)


# Coordinate targets

y_coord_train = (
    data["y_coord_train"]
)

y_coord_dev = (
    data["y_coord_dev"]
)


print("\n" + "=" * 80)
print("STEP 4 — DATA LOADED")
print("=" * 80)

print(
    "Training:",
    X_train_clean.shape
)

print(
    "Development:",
    X_dev_clean.shape
)

print(
    "Observed AP mask:",
    observed_mask_train.shape
)


# ------------------------------------------------------------
# Step 5 — Load Existing Preprocessing Objects
# ------------------------------------------------------------

rssi_scaler = joblib.load(
    PROCESSED_DATA_DIR
    / "uji_minmax_scaler.joblib"
)


coord_scaler = joblib.load(
    BASELINE_MODEL_DIR
    / "coordinate_scaler.joblib"
)


y_coord_train_scaled = (
    coord_scaler.transform(
        y_coord_train
    )
)


# ------------------------------------------------------------
# Step 6 — Load Clean MLP Baselines
#
# These models were trained in Notebook 03.
# ------------------------------------------------------------

clean_mlp_classifier = joblib.load(
    BASELINE_MODEL_DIR
    / "mlp_classifier.joblib"
)


clean_mlp_regressor = joblib.load(
    BASELINE_MODEL_DIR
    / "mlp_regressor.joblib"
)


print(
    "\nClean MLP baseline models loaded."
)


# ============================================================
# PART A — TRAINING AUGMENTATION FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# Step 7 — Generate One Augmented Training Copy
#
# Modes:
#
# "ap"
#     Random AP-loss severity is assigned per fingerprint.
#
# "noise"
#     Random RSSI-noise severity is assigned per fingerprint.
#
# "combined"
#     Both AP loss and RSSI noise are assigned per fingerprint.
#
# Only observed APs are modified.
# ------------------------------------------------------------

def create_augmented_training_copy(
    X_raw,
    observed_mask,
    mode,
    seed
):

    if mode not in [
        "ap",
        "noise",
        "combined"
    ]:

        raise ValueError(
            "Mode must be 'ap', 'noise', or 'combined'."
        )


    rng = np.random.default_rng(
        seed
    )


    X_augmented = (
        X_raw
        .copy()
        .astype(float)
    )


    n_samples = (
        X_augmented.shape[0]
    )


    for row_index in range(
        n_samples
    ):


        detected_indices = np.flatnonzero(
            observed_mask[
                row_index
            ]
        )


        n_detected = len(
            detected_indices
        )


        if n_detected == 0:

            continue


        # ----------------------------------------------------
        # AP-loss augmentation
        # ----------------------------------------------------

        if mode in [
            "ap",
            "combined"
        ]:


            loss_fraction = float(
                rng.choice(
                    TRAIN_AP_LEVELS
                )
            )


            n_to_drop = int(
                np.floor(
                    n_detected
                    * loss_fraction
                )
            )


            if n_to_drop == 0:

                n_to_drop = 1


            n_to_drop = min(
                n_to_drop,
                n_detected
            )


            drop_indices = rng.choice(

                detected_indices,

                size=n_to_drop,

                replace=False
            )


            X_augmented[
                row_index,
                drop_indices
            ] = MISSING_RSSI


        # ----------------------------------------------------
        # RSSI-noise augmentation
        # ----------------------------------------------------

        if mode in [
            "noise",
            "combined"
        ]:


            sigma = float(
                rng.choice(
                    TRAIN_NOISE_LEVELS
                )
            )


            surviving_indices = (
                detected_indices[
                    X_augmented[
                        row_index,
                        detected_indices
                    ]
                    != MISSING_RSSI
                ]
            )


            if (
                len(
                    surviving_indices
                )
                > 0
            ):


                noise = rng.normal(

                    loc=0.0,

                    scale=sigma,

                    size=len(
                        surviving_indices
                    )
                )


                X_augmented[
                    row_index,
                    surviving_indices
                ] = (

                    X_augmented[
                        row_index,
                        surviving_indices
                    ]

                    +

                    noise
                )


                X_augmented[
                    row_index,
                    surviving_indices
                ] = np.clip(

                    X_augmented[
                        row_index,
                        surviving_indices
                    ],

                    OBSERVED_MIN_RSSI,

                    OBSERVED_MAX_RSSI
                )


    return X_augmented


# ------------------------------------------------------------
# Step 8 — Verify Training Augmentation
# ------------------------------------------------------------

example_ap = (
    create_augmented_training_copy(

        X_train_raw[:5],

        observed_mask_train[:5],

        mode="ap",

        seed=142
    )
)


example_noise = (
    create_augmented_training_copy(

        X_train_raw[:5],

        observed_mask_train[:5],

        mode="noise",

        seed=142
    )
)


example_combined = (
    create_augmented_training_copy(

        X_train_raw[:5],

        observed_mask_train[:5],

        mode="combined",

        seed=142
    )
)


print("\n" + "=" * 80)
print("STEP 8 — AUGMENTATION FUNCTION CHECK")
print("=" * 80)

print(
    "AP-only shape:",
    example_ap.shape
)

print(
    "Noise-only shape:",
    example_noise.shape
)

print(
    "Combined shape:",
    example_combined.shape
)


# ------------------------------------------------------------
# Step 9 — Build Augmented Training Dataset
#
# Final training dataset:
#
# clean original
# +
# augmented copy 1
# +
# augmented copy 2
#
# Total size should therefore be:
#
# 15,949 × 3 = 47,847 samples
# ------------------------------------------------------------

def build_augmented_dataset(
    mode
):

    print(
        f"\nBuilding '{mode}' "
        f"augmentation dataset..."
    )


    X_parts = [
        X_train_clean
    ]


    for seed in AUGMENTATION_SEEDS:


        print(
            "Creating augmented copy "
            f"with seed {seed}..."
        )


        X_aug_raw = (
            create_augmented_training_copy(

                X_train_raw,

                observed_mask_train,

                mode=mode,

                seed=seed
            )
        )


        X_aug_scaled = (
            rssi_scaler.transform(
                X_aug_raw
            )
        )


        X_parts.append(
            X_aug_scaled
        )


        del X_aug_raw
        del X_aug_scaled

        gc.collect()


    X_final = np.vstack(
        X_parts
    )


    repetitions = (
        1
        +
        len(
            AUGMENTATION_SEEDS
        )
    )


    y_class_final = np.tile(
        y_class_train,
        repetitions
    )


    y_coord_final = np.tile(
        y_coord_train_scaled,
        (
            repetitions,
            1
        )
    )


    print(
        "Augmented X shape:",
        X_final.shape
    )

    print(
        "Classification target:",
        y_class_final.shape
    )

    print(
        "Regression target:",
        y_coord_final.shape
    )


    return (
        X_final,
        y_class_final,
        y_coord_final
    )


# ============================================================
# PART B — TRAIN ROBUST MODELS
# ============================================================


# ------------------------------------------------------------
# Step 10 — Standard MLP Configuration
#
# Keep architecture consistent with Notebook 03 so that the
# primary methodological difference is augmentation.
# ------------------------------------------------------------

def create_classifier():

    return MLPClassifier(

        hidden_layer_sizes=(
            128,
            64
        ),

        activation="relu",

        solver="adam",

        max_iter=300,

        early_stopping=True,

        validation_fraction=0.10,

        n_iter_no_change=15,

        random_state=RANDOM_STATE
    )


def create_regressor():

    return MLPRegressor(

        hidden_layer_sizes=(
            128,
            64
        ),

        activation="relu",

        solver="adam",

        max_iter=300,

        early_stopping=True,

        validation_fraction=0.10,

        n_iter_no_change=15,

        random_state=RANDOM_STATE
    )


# ------------------------------------------------------------
# Step 11 — Train One Augmentation Regime
# ------------------------------------------------------------

def train_robust_regime(
    regime_name,
    mode
):

    print("\n" + "=" * 80)
    print(
        f"TRAINING REGIME — {regime_name}"
    )
    print("=" * 80)


    (
        X_augmented,
        y_class_augmented,
        y_coord_augmented
    ) = build_augmented_dataset(
        mode
    )


    # --------------------------------------------------------
    # Classification model
    # --------------------------------------------------------

    classifier = (
        create_classifier()
    )


    start = time.time()


    classifier.fit(

        X_augmented,

        y_class_augmented
    )


    classifier_time = (
        time.time()
        - start
    )


    print(
        "Classifier training time:",
        round(
            classifier_time,
            2
        ),
        "seconds"
    )


    print(
        "Classifier iterations:",
        classifier.n_iter_
    )


    # --------------------------------------------------------
    # Regression model
    # --------------------------------------------------------

    regressor = (
        create_regressor()
    )


    start = time.time()


    regressor.fit(

        X_augmented,

        y_coord_augmented
    )


    regressor_time = (
        time.time()
        - start
    )


    print(
        "Regressor training time:",
        round(
            regressor_time,
            2
        ),
        "seconds"
    )


    print(
        "Regressor iterations:",
        regressor.n_iter_
    )


    # --------------------------------------------------------
    # Release large augmented dataset
    # --------------------------------------------------------

    del X_augmented
    del y_class_augmented
    del y_coord_augmented

    gc.collect()


    return (
        classifier,
        regressor,
        classifier_time,
        regressor_time
    )


# ------------------------------------------------------------
# Step 12 — Train AP-Dropout Models
# ------------------------------------------------------------

(
    ap_classifier,
    ap_regressor,
    ap_classifier_time,
    ap_regressor_time
) = train_robust_regime(

    "AP-Dropout Augmentation",

    "ap"
)


# ------------------------------------------------------------
# Step 13 — Save AP-Dropout Models Immediately
# ------------------------------------------------------------

AP_CLASSIFIER_FILE = (
    ROBUST_MODEL_DIR
    / "mlp_ap_dropout_classifier.joblib"
)


AP_REGRESSOR_FILE = (
    ROBUST_MODEL_DIR
    / "mlp_ap_dropout_regressor.joblib"
)


joblib.dump(
    ap_classifier,
    AP_CLASSIFIER_FILE
)


joblib.dump(
    ap_regressor,
    AP_REGRESSOR_FILE
)


# ------------------------------------------------------------
# Step 14 — Train RSSI-Noise Models
# ------------------------------------------------------------

(
    noise_classifier,
    noise_regressor,
    noise_classifier_time,
    noise_regressor_time
) = train_robust_regime(

    "RSSI-Noise Augmentation",

    "noise"
)


# ------------------------------------------------------------
# Step 15 — Save RSSI-Noise Models
# ------------------------------------------------------------

NOISE_CLASSIFIER_FILE = (
    ROBUST_MODEL_DIR
    / "mlp_noise_classifier.joblib"
)


NOISE_REGRESSOR_FILE = (
    ROBUST_MODEL_DIR
    / "mlp_noise_regressor.joblib"
)


joblib.dump(
    noise_classifier,
    NOISE_CLASSIFIER_FILE
)


joblib.dump(
    noise_regressor,
    NOISE_REGRESSOR_FILE
)


# ------------------------------------------------------------
# Step 16 — Train Combined-Augmentation Models
# ------------------------------------------------------------

(
    combined_classifier,
    combined_regressor,
    combined_classifier_time,
    combined_regressor_time
) = train_robust_regime(

    "Combined AP + RSSI Augmentation",

    "combined"
)


# ------------------------------------------------------------
# Step 17 — Save Combined Models
# ------------------------------------------------------------

COMBINED_CLASSIFIER_FILE = (
    ROBUST_MODEL_DIR
    / "mlp_combined_classifier.joblib"
)


COMBINED_REGRESSOR_FILE = (
    ROBUST_MODEL_DIR
    / "mlp_combined_regressor.joblib"
)


joblib.dump(
    combined_classifier,
    COMBINED_CLASSIFIER_FILE
)


joblib.dump(
    combined_regressor,
    COMBINED_REGRESSOR_FILE
)


print(
    "\nAll robust models saved successfully."
)


# ------------------------------------------------------------
# Step 18 — Save Training-Time Summary
# ------------------------------------------------------------

training_summary = pd.DataFrame(
    [
        {
            "Training_Regime":
                "AP Dropout",

            "Classifier_Time_s":
                ap_classifier_time,

            "Regressor_Time_s":
                ap_regressor_time
        },

        {
            "Training_Regime":
                "RSSI Noise",

            "Classifier_Time_s":
                noise_classifier_time,

            "Regressor_Time_s":
                noise_regressor_time
        },

        {
            "Training_Regime":
                "Combined",

            "Classifier_Time_s":
                combined_classifier_time,

            "Regressor_Time_s":
                combined_regressor_time
        }
    ]
)


TRAINING_SUMMARY_FILE = (
    RESULTS_CSV_DIR
    / "07_robust_training_times.csv"
)


training_summary.to_csv(
    TRAINING_SUMMARY_FILE,
    index=False
)


display(
    training_summary.round(2)
)


# ============================================================
# PART C — EVALUATION CORRUPTION FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# Step 19 — Apply AP Loss
# ------------------------------------------------------------

def apply_ap_loss(
    X_raw,
    observed_mask,
    loss_fraction,
    seed
):

    if loss_fraction == 0:

        return X_raw.copy()


    rng = np.random.default_rng(
        seed
    )


    X_corrupted = (
        X_raw
        .copy()
        .astype(float)
    )


    for row_index in range(
        X_corrupted.shape[0]
    ):


        detected_indices = np.flatnonzero(
            observed_mask[
                row_index
            ]
        )


        n_detected = len(
            detected_indices
        )


        if n_detected == 0:

            continue


        n_to_drop = int(
            np.floor(
                n_detected
                * loss_fraction
            )
        )


        if (
            loss_fraction > 0
            and n_to_drop == 0
        ):

            n_to_drop = 1


        drop_indices = rng.choice(

            detected_indices,

            size=min(
                n_to_drop,
                n_detected
            ),

            replace=False
        )


        X_corrupted[
            row_index,
            drop_indices
        ] = MISSING_RSSI


    return X_corrupted


# ------------------------------------------------------------
# Step 20 — Apply RSSI Noise After AP Loss
# ------------------------------------------------------------

def apply_noise_after_loss(
    X_after_loss,
    observed_mask,
    sigma,
    seed
):

    if sigma == 0:

        return X_after_loss.copy()


    rng = np.random.default_rng(
        seed
    )


    X_noisy = (
        X_after_loss
        .copy()
        .astype(float)
    )


    surviving_mask = (
        observed_mask
        &
        (
            X_after_loss
            != MISSING_RSSI
        )
    )


    noise = rng.normal(

        loc=0.0,

        scale=sigma,

        size=X_noisy.shape
    )


    X_noisy[
        surviving_mask
    ] += noise[
        surviving_mask
    ]


    X_noisy[
        surviving_mask
    ] = np.clip(

        X_noisy[
            surviving_mask
        ],

        OBSERVED_MIN_RSSI,

        OBSERVED_MAX_RSSI
    )


    X_noisy[
        ~surviving_mask
    ] = MISSING_RSSI


    return X_noisy


# ------------------------------------------------------------
# Step 21 — Combined Evaluation Corruption
# ------------------------------------------------------------

def create_corrupted_dev(
    ap_loss,
    sigma,
    seed
):

    X_loss = apply_ap_loss(

        X_dev_raw,

        observed_mask_dev,

        loss_fraction=ap_loss,

        seed=seed
    )


    X_corrupted = (
        apply_noise_after_loss(

            X_loss,

            observed_mask_dev,

            sigma=sigma,

            seed=seed + 10000
        )
    )


    return (
        rssi_scaler.transform(
            X_corrupted
        )
    )


# ============================================================
# PART D — EVALUATION FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# Step 22 — Classification Evaluation
# ------------------------------------------------------------

def evaluate_classifier(
    model,
    X_eval
):

    predictions = model.predict(
        X_eval
    )


    return {

        "Accuracy":
            accuracy_score(
                y_class_dev,
                predictions
            ),

        "Precision_Macro":
            precision_score(
                y_class_dev,
                predictions,
                average="macro",
                zero_division=0
            ),

        "Recall_Macro":
            recall_score(
                y_class_dev,
                predictions,
                average="macro",
                zero_division=0
            ),

        "F1_Macro":
            f1_score(
                y_class_dev,
                predictions,
                average="macro",
                zero_division=0
            )
    }


# ------------------------------------------------------------
# Step 23 — Regression Evaluation
# ------------------------------------------------------------

def evaluate_regressor(
    model,
    X_eval
):

    scaled_predictions = (
        model.predict(
            X_eval
        )
    )


    predictions = (
        coord_scaler.inverse_transform(
            scaled_predictions
        )
    )


    coordinate_mae = (
        mean_absolute_error(
            y_coord_dev,
            predictions
        )
    )


    coordinate_rmse = np.sqrt(
        mean_squared_error(
            y_coord_dev,
            predictions
        )
    )


    euclidean_errors = np.sqrt(
        np.sum(
            (
                predictions
                - y_coord_dev
            ) ** 2,
            axis=1
        )
    )


    return {

        "Coordinate_MAE":
            coordinate_mae,

        "Coordinate_RMSE":
            coordinate_rmse,

        "Mean_Localization_Error":
            np.mean(
                euclidean_errors
            ),

        "Median_Localization_Error":
            np.median(
                euclidean_errors
            ),

        "P90_Localization_Error":
            np.percentile(
                euclidean_errors,
                90
            )
    }


# ============================================================
# PART E — ABLATION STUDY
# ============================================================


# ------------------------------------------------------------
# Step 24 — Define Training Variants
# ------------------------------------------------------------

classifier_variants = {

    "Clean":

        clean_mlp_classifier,

    "AP Dropout":

        ap_classifier,

    "RSSI Noise":

        noise_classifier,

    "Combined":

        combined_classifier
}


regressor_variants = {

    "Clean":

        clean_mlp_regressor,

    "AP Dropout":

        ap_regressor,

    "RSSI Noise":

        noise_regressor,

    "Combined":

        combined_regressor
}


# ------------------------------------------------------------
# Step 25 — Define Key Evaluation Conditions
#
# These conditions isolate:
#
# 1. clean performance
# 2. AP-loss robustness
# 3. RSSI-noise robustness
# 4. combined robustness
# ------------------------------------------------------------

KEY_CONDITIONS = [

    {
        "Condition":
            "Clean",

        "AP_Loss":
            0.00,

        "Sigma":
            0
    },

    {
        "Condition":
            "50% AP Loss",

        "AP_Loss":
            0.50,

        "Sigma":
            0
    },

    {
        "Condition":
            "RSSI Noise σ=10",

        "AP_Loss":
            0.00,

        "Sigma":
            10
    },

    {
        "Condition":
            "50% AP Loss + σ=10",

        "AP_Loss":
            0.50,

        "Sigma":
            10
    }
]


# ------------------------------------------------------------
# Step 26 — Run Ablation Study
# ------------------------------------------------------------

classification_records = []

regression_records = []


print("\n" + "=" * 80)
print("STEP 26 — RUNNING AUGMENTATION ABLATION")
print("=" * 80)


for condition in KEY_CONDITIONS:


    condition_name = (
        condition[
            "Condition"
        ]
    )


    ap_loss = (
        condition[
            "AP_Loss"
        ]
    )


    sigma = (
        condition[
            "Sigma"
        ]
    )


    print(
        "\nCondition:",
        condition_name
    )


    for seed in EVALUATION_SEEDS:


        X_eval = (
            create_corrupted_dev(

                ap_loss,

                sigma,

                seed
            )
        )


        # ----------------------------------------------------
        # Classification variants
        # ----------------------------------------------------

        for (
            variant_name,
            model
        ) in classifier_variants.items():


            metrics = (
                evaluate_classifier(
                    model,
                    X_eval
                )
            )


            classification_records.append(
                {

                    "Training_Regime":
                        variant_name,

                    "Condition":
                        condition_name,

                    "AP_Loss_Percent":
                        int(
                            ap_loss
                            * 100
                        ),

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    **metrics
                }
            )


        # ----------------------------------------------------
        # Regression variants
        # ----------------------------------------------------

        for (
            variant_name,
            model
        ) in regressor_variants.items():


            metrics = (
                evaluate_regressor(
                    model,
                    X_eval
                )
            )


            regression_records.append(
                {

                    "Training_Regime":
                        variant_name,

                    "Condition":
                        condition_name,

                    "AP_Loss_Percent":
                        int(
                            ap_loss
                            * 100
                        ),

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    **metrics
                }
            )


# ------------------------------------------------------------
# Step 27 — Create Detailed Ablation Tables
# ------------------------------------------------------------

classification_ablation_detailed = (
    pd.DataFrame(
        classification_records
    )
)


regression_ablation_detailed = (
    pd.DataFrame(
        regression_records
    )
)


CLASS_ABLATION_DETAILED_FILE = (
    RESULTS_CSV_DIR
    / "07_ablation_classification_detailed.csv"
)


REG_ABLATION_DETAILED_FILE = (
    RESULTS_CSV_DIR
    / "07_ablation_regression_detailed.csv"
)


classification_ablation_detailed.to_csv(
    CLASS_ABLATION_DETAILED_FILE,
    index=False
)


regression_ablation_detailed.to_csv(
    REG_ABLATION_DETAILED_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 28 — Summarize Classification Ablation
# ------------------------------------------------------------

classification_ablation = (

    classification_ablation_detailed

    .groupby(
        [
            "Training_Regime",
            "Condition"
        ],
        as_index=False
    )

    .agg(

        Accuracy_Mean=(
            "Accuracy",
            "mean"
        ),

        Accuracy_SD=(
            "Accuracy",
            "std"
        ),

        F1_Mean=(
            "F1_Macro",
            "mean"
        ),

        F1_SD=(
            "F1_Macro",
            "std"
        )
    )
)


classification_ablation = (
    classification_ablation.fillna(0)
)


# ------------------------------------------------------------
# Step 29 — Summarize Regression Ablation
# ------------------------------------------------------------

regression_ablation = (

    regression_ablation_detailed

    .groupby(
        [
            "Training_Regime",
            "Condition"
        ],
        as_index=False
    )

    .agg(

        Mean_Error_Mean=(
            "Mean_Localization_Error",
            "mean"
        ),

        Mean_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),

        Median_Error_Mean=(
            "Median_Localization_Error",
            "mean"
        ),

        P90_Error_Mean=(
            "P90_Localization_Error",
            "mean"
        )
    )
)


regression_ablation = (
    regression_ablation.fillna(0)
)


print("\n" + "=" * 80)
print("STEP 29 — CLASSIFICATION ABLATION")
print("=" * 80)


display(
    classification_ablation.round(4)
)


print("\nRegression ablation:")


display(
    regression_ablation.round(4)
)


# ------------------------------------------------------------
# Step 30 — Save Ablation Summary Tables
# ------------------------------------------------------------

CLASS_ABLATION_FILE = (
    RESULTS_CSV_DIR
    / "07_ablation_classification_summary.csv"
)


REG_ABLATION_FILE = (
    RESULTS_CSV_DIR
    / "07_ablation_regression_summary.csv"
)


classification_ablation.to_csv(
    CLASS_ABLATION_FILE,
    index=False
)


regression_ablation.to_csv(
    REG_ABLATION_FILE,
    index=False
)


# ============================================================
# PART F — DIRECT CLEAN VS COMBINED TRAINING COMPARISON
# ============================================================


# ------------------------------------------------------------
# Step 31 — Severe-Condition Comparison
# ------------------------------------------------------------

severe_condition_name = (
    "50% AP Loss + σ=10"
)


severe_classification = (
    classification_ablation[
        classification_ablation[
            "Condition"
        ]
        == severe_condition_name
    ]
    .copy()
)


severe_regression = (
    regression_ablation[
        regression_ablation[
            "Condition"
        ]
        == severe_condition_name
    ]
    .copy()
)


print("\n" + "=" * 80)
print("STEP 31 — SEVERE-CONDITION ABLATION")
print("=" * 80)


print("\nClassification:")

display(
    severe_classification.round(4)
)


print("\nLocalization:")

display(
    severe_regression.round(4)
)


# ------------------------------------------------------------
# Step 32 — Calculate Improvement Over Clean Training
# ------------------------------------------------------------

clean_severe_class = float(

    severe_classification[
        severe_classification[
            "Training_Regime"
        ]
        == "Clean"
    ][
        "Accuracy_Mean"
    ].iloc[0]
)


combined_severe_class = float(

    severe_classification[
        severe_classification[
            "Training_Regime"
        ]
        == "Combined"
    ][
        "Accuracy_Mean"
    ].iloc[0]
)


classification_improvement = (

    combined_severe_class

    -

    clean_severe_class
)


clean_severe_error = float(

    severe_regression[
        severe_regression[
            "Training_Regime"
        ]
        == "Clean"
    ][
        "Mean_Error_Mean"
    ].iloc[0]
)


combined_severe_error = float(

    severe_regression[
        severe_regression[
            "Training_Regime"
        ]
        == "Combined"
    ][
        "Mean_Error_Mean"
    ].iloc[0]
)


localization_error_reduction = (

    clean_severe_error

    -

    combined_severe_error
)


print(
    "\nCombined-training accuracy improvement:",
    round(
        classification_improvement,
        4
    )
)


print(
    "Combined-training localization "
    "error reduction:",
    round(
        localization_error_reduction,
        4
    )
)


# ============================================================
# PART G — FULL-GRID ROBUSTNESS COMPARISON
# ============================================================


# ------------------------------------------------------------
# Step 33 — Define Full Evaluation Grid
# ------------------------------------------------------------

FULL_AP_LEVELS = [
    0.00,
    0.10,
    0.20,
    0.30,
    0.40,
    0.50
]


FULL_NOISE_LEVELS = [
    0,
    2,
    5,
    10
]


full_grid_records_class = []

full_grid_records_reg = []


print("\n" + "=" * 80)
print("STEP 33 — CLEAN VS COMBINED FULL-GRID EVALUATION")
print("=" * 80)


total_conditions = (

    len(
        FULL_AP_LEVELS
    )

    *

    len(
        FULL_NOISE_LEVELS
    )
)


condition_number = 0


for ap_loss in FULL_AP_LEVELS:


    for sigma in FULL_NOISE_LEVELS:


        condition_number += 1


        print(

            f"\nCondition "
            f"{condition_number}/"
            f"{total_conditions}: "

            f"AP={int(ap_loss * 100)}%, "
            f"σ={sigma}"
        )


        for seed in EVALUATION_SEEDS:


            X_eval = (
                create_corrupted_dev(

                    ap_loss,

                    sigma,

                    seed
                )
            )


            # ------------------------------------------------
            # Clean-trained MLP classifier
            # ------------------------------------------------

            clean_metrics = (
                evaluate_classifier(

                    clean_mlp_classifier,

                    X_eval
                )
            )


            full_grid_records_class.append(
                {

                    "Training_Regime":
                        "Clean",

                    "AP_Loss_Percent":
                        int(
                            ap_loss
                            * 100
                        ),

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    **clean_metrics
                }
            )


            # ------------------------------------------------
            # Combined-trained MLP classifier
            # ------------------------------------------------

            combined_metrics = (
                evaluate_classifier(

                    combined_classifier,

                    X_eval
                )
            )


            full_grid_records_class.append(
                {

                    "Training_Regime":
                        "Combined",

                    "AP_Loss_Percent":
                        int(
                            ap_loss
                            * 100
                        ),

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    **combined_metrics
                }
            )


            # ------------------------------------------------
            # Clean-trained MLP regressor
            # ------------------------------------------------

            clean_reg_metrics = (
                evaluate_regressor(

                    clean_mlp_regressor,

                    X_eval
                )
            )


            full_grid_records_reg.append(
                {

                    "Training_Regime":
                        "Clean",

                    "AP_Loss_Percent":
                        int(
                            ap_loss
                            * 100
                        ),

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    **clean_reg_metrics
                }
            )


            # ------------------------------------------------
            # Combined-trained MLP regressor
            # ------------------------------------------------

            combined_reg_metrics = (
                evaluate_regressor(

                    combined_regressor,

                    X_eval
                )
            )


            full_grid_records_reg.append(
                {

                    "Training_Regime":
                        "Combined",

                    "AP_Loss_Percent":
                        int(
                            ap_loss
                            * 100
                        ),

                    "RSSI_Noise_Sigma_dB":
                        sigma,

                    "Seed":
                        seed,

                    **combined_reg_metrics
                }
            )


# ------------------------------------------------------------
# Step 34 — Full-Grid Summary
# ------------------------------------------------------------

full_class_df = pd.DataFrame(
    full_grid_records_class
)


full_reg_df = pd.DataFrame(
    full_grid_records_reg
)


full_class_summary = (

    full_class_df

    .groupby(
        [
            "Training_Regime",
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ],
        as_index=False
    )

    .agg(

        Accuracy_Mean=(
            "Accuracy",
            "mean"
        ),

        Accuracy_SD=(
            "Accuracy",
            "std"
        ),

        F1_Mean=(
            "F1_Macro",
            "mean"
        )
    )

    .fillna(0)
)


full_reg_summary = (

    full_reg_df

    .groupby(
        [
            "Training_Regime",
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ],
        as_index=False
    )

    .agg(

        Mean_Error_Mean=(
            "Mean_Localization_Error",
            "mean"
        ),

        Mean_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),

        Median_Error_Mean=(
            "Median_Localization_Error",
            "mean"
        )
    )

    .fillna(0)
)


FULL_CLASS_FILE = (
    RESULTS_CSV_DIR
    / "07_clean_vs_robust_fullgrid_classification.csv"
)


FULL_REG_FILE = (
    RESULTS_CSV_DIR
    / "07_clean_vs_robust_fullgrid_regression.csv"
)


full_class_summary.to_csv(
    FULL_CLASS_FILE,
    index=False
)


full_reg_summary.to_csv(
    FULL_REG_FILE,
    index=False
)


# ============================================================
# PART H — IMPROVEMENT HEATMAPS
# ============================================================


# ------------------------------------------------------------
# Step 35 — Calculate Classification Accuracy Improvement
#
# Improvement =
#
# robust accuracy - clean accuracy
#
# Positive values mean robust training helped.
# ------------------------------------------------------------

clean_class_grid = (

    full_class_summary[
        full_class_summary[
            "Training_Regime"
        ]
        == "Clean"
    ][
        [
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB",
            "Accuracy_Mean"
        ]
    ]

    .rename(
        columns={
            "Accuracy_Mean":
                "Clean_Accuracy"
        }
    )
)


robust_class_grid = (

    full_class_summary[
        full_class_summary[
            "Training_Regime"
        ]
        == "Combined"
    ][
        [
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB",
            "Accuracy_Mean"
        ]
    ]

    .rename(
        columns={
            "Accuracy_Mean":
                "Robust_Accuracy"
        }
    )
)


classification_improvement_grid = (
    clean_class_grid.merge(

        robust_class_grid,

        on=[
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ]
    )
)


classification_improvement_grid[
    "Accuracy_Improvement"
] = (

    classification_improvement_grid[
        "Robust_Accuracy"
    ]

    -

    classification_improvement_grid[
        "Clean_Accuracy"
    ]
)


# ------------------------------------------------------------
# Step 36 — Calculate Localization Error Reduction
#
# Error reduction =
#
# clean error - robust error
#
# Positive values mean robust training helped.
# ------------------------------------------------------------

clean_reg_grid = (

    full_reg_summary[
        full_reg_summary[
            "Training_Regime"
        ]
        == "Clean"
    ][
        [
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB",
            "Mean_Error_Mean"
        ]
    ]

    .rename(
        columns={
            "Mean_Error_Mean":
                "Clean_Error"
        }
    )
)


robust_reg_grid = (

    full_reg_summary[
        full_reg_summary[
            "Training_Regime"
        ]
        == "Combined"
    ][
        [
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB",
            "Mean_Error_Mean"
        ]
    ]

    .rename(
        columns={
            "Mean_Error_Mean":
                "Robust_Error"
        }
    )
)


regression_improvement_grid = (
    clean_reg_grid.merge(

        robust_reg_grid,

        on=[
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ]
    )
)


regression_improvement_grid[
    "Error_Reduction"
] = (

    regression_improvement_grid[
        "Clean_Error"
    ]

    -

    regression_improvement_grid[
        "Robust_Error"
    ]
)


# ------------------------------------------------------------
# Step 37 — Save Improvement Data
# ------------------------------------------------------------

CLASS_IMPROVEMENT_FILE = (
    RESULTS_CSV_DIR
    / "07_classification_improvement_grid.csv"
)


REG_IMPROVEMENT_FILE = (
    RESULTS_CSV_DIR
    / "07_localization_improvement_grid.csv"
)


classification_improvement_grid.to_csv(
    CLASS_IMPROVEMENT_FILE,
    index=False
)


regression_improvement_grid.to_csv(
    REG_IMPROVEMENT_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 38 — Classification Improvement Heatmap
# ------------------------------------------------------------

class_heatmap = (

    classification_improvement_grid

    .pivot(

        index="RSSI_Noise_Sigma_dB",

        columns="AP_Loss_Percent",

        values="Accuracy_Improvement"
    )

    .sort_index()

    .sort_index(
        axis=1
    )
)


fig, ax = plt.subplots(
    figsize=(9, 5)
)


image = ax.imshow(

    class_heatmap.values,

    aspect="auto",

    origin="lower"
)


ax.set_xticks(
    range(
        len(
            class_heatmap.columns
        )
    )
)


ax.set_xticklabels(
    class_heatmap.columns
)


ax.set_yticks(
    range(
        len(
            class_heatmap.index
        )
    )
)


ax.set_yticklabels(
    class_heatmap.index
)


ax.set_xlabel(
    "Detected APs Removed (%)"
)

ax.set_ylabel(
    "RSSI Noise σ (dB)"
)


ax.set_title(
    "Accuracy Improvement from Combined Robust Training"
)


for row in range(
    class_heatmap.shape[0]
):

    for col in range(
        class_heatmap.shape[1]
    ):


        ax.text(

            col,

            row,

            f"{class_heatmap.iloc[row, col]:+.3f}",

            ha="center",

            va="center"
        )


fig.colorbar(

    image,

    ax=ax,

    label="Accuracy Improvement"
)


plt.tight_layout()


CLASS_HEATMAP_FILE = (
    RESULTS_FIGURES_DIR
    / "07_robust_training_accuracy_improvement_heatmap.png"
)


plt.savefig(
    CLASS_HEATMAP_FILE,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ------------------------------------------------------------
# Step 39 — Localization Improvement Heatmap
# ------------------------------------------------------------

reg_heatmap = (

    regression_improvement_grid

    .pivot(

        index="RSSI_Noise_Sigma_dB",

        columns="AP_Loss_Percent",

        values="Error_Reduction"
    )

    .sort_index()

    .sort_index(
        axis=1
    )
)


fig, ax = plt.subplots(
    figsize=(9, 5)
)


image = ax.imshow(

    reg_heatmap.values,

    aspect="auto",

    origin="lower"
)


ax.set_xticks(
    range(
        len(
            reg_heatmap.columns
        )
    )
)


ax.set_xticklabels(
    reg_heatmap.columns
)


ax.set_yticks(
    range(
        len(
            reg_heatmap.index
        )
    )
)


ax.set_yticklabels(
    reg_heatmap.index
)


ax.set_xlabel(
    "Detected APs Removed (%)"
)

ax.set_ylabel(
    "RSSI Noise σ (dB)"
)


ax.set_title(
    "Localization Error Reduction from Combined Robust Training"
)


for row in range(
    reg_heatmap.shape[0]
):

    for col in range(
        reg_heatmap.shape[1]
    ):


        ax.text(

            col,

            row,

            f"{reg_heatmap.iloc[row, col]:+.2f}",

            ha="center",

            va="center"
        )


fig.colorbar(

    image,

    ax=ax,

    label="Mean Localization Error Reduction"
)


plt.tight_layout()


REG_HEATMAP_FILE = (
    RESULTS_FIGURES_DIR
    / "07_robust_training_error_reduction_heatmap.png"
)


plt.savefig(
    REG_HEATMAP_FILE,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# PART I — ABLATION BAR CHARTS
# ============================================================


# ------------------------------------------------------------
# Step 40 — Classification Severe-Condition Bar Chart
# ------------------------------------------------------------

class_plot = (

    severe_classification

    .set_index(
        "Training_Regime"
    )[
        "Accuracy_Mean"
    ]
)


class_plot.plot(
    kind="bar",
    figsize=(9, 5)
)


plt.ylabel(
    "Classification Accuracy"
)

plt.xlabel(
    "Training Regime"
)

plt.title(
    "Augmentation Ablation Under "
    "50% AP Loss + σ=10 dB"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y",
    alpha=0.2
)

plt.tight_layout()


CLASS_ABLATION_FIGURE = (
    RESULTS_FIGURES_DIR
    / "07_classification_ablation_severe_condition.png"
)


plt.savefig(
    CLASS_ABLATION_FIGURE,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ------------------------------------------------------------
# Step 41 — Regression Severe-Condition Bar Chart
# ------------------------------------------------------------

reg_plot = (

    severe_regression

    .set_index(
        "Training_Regime"
    )[
        "Mean_Error_Mean"
    ]
)


reg_plot.plot(
    kind="bar",
    figsize=(9, 5)
)


plt.ylabel(
    "Mean Localization Error "
    "(dataset coordinate units)"
)

plt.xlabel(
    "Training Regime"
)

plt.title(
    "Localization Ablation Under "
    "50% AP Loss + σ=10 dB"
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y",
    alpha=0.2
)

plt.tight_layout()


REG_ABLATION_FIGURE = (
    RESULTS_FIGURES_DIR
    / "07_localization_ablation_severe_condition.png"
)


plt.savefig(
    REG_ABLATION_FIGURE,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# PART J — FINAL SUMMARY
# ============================================================


# ------------------------------------------------------------
# Step 42 — Identify Strongest Severe Classification Regime
# ------------------------------------------------------------

best_severe_classifier_row = (

    severe_classification.loc[
        severe_classification[
            "Accuracy_Mean"
        ].idxmax()
    ]
)


# ------------------------------------------------------------
# Step 43 — Identify Lowest-Error Severe Regression Regime
# ------------------------------------------------------------

best_severe_regressor_row = (

    severe_regression.loc[
        severe_regression[
            "Mean_Error_Mean"
        ].idxmin()
    ]
)


# ------------------------------------------------------------
# Step 44 — Final Verification
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("NOTEBOOK 07 COMPLETED SUCCESSFULLY")
print("=" * 80)


print("\nAUGMENTATION STUDY DESIGN")

print(
    "Clean training samples:",
    len(
        X_train_clean
    )
)

print(
    "Augmented copies per regime:",
    len(
        AUGMENTATION_SEEDS
    )
)

print(
    "Approximate samples per "
    "augmented training regime:",
    len(
        X_train_clean
    )
    *
    (
        1
        +
        len(
            AUGMENTATION_SEEDS
        )
    )
)


print(
    "\nSEVERE CONDITION:"
)

print(
    "50% AP loss + "
    "RSSI noise σ=10 dB"
)


print(
    "\nBEST CLASSIFICATION TRAINING REGIME"
)

print(
    "Training regime:",
    best_severe_classifier_row[
        "Training_Regime"
    ]
)

print(
    "Accuracy:",
    round(
        float(
            best_severe_classifier_row[
                "Accuracy_Mean"
            ]
        ),
        4
    )
)

print(
    "Accuracy SD:",
    round(
        float(
            best_severe_classifier_row[
                "Accuracy_SD"
            ]
        ),
        4
    )
)

print(
    "Macro F1:",
    round(
        float(
            best_severe_classifier_row[
                "F1_Mean"
            ]
        ),
        4
    )
)


print(
    "\nBEST LOCALIZATION TRAINING REGIME"
)

print(
    "Training regime:",
    best_severe_regressor_row[
        "Training_Regime"
    ]
)

print(
    "Mean localization error:",
    round(
        float(
            best_severe_regressor_row[
                "Mean_Error_Mean"
            ]
        ),
        4
    )
)

print(
    "Localization error SD:",
    round(
        float(
            best_severe_regressor_row[
                "Mean_Error_SD"
            ]
        ),
        4
    )
)


print(
    "\nCOMBINED AUGMENTATION VS CLEAN BASELINE"
)

print(
    "Severe-condition clean accuracy:",
    round(
        clean_severe_class,
        4
    )
)

print(
    "Severe-condition combined "
    "training accuracy:",
    round(
        combined_severe_class,
        4
    )
)

print(
    "Accuracy improvement:",
    round(
        classification_improvement,
        4
    )
)


print(
    "\nClean mean localization error:",
    round(
        clean_severe_error,
        4
    )
)

print(
    "Combined-training mean "
    "localization error:",
    round(
        combined_severe_error,
        4
    )
)

print(
    "Localization error reduction:",
    round(
        localization_error_reduction,
        4
    )
)


print("\nSaved robust models:")

print(
    "1.",
    AP_CLASSIFIER_FILE
)

print(
    "2.",
    AP_REGRESSOR_FILE
)

print(
    "3.",
    NOISE_CLASSIFIER_FILE
)

print(
    "4.",
    NOISE_REGRESSOR_FILE
)

print(
    "5.",
    COMBINED_CLASSIFIER_FILE
)

print(
    "6.",
    COMBINED_REGRESSOR_FILE
)


print("\nSaved result files:")

print(
    "1.",
    CLASS_ABLATION_FILE
)

print(
    "2.",
    REG_ABLATION_FILE
)

print(
    "3.",
    FULL_CLASS_FILE
)

print(
    "4.",
    FULL_REG_FILE
)

print(
    "5.",
    CLASS_IMPROVEMENT_FILE
)

print(
    "6.",
    REG_IMPROVEMENT_FILE
)


print("\nSaved figures:")

print(
    "1.",
    CLASS_HEATMAP_FILE
)

print(
    "2.",
    REG_HEATMAP_FILE
)

print(
    "3.",
    CLASS_ABLATION_FIGURE
)

print(
    "4.",
    REG_ABLATION_FIGURE
)


print("\nNotebook 07 is complete.")

print(
    "Next notebook: "
    "08_final_evaluation.ipynb"
)
```
