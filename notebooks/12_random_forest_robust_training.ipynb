```python
# ============================================================
# NOTEBOOK 12
# Random Forest Robust-Training Control
#
# Reviewer concern addressed:
#   Corruption-aware augmentation was originally applied only
#   to MLP. This notebook tests whether the same augmentation
#   strategy also benefits Random Forest.
#
# Regimes:
#   1. Clean
#   2. Repeated Clean (equal-volume control)
#   3. AP Dropout
#   4. Combined AP Dropout + RSSI Noise
#
# IMPORTANT:
#   This is a reviewer-requested post-hoc experiment.
#   It does not replace the original frozen final models.
#   Round-two export adds Mean, Median, and P90 localization error.
# ============================================================


# ------------------------------------------------------------
# Step 1 — Imports
# ------------------------------------------------------------

from pathlib import Path
import json
import time
import warnings

import numpy as np
import pandas as pd
import joblib
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import accuracy_score, f1_score

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# Step 2 — Project paths
# ------------------------------------------------------------

PROJECT_ROOT = (
    Path.home()
    / "Documents"
    / "ADU"
    / "wifi-localization-research"
)

TRAIN_FILE = PROJECT_ROOT / "data" / "raw" / "trainingData.csv"
TEST_FILE = PROJECT_ROOT / "data" / "raw" / "validationData.csv"

MODEL_11_DIR = PROJECT_ROOT / "models" / "11_equal_volume_control"
MODEL_DIR = PROJECT_ROOT / "models" / "12_random_forest_robust_training"

RESULTS_CSV = PROJECT_ROOT / "results" / "csv"
RESULTS_FIG = PROJECT_ROOT / "results" / "figures"

MODEL_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_CSV.mkdir(parents=True, exist_ok=True)
RESULTS_FIG.mkdir(parents=True, exist_ok=True)

assert TRAIN_FILE.exists(), TRAIN_FILE
assert TEST_FILE.exists(), TEST_FILE

print("PROJECT ROOT:", PROJECT_ROOT)


# ------------------------------------------------------------
# Step 3 — Experiment constants
# ------------------------------------------------------------

RANDOM_STATE = 42

DEV_SIZE = 0.20
EARLY_STOP_SIZE = 0.10

MISSING_ORIGINAL = 100
MISSING_RSSI = -110.0

MIN_OBSERVED_RSSI = -104.0
MAX_OBSERVED_RSSI = 0.0

AP_LEVELS = [0.10, 0.20, 0.30, 0.40, 0.50]
NOISE_LEVELS = [2.0, 5.0, 10.0]

AUGMENTATION_SEEDS = [142, 242]

EVAL_SEEDS = [42, 43, 44, 45, 46]

RF_PARAMS = {
    "n_estimators": 300,
    "random_state": RANDOM_STATE,
    "n_jobs": -1
}

REGIMES = [
    "Clean",
    "Repeated Clean",
    "AP Dropout",
    "Combined"
]


# ------------------------------------------------------------
# Step 4 — Load dataset
# ------------------------------------------------------------

train_df = pd.read_csv(TRAIN_FILE)
test_df = pd.read_csv(TEST_FILE)

wifi_cols = [
    c for c in train_df.columns
    if c.upper().startswith("WAP")
]

assert len(train_df) == 19937
assert len(test_df) == 1111
assert len(wifi_cols) == 520

print("\nDATASET")
print("Training rows:", len(train_df))
print("Final-test rows:", len(test_df))
print("Wi-Fi features:", len(wifi_cols))


# ------------------------------------------------------------
# Step 5 — Labels
# ------------------------------------------------------------

def make_class_labels(df):

    return (
        df["BUILDINGID"].astype(int).astype(str)
        + "_"
        + df["FLOOR"].astype(int).astype(str)
    ).to_numpy()


y_all_cls = make_class_labels(train_df)
y_test_cls = make_class_labels(test_df)

COORD_COLS = [
    "LONGITUDE",
    "LATITUDE"
]

y_all_reg = (
    train_df[COORD_COLS]
    .to_numpy(dtype=np.float64)
)

y_test_reg = (
    test_df[COORD_COLS]
    .to_numpy(dtype=np.float64)
)


# ------------------------------------------------------------
# Step 6 — RSSI matrices and masks
# ------------------------------------------------------------

X_all_original = (
    train_df[wifi_cols]
    .to_numpy(dtype=np.float64)
)

X_test_original = (
    test_df[wifi_cols]
    .to_numpy(dtype=np.float64)
)

mask_all = (
    X_all_original != MISSING_ORIGINAL
)

mask_test = (
    X_test_original != MISSING_ORIGINAL
)

X_all_raw = X_all_original.copy()
X_test_raw = X_test_original.copy()

X_all_raw[~mask_all] = MISSING_RSSI
X_test_raw[~mask_test] = MISSING_RSSI


# ------------------------------------------------------------
# Step 7 — Reproduce Notebook 11 data split exactly
# ------------------------------------------------------------

all_indices = np.arange(
    len(train_df)
)

train_indices, dev_indices = train_test_split(
    all_indices,
    test_size=DEV_SIZE,
    random_state=RANDOM_STATE,
    stratify=y_all_cls
)

fit_indices, stop_indices = train_test_split(
    train_indices,
    test_size=EARLY_STOP_SIZE,
    random_state=RANDOM_STATE,
    stratify=y_all_cls[train_indices]
)

X_fit_raw = X_all_raw[fit_indices]
X_dev_raw = X_all_raw[dev_indices]

mask_fit = mask_all[fit_indices]
mask_dev = mask_all[dev_indices]

y_fit_cls = y_all_cls[fit_indices]
y_dev_cls = y_all_cls[dev_indices]

y_fit_reg = y_all_reg[fit_indices]
y_dev_reg = y_all_reg[dev_indices]

print("\nCONTROL SPLIT")
print("RF fitting fingerprints:", len(fit_indices))
print("Held-out early-stop fingerprints:", len(stop_indices))
print("Development fingerprints:", len(dev_indices))
print("Final-test fingerprints:", len(test_df))


# ------------------------------------------------------------
# Step 8 — Load the exact Notebook 11 feature scaler
# ------------------------------------------------------------

feature_scaler_path = (
    MODEL_11_DIR
    / "review_feature_scaler.joblib"
)

assert feature_scaler_path.exists(), feature_scaler_path

feature_scaler = joblib.load(
    feature_scaler_path
)

print("\nLoaded Notebook 11 feature scaler.")


# ------------------------------------------------------------
# Step 9 — Corruption functions
# ------------------------------------------------------------

def apply_ap_dropout(
    X_raw,
    observed_mask,
    rng,
    fixed_fraction=None
):

    X_corrupt = X_raw.copy()
    surviving_mask = observed_mask.copy()

    for i in range(
        X_raw.shape[0]
    ):

        observed = np.flatnonzero(
            observed_mask[i]
        )

        if len(observed) == 0:
            continue

        if fixed_fraction is None:

            p = float(
                rng.choice(AP_LEVELS)
            )

        else:

            p = float(
                fixed_fraction
            )

        if p <= 0:
            continue

        n_drop = int(
            np.floor(
                p * len(observed)
            )
        )

        n_drop = max(
            1,
            n_drop
        )

        n_drop = min(
            n_drop,
            len(observed)
        )

        drop_idx = rng.choice(
            observed,
            size=n_drop,
            replace=False
        )

        X_corrupt[
            i,
            drop_idx
        ] = MISSING_RSSI

        surviving_mask[
            i,
            drop_idx
        ] = False

    return (
        X_corrupt,
        surviving_mask
    )


def apply_rssi_noise(
    X_raw,
    surviving_mask,
    rng,
    fixed_sigma=None
):

    X_corrupt = X_raw.copy()

    for i in range(
        X_raw.shape[0]
    ):

        observed = np.flatnonzero(
            surviving_mask[i]
        )

        if len(observed) == 0:
            continue

        if fixed_sigma is None:

            sigma = float(
                rng.choice(
                    NOISE_LEVELS
                )
            )

        else:

            sigma = float(
                fixed_sigma
            )

        if sigma <= 0:
            continue

        noise = rng.normal(
            loc=0.0,
            scale=sigma,
            size=len(observed)
        )

        X_corrupt[
            i,
            observed
        ] += noise

        X_corrupt[
            i,
            observed
        ] = np.clip(
            X_corrupt[
                i,
                observed
            ],
            MIN_OBSERVED_RSSI,
            MAX_OBSERVED_RSSI
        )

    X_corrupt[
        ~surviving_mask
    ] = MISSING_RSSI

    return X_corrupt


def create_augmented_copy(
    X_raw,
    observed_mask,
    mode,
    seed
):

    rng = np.random.default_rng(
        seed
    )

    if mode == "ap_dropout":

        X_aug, _ = apply_ap_dropout(
            X_raw,
            observed_mask,
            rng,
            fixed_fraction=None
        )

        return X_aug

    if mode == "combined":

        X_aug, surviving = apply_ap_dropout(
            X_raw,
            observed_mask,
            rng,
            fixed_fraction=None
        )

        X_aug = apply_rssi_noise(
            X_aug,
            surviving,
            rng,
            fixed_sigma=None
        )

        return X_aug

    raise ValueError(
        mode
    )


# ------------------------------------------------------------
# Step 10 — Build equal-volume RF training regimes
# ------------------------------------------------------------

def build_training_regime(
    regime
):

    if regime == "Clean":

        return (
            X_fit_raw.copy(),
            y_fit_cls.copy(),
            y_fit_reg.copy()
        )

    if regime == "Repeated Clean":

        X = np.concatenate(
            [
                X_fit_raw,
                X_fit_raw,
                X_fit_raw
            ],
            axis=0
        )

        yc = np.concatenate(
            [
                y_fit_cls,
                y_fit_cls,
                y_fit_cls
            ],
            axis=0
        )

        yr = np.concatenate(
            [
                y_fit_reg,
                y_fit_reg,
                y_fit_reg
            ],
            axis=0
        )

        return X, yc, yr

    if regime == "AP Dropout":

        copies = [
            X_fit_raw
        ]

        for seed in AUGMENTATION_SEEDS:

            copies.append(
                create_augmented_copy(
                    X_fit_raw,
                    mask_fit,
                    mode="ap_dropout",
                    seed=seed
                )
            )

        X = np.concatenate(
            copies,
            axis=0
        )

        yc = np.tile(
            y_fit_cls,
            len(copies)
        )

        yr = np.tile(
            y_fit_reg,
            (
                len(copies),
                1
            )
        )

        return X, yc, yr

    if regime == "Combined":

        copies = [
            X_fit_raw
        ]

        for seed in AUGMENTATION_SEEDS:

            copies.append(
                create_augmented_copy(
                    X_fit_raw,
                    mask_fit,
                    mode="combined",
                    seed=seed
                )
            )

        X = np.concatenate(
            copies,
            axis=0
        )

        yc = np.tile(
            y_fit_cls,
            len(copies)
        )

        yr = np.tile(
            y_fit_reg,
            (
                len(copies),
                1
            )
        )

        return X, yc, yr

    raise ValueError(
        regime
    )


print("\nTRAINING REGIME SIZES")

for regime in REGIMES:

    X_tmp, _, _ = (
        build_training_regime(
            regime
        )
    )

    print(
        f"{regime:15s}:",
        len(X_tmp)
    )


# ------------------------------------------------------------
# Step 11 — Train RF classifier and regressor
# ------------------------------------------------------------

classifier_models = {}
regressor_models = {}

training_rows = []

for regime in REGIMES:

    print(
        "\n"
        + "=" * 75
    )

    print(
        "TRAINING RANDOM FOREST:",
        regime
    )

    print(
        "=" * 75
    )

    X_regime_raw, y_cls_regime, y_reg_regime = (
        build_training_regime(
            regime
        )
    )

    # Match the original project's common scaled-input convention.
    X_regime = feature_scaler.transform(
        X_regime_raw
    )

    start = time.time()

    classifier = RandomForestClassifier(
        **RF_PARAMS
    )

    classifier.fit(
        X_regime,
        y_cls_regime
    )

    classifier_models[
        regime
    ] = classifier

    joblib.dump(
        classifier,
        MODEL_DIR
        / (
            regime.lower()
            .replace(" ", "_")
            + "_rf_classifier.joblib"
        )
    )

    classifier_time = (
        time.time()
        - start
    )

    print(
        "Classifier trained."
    )

    start = time.time()

    regressor = RandomForestRegressor(
        **RF_PARAMS
    )

    regressor.fit(
        X_regime,
        y_reg_regime
    )

    regressor_models[
        regime
    ] = regressor

    joblib.dump(
        regressor,
        MODEL_DIR
        / (
            regime.lower()
            .replace(" ", "_")
            + "_rf_regressor.joblib"
        )
    )

    regressor_time = (
        time.time()
        - start
    )

    print(
        "Regressor trained."
    )

    training_rows.append(
        {
            "Training_Regime":
                regime,

            "Training_Samples":
                len(X_regime),

            "Classifier_Training_Seconds":
                classifier_time,

            "Regressor_Training_Seconds":
                regressor_time
        }
    )


training_df = pd.DataFrame(
    training_rows
)

training_df.to_csv(
    RESULTS_CSV
    / "12_rf_training_summary.csv",
    index=False
)

print("\nRF TRAINING SUMMARY")

display(
    training_df.round(2)
)


# ------------------------------------------------------------
# Step 12 — Localization metric
# ------------------------------------------------------------

def mean_euclidean_error(
    y_true,
    y_pred
):

    return np.mean(
        np.linalg.norm(
            y_pred - y_true,
            axis=1
        )
    )


# ------------------------------------------------------------
# Step 13 — Evaluate one corruption condition
# ------------------------------------------------------------

def evaluate_condition(
    X_raw,
    observed_mask,
    y_cls,
    y_reg,
    ap_loss,
    sigma,
    seed
):

    rng = np.random.default_rng(
        seed
    )

    X_corr = X_raw.copy()
    surviving = observed_mask.copy()

    if ap_loss > 0:

        X_corr, surviving = apply_ap_dropout(
            X_corr,
            surviving,
            rng,
            fixed_fraction=ap_loss
        )

    if sigma > 0:

        X_corr = apply_rssi_noise(
            X_corr,
            surviving,
            rng,
            fixed_sigma=sigma
        )

    X_scaled = feature_scaler.transform(
        X_corr
    )

    rows = []

    for regime in REGIMES:

        classifier = (
            classifier_models[
                regime
            ]
        )

        pred_cls = classifier.predict(
            X_scaled
        )

        accuracy = accuracy_score(
            y_cls,
            pred_cls
        )

        macro_f1 = f1_score(
            y_cls,
            pred_cls,
            average="macro"
        )

        regressor = (
            regressor_models[
                regime
            ]
        )

        pred_reg = regressor.predict(
            X_scaled
        )

        localization_errors = np.linalg.norm(
            pred_reg - y_reg,
            axis=1
        )

        mean_error = float(
            np.mean(localization_errors)
        )

        median_error = float(
            np.median(localization_errors)
        )

        p90_error = float(
            np.percentile(
                localization_errors,
                90
            )
        )

        rows.append(
            {
                "Training_Regime":
                    regime,

                "AP_Loss":
                    ap_loss,

                "RSSI_Sigma":
                    sigma,

                "Seed":
                    seed,

                "Accuracy":
                    accuracy,

                "Macro_F1":
                    macro_f1,

                "Mean_Localization_Error":
                    mean_error,

                "Median_Localization_Error":
                    median_error,

                "P90_Localization_Error":
                    p90_error
            }
        )

    return rows


# ------------------------------------------------------------
# Step 14 — Key reviewer conditions
# ------------------------------------------------------------

KEY_CONDITIONS = [
    (
        "Clean",
        0.0,
        0.0
    ),
    (
        "50% AP Loss",
        0.50,
        0.0
    ),
    (
        "RSSI sigma=10",
        0.0,
        10.0
    ),
    (
        "50% AP Loss + sigma=10",
        0.50,
        10.0
    )
]


# ------------------------------------------------------------
# Step 15 — Development evaluation
# ------------------------------------------------------------

dev_rows = []

for (
    condition_name,
    ap_loss,
    sigma
) in KEY_CONDITIONS:

    seeds = (
        [42]
        if (
            ap_loss == 0
            and sigma == 0
        )
        else EVAL_SEEDS
    )

    for seed in seeds:

        rows = evaluate_condition(
            X_dev_raw,
            mask_dev,
            y_dev_cls,
            y_dev_reg,
            ap_loss,
            sigma,
            seed
        )

        for row in rows:

            row["Dataset"] = (
                "Development"
            )

            row["Condition"] = (
                condition_name
            )

        dev_rows.extend(
            rows
        )


dev_detailed = pd.DataFrame(
    dev_rows
)

dev_summary = (
    dev_detailed
    .groupby(
        [
            "Dataset",
            "Condition",
            "Training_Regime"
        ],
        as_index=False
    )
    .agg(
        Accuracy=(
            "Accuracy",
            "mean"
        ),
        Accuracy_SD=(
            "Accuracy",
            "std"
        ),
        Macro_F1=(
            "Macro_F1",
            "mean"
        ),
        Mean_Localization_Error=(
            "Mean_Localization_Error",
            "mean"
        ),
        Localization_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),
        Median_Localization_Error=(
            "Median_Localization_Error",
            "mean"
        ),
        Median_Localization_Error_SD=(
            "Median_Localization_Error",
            "std"
        ),
        P90_Localization_Error=(
            "P90_Localization_Error",
            "mean"
        ),
        P90_Localization_Error_SD=(
            "P90_Localization_Error",
            "std"
        )
    )
)


dev_detailed.to_csv(
    RESULTS_CSV
    / "12_rf_development_detailed.csv",
    index=False
)

dev_summary.to_csv(
    RESULTS_CSV
    / "12_rf_development_summary.csv",
    index=False
)

print(
    "\nRF DEVELOPMENT RESULTS"
)

display(
    dev_summary.round(4)
)


# ------------------------------------------------------------
# Step 16 — Reviewer-requested post-hoc final-test evaluation
# ------------------------------------------------------------

test_rows = []

for (
    condition_name,
    ap_loss,
    sigma
) in KEY_CONDITIONS:

    seeds = (
        [42]
        if (
            ap_loss == 0
            and sigma == 0
        )
        else EVAL_SEEDS
    )

    for seed in seeds:

        rows = evaluate_condition(
            X_test_raw,
            mask_test,
            y_test_cls,
            y_test_reg,
            ap_loss,
            sigma,
            seed
        )

        for row in rows:

            row["Dataset"] = (
                "Final Test — Post-hoc Reviewer Control"
            )

            row["Condition"] = (
                condition_name
            )

        test_rows.extend(
            rows
        )


test_detailed = pd.DataFrame(
    test_rows
)

test_summary = (
    test_detailed
    .groupby(
        [
            "Dataset",
            "Condition",
            "Training_Regime"
        ],
        as_index=False
    )
    .agg(
        Accuracy=(
            "Accuracy",
            "mean"
        ),
        Accuracy_SD=(
            "Accuracy",
            "std"
        ),
        Macro_F1=(
            "Macro_F1",
            "mean"
        ),
        Mean_Localization_Error=(
            "Mean_Localization_Error",
            "mean"
        ),
        Localization_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),
        Median_Localization_Error=(
            "Median_Localization_Error",
            "mean"
        ),
        Median_Localization_Error_SD=(
            "Median_Localization_Error",
            "std"
        ),
        P90_Localization_Error=(
            "P90_Localization_Error",
            "mean"
        ),
        P90_Localization_Error_SD=(
            "P90_Localization_Error",
            "std"
        )
    )
)


test_detailed.to_csv(
    RESULTS_CSV
    / "12_rf_final_test_detailed.csv",
    index=False
)

test_summary.to_csv(
    RESULTS_CSV
    / "12_rf_final_test_summary.csv",
    index=False
)


print(
    "\nRF FINAL-TEST POST-HOC RESULTS"
)

display(
    test_summary.round(4)
)


# ------------------------------------------------------------
# Step 17 — Full 24-condition RF corruption grid
# ------------------------------------------------------------

FULL_AP_LEVELS = [
    0.0,
    0.10,
    0.20,
    0.30,
    0.40,
    0.50
]

FULL_NOISE_LEVELS = [
    0.0,
    2.0,
    5.0,
    10.0
]

grid_rows = []

for ap_loss in FULL_AP_LEVELS:

    for sigma in FULL_NOISE_LEVELS:

        seeds = (
            [42]
            if (
                ap_loss == 0
                and sigma == 0
            )
            else EVAL_SEEDS
        )

        for seed in seeds:

            rows = evaluate_condition(
                X_test_raw,
                mask_test,
                y_test_cls,
                y_test_reg,
                ap_loss,
                sigma,
                seed
            )

            grid_rows.extend(
                rows
            )


grid_detailed = pd.DataFrame(
    grid_rows
)

grid_summary = (
    grid_detailed
    .groupby(
        [
            "AP_Loss",
            "RSSI_Sigma",
            "Training_Regime"
        ],
        as_index=False
    )
    .agg(
        Accuracy=(
            "Accuracy",
            "mean"
        ),
        Macro_F1=(
            "Macro_F1",
            "mean"
        ),
        Mean_Localization_Error=(
            "Mean_Localization_Error",
            "mean"
        ),
        Median_Localization_Error=(
            "Median_Localization_Error",
            "mean"
        ),
        P90_Localization_Error=(
            "P90_Localization_Error",
            "mean"
        )
    )
)


grid_detailed.to_csv(
    RESULTS_CSV
    / "12_rf_full_grid_detailed.csv",
    index=False
)

grid_summary.to_csv(
    RESULTS_CSV
    / "12_rf_full_grid_summary.csv",
    index=False
)


# ------------------------------------------------------------
# Step 18 — Direct reviewer answer
# ------------------------------------------------------------

severe = test_summary[
    test_summary["Condition"]
    == "50% AP Loss + sigma=10"
].copy()


def value(
    regime,
    column
):

    return float(
        severe.loc[
            severe["Training_Regime"]
            == regime,
            column
        ].iloc[0]
    )


clean_acc = value(
    "Clean",
    "Accuracy"
)

repeated_acc = value(
    "Repeated Clean",
    "Accuracy"
)

dropout_acc = value(
    "AP Dropout",
    "Accuracy"
)

combined_acc = value(
    "Combined",
    "Accuracy"
)


clean_err = value(
    "Clean",
    "Mean_Localization_Error"
)

repeated_err = value(
    "Repeated Clean",
    "Mean_Localization_Error"
)

dropout_err = value(
    "AP Dropout",
    "Mean_Localization_Error"
)

combined_err = value(
    "Combined",
    "Mean_Localization_Error"
)


print(
    "\n"
    + "=" * 80
)

print(
    "NOTEBOOK 12 — RANDOM FOREST REVIEWER CONTROL"
)

print(
    "=" * 80
)

print(
    "\nSEVERE CONDITION:"
)

print(
    "50% AP loss + RSSI Gaussian noise sigma=10 dB"
)

print(
    "\nCLASSIFICATION"
)

print(
    f"Clean RF           = {clean_acc:.4f}"
)

print(
    f"Repeated Clean RF  = {repeated_acc:.4f}"
)

print(
    f"AP-Dropout RF      = {dropout_acc:.4f}"
)

print(
    f"Combined RF        = {combined_acc:.4f}"
)

print(
    "\nAP-Dropout gain over equal-volume Repeated Clean:"
)

print(
    f"{dropout_acc - repeated_acc:.4f}"
)

print(
    "Percentage points:"
)

print(
    f"{100 * (dropout_acc - repeated_acc):.2f}"
)


print(
    "\nLOCALIZATION"
)

print(
    f"Clean RF           = {clean_err:.4f}"
)

print(
    f"Repeated Clean RF  = {repeated_err:.4f}"
)

print(
    f"AP-Dropout RF      = {dropout_err:.4f}"
)

print(
    f"Combined RF        = {combined_err:.4f}"
)

print(
    "\nCombined RF error reduction vs equal-volume Repeated Clean:"
)

print(
    f"{repeated_err - combined_err:.4f}"
)


# ------------------------------------------------------------
# Step 19 — Grid-level comparison
# ------------------------------------------------------------

overall_grid = (
    grid_summary
    .groupby(
        "Training_Regime",
        as_index=False
    )
    .agg(
        Mean_Accuracy_Across_Grid=(
            "Accuracy",
            "mean"
        ),
        Mean_F1_Across_Grid=(
            "Macro_F1",
            "mean"
        ),
        Mean_Error_Across_Grid=(
            "Mean_Localization_Error",
            "mean"
        ),
        Mean_Median_Error_Across_Grid=(
            "Median_Localization_Error",
            "mean"
        ),
        Mean_P90_Error_Across_Grid=(
            "P90_Localization_Error",
            "mean"
        )
    )
)


overall_grid.to_csv(
    RESULTS_CSV
    / "12_rf_grid_overall_summary.csv",
    index=False
)

print(
    "\nFULL 24-CONDITION GRID"
)

display(
    overall_grid.round(4)
)


# ------------------------------------------------------------
# Step 20 — Severe-condition figures
# ------------------------------------------------------------

plot_order = [
    "Clean",
    "Repeated Clean",
    "AP Dropout",
    "Combined"
]

plot_data = (
    severe
    .set_index(
        "Training_Regime"
    )
    .loc[
        plot_order
    ]
)


plt.figure(
    figsize=(8, 5)
)

plt.bar(
    plot_data.index,
    plot_data["Accuracy"]
)

plt.ylabel(
    "Classification Accuracy"
)

plt.title(
    "Random Forest Robust Training Under Severe Corruption"
)

plt.xticks(
    rotation=15
)

plt.tight_layout()

plt.savefig(
    RESULTS_FIG
    / "12_rf_severe_classification.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


plt.figure(
    figsize=(8, 5)
)

plt.bar(
    plot_data.index,
    plot_data["Mean_Localization_Error"]
)

plt.ylabel(
    "Mean Localization Error (coordinate units)"
)

plt.title(
    "Random Forest Localization Under Severe Corruption"
)

plt.xticks(
    rotation=15
)

plt.tight_layout()

plt.savefig(
    RESULTS_FIG
    / "12_rf_severe_localization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 21 — Save metadata
# ------------------------------------------------------------

metadata = {
    "experiment":
        "Reviewer-requested Random Forest augmentation control",

    "post_hoc":
        True,

    "random_state":
        RANDOM_STATE,

    "rf_parameters":
        RF_PARAMS,

    "augmentation_seeds":
        AUGMENTATION_SEEDS,

    "evaluation_seeds":
        EVAL_SEEDS,

    "training_regimes":
        REGIMES,

    "note":
        (
            "Training data and corruption rules mirror Notebook 11. "
            "Random Forest robust training is a reviewer-requested "
            "post-hoc architecture-generalization control."
        )
}


with open(
    RESULTS_CSV
    / "12_rf_metadata.json",
    "w"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )


# ------------------------------------------------------------
# Step 22 — Completion
# ------------------------------------------------------------

print(
    "\n"
    + "=" * 80
)

print(
    "NOTEBOOK 12 COMPLETED SUCCESSFULLY"
)

print(
    "=" * 80
)

print(
    "\nSaved models:"
)

for path in sorted(
    MODEL_DIR.glob("*.joblib")
):

    print(
        "-",
        path.name
    )


print(
    "\nSaved CSV files:"
)

for path in sorted(
    RESULTS_CSV.glob("12_*")
):

    print(
        "-",
        path.name
    )


print(
    "\nSaved figures:"
)

for path in sorted(
    RESULTS_FIG.glob("12_*")
):

    print(
        "-",
        path.name
    )

print(
    "\nIMPORTANT:"
)

print(
    "These Random Forest results are reviewer-requested "
    "post-hoc controls and do not replace the original frozen "
    "model-selection results."
)
```
