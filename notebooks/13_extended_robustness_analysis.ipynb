```python
# ============================================================
# NOTEBOOK 13
# Extended Robustness Analysis for Reviewer Revision
#
# Reviewer concerns addressed:
#
# Major concern #2:
#   Add a structured/global AP-outage corruption model.
#
# Major concern #6:
#   Increase corruption realizations from 5 to 20 seeds.
#
# Minor concerns:
#   - Confusion matrices
#   - Spatial localization-error visualizations
#   - Explicit random seeds / bootstrap procedure
#
# IMPORTANT:
#   These are reviewer-requested post-hoc analyses.
#   They do NOT replace the originally frozen experiment.
#   Round-two export adds Macro-F1 plus Mean, Median, and P90
#   localization-error summaries to machine-readable outputs.
# ============================================================


# ------------------------------------------------------------
# Step 1 — Imports
# ------------------------------------------------------------

from pathlib import Path
import json
import platform
import os
import sys
import warnings

import numpy as np
import pandas as pd
import joblib
import matplotlib.pyplot as plt

import sklearn

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# Step 2 — Project paths
# ------------------------------------------------------------

PROJECT_ROOT = (
    Path.home()
    / "Documents"
    / "ADU"
    / "wifi-localization-research"
)

TRAIN_FILE = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "trainingData.csv"
)

TEST_FILE = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "validationData.csv"
)

MODEL_11_DIR = (
    PROJECT_ROOT
    / "models"
    / "11_equal_volume_control"
)

MODEL_12_DIR = (
    PROJECT_ROOT
    / "models"
    / "12_random_forest_robust_training"
)

RESULTS_CSV = (
    PROJECT_ROOT
    / "results"
    / "csv"
)

RESULTS_FIG = (
    PROJECT_ROOT
    / "results"
    / "figures"
)

RESULTS_CSV.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_FIG.mkdir(
    parents=True,
    exist_ok=True
)

assert TRAIN_FILE.exists()
assert TEST_FILE.exists()

print("PROJECT ROOT:", PROJECT_ROOT)


# ------------------------------------------------------------
# Step 3 — Constants
# ------------------------------------------------------------

RANDOM_STATE = 42

MISSING_ORIGINAL = 100
MISSING_RSSI = -110.0

MIN_OBSERVED_RSSI = -104.0
MAX_OBSERVED_RSSI = 0.0

# Reviewer-requested increase from 5 to 20 corruption seeds.
EVAL_SEEDS_20 = list(
    range(42, 62)
)

# Structured/global AP-outage levels.
GLOBAL_OUTAGE_LEVELS = [
    0.10,
    0.30,
    0.50
]

# Evaluate structured outage both without and with RSSI noise.
GLOBAL_NOISE_LEVELS = [
    0.0,
    10.0
]

SEVERE_AP_LOSS = 0.50
SEVERE_SIGMA = 10.0

BOOTSTRAP_REPS = 5000
BOOTSTRAP_SEED = 13042

print("\nEXTENDED EXPERIMENT SETTINGS")
print("Evaluation seeds:", EVAL_SEEDS_20)
print("Number of seeds:", len(EVAL_SEEDS_20))
print("Global outage levels:", GLOBAL_OUTAGE_LEVELS)
print("Global noise levels:", GLOBAL_NOISE_LEVELS)
print("Bootstrap repetitions:", BOOTSTRAP_REPS)


# ------------------------------------------------------------
# Step 4 — Load raw dataset
# ------------------------------------------------------------

train_df = pd.read_csv(
    TRAIN_FILE
)

test_df = pd.read_csv(
    TEST_FILE
)

wifi_cols = [
    c for c in train_df.columns
    if c.upper().startswith("WAP")
]

assert len(train_df) == 19937
assert len(test_df) == 1111
assert len(wifi_cols) == 520

print("\nDATASET")
print("Training fingerprints:", len(train_df))
print("Final-test fingerprints:", len(test_df))
print("Wi-Fi AP features:", len(wifi_cols))


# ------------------------------------------------------------
# Step 5 — Targets
# ------------------------------------------------------------

def make_class_labels(df):

    return (
        df["BUILDINGID"]
        .astype(int)
        .astype(str)
        + "_"
        + df["FLOOR"]
        .astype(int)
        .astype(str)
    ).to_numpy()


y_test_cls = make_class_labels(
    test_df
)

COORD_COLS = [
    "LONGITUDE",
    "LATITUDE"
]

y_test_reg = (
    test_df[
        COORD_COLS
    ]
    .to_numpy(
        dtype=np.float64
    )
)


# ------------------------------------------------------------
# Step 6 — Raw RSSI and observation mask
# ------------------------------------------------------------

X_test_original = (
    test_df[
        wifi_cols
    ]
    .to_numpy(
        dtype=np.float64
    )
)

mask_test = (
    X_test_original
    != MISSING_ORIGINAL
)

X_test_raw = (
    X_test_original.copy()
)

X_test_raw[
    ~mask_test
] = MISSING_RSSI

observed_counts = (
    mask_test.sum(
        axis=1
    )
)

print("\nFINAL TEST OBSERVATION STATISTICS")
print(
    "Mean observed APs:",
    round(
        observed_counts.mean(),
        2
    )
)
print(
    "Median observed APs:",
    float(
        np.median(
            observed_counts
        )
    )
)


# ------------------------------------------------------------
# Step 7 — Load Notebook 11 scalers
# ------------------------------------------------------------

feature_scaler = joblib.load(
    MODEL_11_DIR
    / "review_feature_scaler.joblib"
)

coordinate_scaler = joblib.load(
    MODEL_11_DIR
    / "review_coordinate_scaler.joblib"
)

print("\nLoaded reviewer-control scalers.")


# ------------------------------------------------------------
# Step 8 — Load MLP equal-volume models
#
# Baseline:
#   Repeated Clean
#
# Robust:
#   AP Dropout for classification
#   Combined for localization
# ------------------------------------------------------------

mlp_repeated_clf = joblib.load(
    MODEL_11_DIR
    / "repeated_clean_classifier.joblib"
)

mlp_dropout_clf = joblib.load(
    MODEL_11_DIR
    / "ap_dropout_classifier.joblib"
)

mlp_repeated_reg = joblib.load(
    MODEL_11_DIR
    / "repeated_clean_regressor.joblib"
)

mlp_combined_reg = joblib.load(
    MODEL_11_DIR
    / "combined_regressor.joblib"
)


# ------------------------------------------------------------
# Step 9 — Load Random Forest equal-volume models
# ------------------------------------------------------------

rf_repeated_clf = joblib.load(
    MODEL_12_DIR
    / "repeated_clean_rf_classifier.joblib"
)

rf_dropout_clf = joblib.load(
    MODEL_12_DIR
    / "ap_dropout_rf_classifier.joblib"
)

rf_repeated_reg = joblib.load(
    MODEL_12_DIR
    / "repeated_clean_rf_regressor.joblib"
)

rf_combined_reg = joblib.load(
    MODEL_12_DIR
    / "combined_rf_regressor.joblib"
)

print("Loaded MLP and Random Forest reviewer-control models.")


# ------------------------------------------------------------
# Step 10 — Independent per-fingerprint AP dropout
# ------------------------------------------------------------

def apply_independent_ap_dropout(
    X_raw,
    observed_mask,
    rng,
    fraction
):

    X_corrupt = (
        X_raw.copy()
    )

    surviving_mask = (
        observed_mask.copy()
    )

    for i in range(
        X_raw.shape[0]
    ):

        observed = np.flatnonzero(
            observed_mask[i]
        )

        if len(observed) == 0:
            continue

        n_drop = int(
            np.floor(
                fraction
                * len(observed)
            )
        )

        if fraction > 0:

            n_drop = max(
                1,
                n_drop
            )

        n_drop = min(
            n_drop,
            len(observed)
        )

        if n_drop == 0:
            continue

        selected = rng.choice(
            observed,
            size=n_drop,
            replace=False
        )

        X_corrupt[
            i,
            selected
        ] = MISSING_RSSI

        surviving_mask[
            i,
            selected
        ] = False

    return (
        X_corrupt,
        surviving_mask
    )


# ------------------------------------------------------------
# Step 11 — Gaussian RSSI noise
# ------------------------------------------------------------

def apply_rssi_noise(
    X_raw,
    surviving_mask,
    rng,
    sigma
):

    X_corrupt = (
        X_raw.copy()
    )

    if sigma <= 0:

        return X_corrupt

    for i in range(
        X_corrupt.shape[0]
    ):

        observed = np.flatnonzero(
            surviving_mask[i]
        )

        if len(observed) == 0:
            continue

        noise = rng.normal(
            loc=0.0,
            scale=sigma,
            size=len(observed)
        )

        X_corrupt[
            i,
            observed
        ] += noise

        X_corrupt[
            i,
            observed
        ] = np.clip(
            X_corrupt[
                i,
                observed
            ],
            MIN_OBSERVED_RSSI,
            MAX_OBSERVED_RSSI
        )

    X_corrupt[
        ~surviving_mask
    ] = MISSING_RSSI

    return X_corrupt


# ------------------------------------------------------------
# Step 12 — Structured / global AP outage
#
# A selected AP identity disappears consistently wherever
# it is observed, rather than being sampled independently
# for every fingerprint.
# ------------------------------------------------------------

def apply_global_ap_outage(
    X_raw,
    observed_mask,
    rng,
    outage_fraction
):

    X_corrupt = (
        X_raw.copy()
    )

    surviving_mask = (
        observed_mask.copy()
    )

    # AP identities that appear at least once in the
    # evaluation set.
    active_ap_ids = np.flatnonzero(
        observed_mask.any(
            axis=0
        )
    )

    n_outage = int(
        np.floor(
            outage_fraction
            * len(active_ap_ids)
        )
    )

    if outage_fraction > 0:

        n_outage = max(
            1,
            n_outage
        )

    selected_ap_ids = rng.choice(
        active_ap_ids,
        size=n_outage,
        replace=False
    )

    # Same AP identities are removed from every fingerprint.
    X_corrupt[
        :,
        selected_ap_ids
    ] = MISSING_RSSI

    surviving_mask[
        :,
        selected_ap_ids
    ] = False

    # Measure actual fraction of observed evidence removed
    # from each fingerprint.
    removed_counts = (
        observed_mask[
            :,
            selected_ap_ids
        ]
        .sum(
            axis=1
        )
    )

    effective_fraction = np.divide(
        removed_counts,
        observed_counts,
        out=np.zeros_like(
            removed_counts,
            dtype=float
        ),
        where=observed_counts > 0
    )

    return (
        X_corrupt,
        surviving_mask,
        selected_ap_ids,
        effective_fraction
    )


# ------------------------------------------------------------
# Step 13 — Prediction helper
# ------------------------------------------------------------

def inverse_regression(
    model,
    X_scaled
):

    pred_scaled = (
        model.predict(
            X_scaled
        )
    )

    return (
        coordinate_scaler
        .inverse_transform(
            pred_scaled
        )
    )


def euclidean_errors(
    y_true,
    y_pred
):

    return np.linalg.norm(
        y_pred - y_true,
        axis=1
    )


# ------------------------------------------------------------
# Step 14 — Evaluate one corrupted matrix
# ------------------------------------------------------------

def evaluate_models(
    X_corrupt_raw
):

    X_scaled = (
        feature_scaler
        .transform(
            X_corrupt_raw
        )
    )

    results = {}

    # --------------------------------------------------------
    # MLP classification
    # --------------------------------------------------------

    mlp_base_pred = (
        mlp_repeated_clf
        .predict(
            X_scaled
        )
    )

    mlp_robust_pred = (
        mlp_dropout_clf
        .predict(
            X_scaled
        )
    )

    # --------------------------------------------------------
    # MLP localization
    # --------------------------------------------------------

    mlp_base_reg = inverse_regression(
        mlp_repeated_reg,
        X_scaled
    )

    mlp_robust_reg = inverse_regression(
        mlp_combined_reg,
        X_scaled
    )

    mlp_base_err = euclidean_errors(
        y_test_reg,
        mlp_base_reg
    )

    mlp_robust_err = euclidean_errors(
        y_test_reg,
        mlp_robust_reg
    )

    # --------------------------------------------------------
    # RF classification
    # --------------------------------------------------------

    rf_base_pred = (
        rf_repeated_clf
        .predict(
            X_scaled
        )
    )

    rf_robust_pred = (
        rf_dropout_clf
        .predict(
            X_scaled
        )
    )

    # --------------------------------------------------------
    # RF localization
    # --------------------------------------------------------

    rf_base_reg = (
        rf_repeated_reg
        .predict(
            X_scaled
        )
    )

    rf_robust_reg = (
        rf_combined_reg
        .predict(
            X_scaled
        )
    )

    rf_base_err = euclidean_errors(
        y_test_reg,
        rf_base_reg
    )

    rf_robust_err = euclidean_errors(
        y_test_reg,
        rf_robust_reg
    )

    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------

    results[
        "MLP"
    ] = {
        "baseline_accuracy":
            accuracy_score(
                y_test_cls,
                mlp_base_pred
            ),

        "robust_accuracy":
            accuracy_score(
                y_test_cls,
                mlp_robust_pred
            ),

        "baseline_f1":
            f1_score(
                y_test_cls,
                mlp_base_pred,
                average="macro"
            ),

        "robust_f1":
            f1_score(
                y_test_cls,
                mlp_robust_pred,
                average="macro"
            ),

        "baseline_mean_error":
            float(np.mean(mlp_base_err)),

        "robust_mean_error":
            float(np.mean(mlp_robust_err)),

        "baseline_median_error":
            float(np.median(mlp_base_err)),

        "robust_median_error":
            float(np.median(mlp_robust_err)),

        "baseline_p90_error":
            float(np.percentile(mlp_base_err, 90)),

        "robust_p90_error":
            float(np.percentile(mlp_robust_err, 90)),

        "baseline_correct":
            (
                mlp_base_pred
                == y_test_cls
            ).astype(float),

        "robust_correct":
            (
                mlp_robust_pred
                == y_test_cls
            ).astype(float),

        "baseline_errors":
            mlp_base_err,

        "robust_errors":
            mlp_robust_err,

        "baseline_class_predictions":
            mlp_base_pred,

        "robust_class_predictions":
            mlp_robust_pred,

        "baseline_coordinates":
            mlp_base_reg,

        "robust_coordinates":
            mlp_robust_reg
    }


    results[
        "RF"
    ] = {
        "baseline_accuracy":
            accuracy_score(
                y_test_cls,
                rf_base_pred
            ),

        "robust_accuracy":
            accuracy_score(
                y_test_cls,
                rf_robust_pred
            ),

        "baseline_f1":
            f1_score(
                y_test_cls,
                rf_base_pred,
                average="macro"
            ),

        "robust_f1":
            f1_score(
                y_test_cls,
                rf_robust_pred,
                average="macro"
            ),

        "baseline_mean_error":
            float(np.mean(rf_base_err)),

        "robust_mean_error":
            float(np.mean(rf_robust_err)),

        "baseline_median_error":
            float(np.median(rf_base_err)),

        "robust_median_error":
            float(np.median(rf_robust_err)),

        "baseline_p90_error":
            float(np.percentile(rf_base_err, 90)),

        "robust_p90_error":
            float(np.percentile(rf_robust_err, 90)),

        "baseline_correct":
            (
                rf_base_pred
                == y_test_cls
            ).astype(float),

        "robust_correct":
            (
                rf_robust_pred
                == y_test_cls
            ).astype(float),

        "baseline_errors":
            rf_base_err,

        "robust_errors":
            rf_robust_err,

        "baseline_class_predictions":
            rf_base_pred,

        "robust_class_predictions":
            rf_robust_pred,

        "baseline_coordinates":
            rf_base_reg,

        "robust_coordinates":
            rf_robust_reg
    }

    return results


# ------------------------------------------------------------
# Step 15 — 20-seed severe independent corruption
# ------------------------------------------------------------

severe_rows = []

paired_effects = {
    "MLP": {
        "classification": [],
        "localization": []
    },
    "RF": {
        "classification": [],
        "localization": []
    }
}

last_severe_results = None

for seed in EVAL_SEEDS_20:

    rng = np.random.default_rng(
        seed
    )

    X_corr, surviving = (
        apply_independent_ap_dropout(
            X_test_raw,
            mask_test,
            rng,
            fraction=SEVERE_AP_LOSS
        )
    )

    X_corr = apply_rssi_noise(
        X_corr,
        surviving,
        rng,
        sigma=SEVERE_SIGMA
    )

    results = evaluate_models(
        X_corr
    )

    last_severe_results = (
        results
    )

    for architecture in [
        "MLP",
        "RF"
    ]:

        r = results[
            architecture
        ]

        severe_rows.append(
            {
                "Architecture":
                    architecture,

                "Seed":
                    seed,

                "AP_Loss":
                    SEVERE_AP_LOSS,

                "RSSI_Sigma":
                    SEVERE_SIGMA,

                "Baseline_Accuracy":
                    r[
                        "baseline_accuracy"
                    ],

                "Robust_Accuracy":
                    r[
                        "robust_accuracy"
                    ],

                "Accuracy_Gain":
                    (
                        r[
                            "robust_accuracy"
                        ]
                        -
                        r[
                            "baseline_accuracy"
                        ]
                    ),

                "Baseline_F1":
                    r[
                        "baseline_f1"
                    ],

                "Robust_F1":
                    r[
                        "robust_f1"
                    ],

                "Baseline_Error":
                    r[
                        "baseline_mean_error"
                    ],

                "Robust_Error":
                    r[
                        "robust_mean_error"
                    ],

                "Error_Reduction":
                    (
                        r[
                            "baseline_mean_error"
                        ]
                        -
                        r[
                            "robust_mean_error"
                        ]
                    ),

                "Baseline_Median_Error":
                    r[
                        "baseline_median_error"
                    ],

                "Robust_Median_Error":
                    r[
                        "robust_median_error"
                    ],

                "Median_Error_Reduction":
                    (
                        r[
                            "baseline_median_error"
                        ]
                        -
                        r[
                            "robust_median_error"
                        ]
                    ),

                "Baseline_P90_Error":
                    r[
                        "baseline_p90_error"
                    ],

                "Robust_P90_Error":
                    r[
                        "robust_p90_error"
                    ],

                "P90_Error_Reduction":
                    (
                        r[
                            "baseline_p90_error"
                        ]
                        -
                        r[
                            "robust_p90_error"
                        ]
                    )
            }
        )

        paired_effects[
            architecture
        ][
            "classification"
        ].append(
            (
                r[
                    "robust_correct"
                ]
                -
                r[
                    "baseline_correct"
                ]
            )
        )

        paired_effects[
            architecture
        ][
            "localization"
        ].append(
            (
                r[
                    "baseline_errors"
                ]
                -
                r[
                    "robust_errors"
                ]
            )
        )


severe_df = pd.DataFrame(
    severe_rows
)

severe_df.to_csv(
    RESULTS_CSV
    / "13_twenty_seed_severe_results.csv",
    index=False
)


# ------------------------------------------------------------
# Step 16 — 20-seed summary
# ------------------------------------------------------------

severe_summary = (
    severe_df
    .groupby(
        "Architecture",
        as_index=False
    )
    .agg(
        Baseline_Accuracy=(
            "Baseline_Accuracy",
            "mean"
        ),

        Robust_Accuracy=(
            "Robust_Accuracy",
            "mean"
        ),

        Accuracy_Gain=(
            "Accuracy_Gain",
            "mean"
        ),

        Accuracy_Gain_SD=(
            "Accuracy_Gain",
            "std"
        ),

        Baseline_F1=(
            "Baseline_F1",
            "mean"
        ),

        Robust_F1=(
            "Robust_F1",
            "mean"
        ),

        Baseline_Error=(
            "Baseline_Error",
            "mean"
        ),

        Robust_Error=(
            "Robust_Error",
            "mean"
        ),

        Error_Reduction=(
            "Error_Reduction",
            "mean"
        ),

        Error_Reduction_SD=(
            "Error_Reduction",
            "std"
        ),

        Baseline_Median_Error=(
            "Baseline_Median_Error",
            "mean"
        ),

        Robust_Median_Error=(
            "Robust_Median_Error",
            "mean"
        ),

        Median_Error_Reduction=(
            "Median_Error_Reduction",
            "mean"
        ),

        Baseline_P90_Error=(
            "Baseline_P90_Error",
            "mean"
        ),

        Robust_P90_Error=(
            "Robust_P90_Error",
            "mean"
        ),

        P90_Error_Reduction=(
            "P90_Error_Reduction",
            "mean"
        )
    )
)

severe_summary.to_csv(
    RESULTS_CSV
    / "13_twenty_seed_severe_summary.csv",
    index=False
)

print("\n20-SEED SEVERE-CONDITION SUMMARY")
display(
    severe_summary.round(4)
)


# ------------------------------------------------------------
# Step 17 — Hierarchical paired bootstrap
#
# Procedure:
#   1. Resample corruption seeds with replacement.
#   2. For each selected seed, resample fingerprints with
#      replacement.
#   3. Compute mean paired effect.
#
# Positive classification effect = robust better.
# Positive localization effect = robust lower error.
# ------------------------------------------------------------

def hierarchical_bootstrap(
    seed_sample_effects,
    n_boot=5000,
    random_seed=13042
):

    effects = np.asarray(
        seed_sample_effects,
        dtype=float
    )

    n_seeds, n_samples = (
        effects.shape
    )

    rng = np.random.default_rng(
        random_seed
    )

    boot = np.empty(
        n_boot,
        dtype=float
    )

    for b in range(
        n_boot
    ):

        selected_seeds = rng.integers(
            0,
            n_seeds,
            size=n_seeds
        )

        seed_means = []

        for seed_idx in selected_seeds:

            sample_idx = rng.integers(
                0,
                n_samples,
                size=n_samples
            )

            seed_means.append(
                effects[
                    seed_idx,
                    sample_idx
                ].mean()
            )

        boot[b] = np.mean(
            seed_means
        )

    estimate = (
        effects.mean()
    )

    lower, upper = np.percentile(
        boot,
        [
            2.5,
            97.5
        ]
    )

    return (
        estimate,
        lower,
        upper,
        boot
    )


bootstrap_rows = []

bootstrap_distributions = {}

for architecture in [
    "MLP",
    "RF"
]:

    cls_est, cls_low, cls_high, cls_boot = (
        hierarchical_bootstrap(
            paired_effects[
                architecture
            ][
                "classification"
            ],
            n_boot=BOOTSTRAP_REPS,
            random_seed=BOOTSTRAP_SEED
        )
    )

    reg_est, reg_low, reg_high, reg_boot = (
        hierarchical_bootstrap(
            paired_effects[
                architecture
            ][
                "localization"
            ],
            n_boot=BOOTSTRAP_REPS,
            random_seed=(
                BOOTSTRAP_SEED
                + 1
            )
        )
    )

    bootstrap_distributions[
        architecture
    ] = {
        "classification":
            cls_boot,

        "localization":
            reg_boot
    }

    bootstrap_rows.append(
        {
            "Architecture":
                architecture,

            "Metric":
                "Accuracy gain",

            "Estimate":
                cls_est,

            "CI_Lower":
                cls_low,

            "CI_Upper":
                cls_high
        }
    )

    bootstrap_rows.append(
        {
            "Architecture":
                architecture,

            "Metric":
                "Localization error reduction",

            "Estimate":
                reg_est,

            "CI_Lower":
                reg_low,

            "CI_Upper":
                reg_high
        }
    )


bootstrap_df = pd.DataFrame(
    bootstrap_rows
)

bootstrap_df.to_csv(
    RESULTS_CSV
    / "13_twenty_seed_bootstrap_summary.csv",
    index=False
)

print("\n20-SEED HIERARCHICAL BOOTSTRAP")
display(
    bootstrap_df.round(4)
)


# ------------------------------------------------------------
# Step 18 — Structured/global AP-outage experiment
# ------------------------------------------------------------

global_rows = []

for outage_fraction in GLOBAL_OUTAGE_LEVELS:

    for sigma in GLOBAL_NOISE_LEVELS:

        for seed in EVAL_SEEDS_20:

            rng = np.random.default_rng(
                seed
            )

            (
                X_global,
                surviving,
                selected_aps,
                effective_fraction
            ) = apply_global_ap_outage(
                X_test_raw,
                mask_test,
                rng,
                outage_fraction
            )

            if sigma > 0:

                X_global = apply_rssi_noise(
                    X_global,
                    surviving,
                    rng,
                    sigma
                )

            results = evaluate_models(
                X_global
            )

            for architecture in [
                "MLP",
                "RF"
            ]:

                r = results[
                    architecture
                ]

                global_rows.append(
                    {
                        "Architecture":
                            architecture,

                        "Outage_Fraction_Global_AP_IDs":
                            outage_fraction,

                        "RSSI_Sigma":
                            sigma,

                        "Seed":
                            seed,

                        "Selected_Global_APs":
                            len(
                                selected_aps
                            ),

                        "Mean_Effective_Observed_AP_Loss":
                            effective_fraction.mean(),

                        "Median_Effective_Observed_AP_Loss":
                            np.median(
                                effective_fraction
                            ),

                        "Baseline_Accuracy":
                            r[
                                "baseline_accuracy"
                            ],

                        "Robust_Accuracy":
                            r[
                                "robust_accuracy"
                            ],

                        "Accuracy_Gain":
                            (
                                r[
                                    "robust_accuracy"
                                ]
                                -
                                r[
                                    "baseline_accuracy"
                                ]
                            ),

                        "Baseline_F1":
                            r[
                                "baseline_f1"
                            ],

                        "Robust_F1":
                            r[
                                "robust_f1"
                            ],

                        "Baseline_Error":
                            r[
                                "baseline_mean_error"
                            ],

                        "Robust_Error":
                            r[
                                "robust_mean_error"
                            ],

                        "Error_Reduction":
                            (
                                r[
                                    "baseline_mean_error"
                                ]
                                -
                                r[
                                    "robust_mean_error"
                                ]
                            ),

                        "Baseline_Median_Error":
                            r[
                                "baseline_median_error"
                            ],

                        "Robust_Median_Error":
                            r[
                                "robust_median_error"
                            ],

                        "Median_Error_Reduction":
                            (
                                r[
                                    "baseline_median_error"
                                ]
                                -
                                r[
                                    "robust_median_error"
                                ]
                            ),

                        "Baseline_P90_Error":
                            r[
                                "baseline_p90_error"
                            ],

                        "Robust_P90_Error":
                            r[
                                "robust_p90_error"
                            ],

                        "P90_Error_Reduction":
                            (
                                r[
                                    "baseline_p90_error"
                                ]
                                -
                                r[
                                    "robust_p90_error"
                                ]
                            )
                    }
                )


global_df = pd.DataFrame(
    global_rows
)

global_df.to_csv(
    RESULTS_CSV
    / "13_global_ap_outage_detailed.csv",
    index=False
)


# ------------------------------------------------------------
# Step 19 — Structured outage summary
# ------------------------------------------------------------

global_summary = (
    global_df
    .groupby(
        [
            "Architecture",
            "Outage_Fraction_Global_AP_IDs",
            "RSSI_Sigma"
        ],
        as_index=False
    )
    .agg(
        Mean_Effective_Observed_AP_Loss=(
            "Mean_Effective_Observed_AP_Loss",
            "mean"
        ),

        Baseline_Accuracy=(
            "Baseline_Accuracy",
            "mean"
        ),

        Robust_Accuracy=(
            "Robust_Accuracy",
            "mean"
        ),

        Accuracy_Gain=(
            "Accuracy_Gain",
            "mean"
        ),

        Accuracy_Gain_SD=(
            "Accuracy_Gain",
            "std"
        ),

        Baseline_F1=(
            "Baseline_F1",
            "mean"
        ),

        Robust_F1=(
            "Robust_F1",
            "mean"
        ),

        Baseline_Error=(
            "Baseline_Error",
            "mean"
        ),

        Robust_Error=(
            "Robust_Error",
            "mean"
        ),

        Error_Reduction=(
            "Error_Reduction",
            "mean"
        ),

        Error_Reduction_SD=(
            "Error_Reduction",
            "std"
        ),

        Baseline_Median_Error=(
            "Baseline_Median_Error",
            "mean"
        ),

        Robust_Median_Error=(
            "Robust_Median_Error",
            "mean"
        ),

        Median_Error_Reduction=(
            "Median_Error_Reduction",
            "mean"
        ),

        Baseline_P90_Error=(
            "Baseline_P90_Error",
            "mean"
        ),

        Robust_P90_Error=(
            "Robust_P90_Error",
            "mean"
        ),

        P90_Error_Reduction=(
            "P90_Error_Reduction",
            "mean"
        )
    )
)

global_summary.to_csv(
    RESULTS_CSV
    / "13_global_ap_outage_summary.csv",
    index=False
)

print("\nSTRUCTURED / GLOBAL AP-OUTAGE SUMMARY")
display(
    global_summary.round(4)
)


# ------------------------------------------------------------
# Step 20 — Confusion matrices
#
# Use the final severe corruption draw only for visualization.
# It is explicitly illustrative, not a new primary estimate.
# ------------------------------------------------------------

labels = np.unique(
    y_test_cls
)

for architecture in [
    "MLP",
    "RF"
]:

    results = (
        last_severe_results[
            architecture
        ]
    )

    baseline_pred = (
        results[
            "baseline_class_predictions"
        ]
    )

    robust_pred = (
        results[
            "robust_class_predictions"
        ]
    )

    # Baseline.
    cm = confusion_matrix(
        y_test_cls,
        baseline_pred,
        labels=labels,
        normalize="true"
    )

    fig, ax = plt.subplots(
        figsize=(9, 8)
    )

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=labels
    )

    disp.plot(
        ax=ax,
        xticks_rotation=90,
        values_format=".2f",
        colorbar=False
    )

    ax.set_title(
        f"{architecture} Equal-Volume Baseline — Severe Condition"
    )

    plt.tight_layout()

    plt.savefig(
        RESULTS_FIG
        / f"13_{architecture.lower()}_baseline_severe_confusion.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    # Robust.
    cm = confusion_matrix(
        y_test_cls,
        robust_pred,
        labels=labels,
        normalize="true"
    )

    fig, ax = plt.subplots(
        figsize=(9, 8)
    )

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=labels
    )

    disp.plot(
        ax=ax,
        xticks_rotation=90,
        values_format=".2f",
        colorbar=False
    )

    ax.set_title(
        f"{architecture} Robust Model — Severe Condition"
    )

    plt.tight_layout()

    plt.savefig(
        RESULTS_FIG
        / f"13_{architecture.lower()}_robust_severe_confusion.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# Step 21 — Spatial localization-error visualizations
#
# Again, these are illustrative plots using the final
# severe corruption draw.
# ------------------------------------------------------------

for architecture in [
    "MLP",
    "RF"
]:

    results = (
        last_severe_results[
            architecture
        ]
    )

    baseline_errors = (
        results[
            "baseline_errors"
        ]
    )

    robust_errors = (
        results[
            "robust_errors"
        ]
    )

    # Equal-volume baseline.
    plt.figure(
        figsize=(8, 6)
    )

    scatter = plt.scatter(
        y_test_reg[:, 0],
        y_test_reg[:, 1],
        c=baseline_errors,
        s=14
    )

    plt.colorbar(
        scatter,
        label="Localization error (coordinate units)"
    )

    plt.xlabel(
        "LONGITUDE coordinate"
    )

    plt.ylabel(
        "LATITUDE coordinate"
    )

    plt.title(
        f"{architecture} Equal-Volume Baseline Localization Error"
    )

    plt.tight_layout()

    plt.savefig(
        RESULTS_FIG
        / f"13_{architecture.lower()}_baseline_spatial_error.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    # Robust.
    plt.figure(
        figsize=(8, 6)
    )

    scatter = plt.scatter(
        y_test_reg[:, 0],
        y_test_reg[:, 1],
        c=robust_errors,
        s=14
    )

    plt.colorbar(
        scatter,
        label="Localization error (coordinate units)"
    )

    plt.xlabel(
        "LONGITUDE coordinate"
    )

    plt.ylabel(
        "LATITUDE coordinate"
    )

    plt.title(
        f"{architecture} Robust Localization Error"
    )

    plt.tight_layout()

    plt.savefig(
        RESULTS_FIG
        / f"13_{architecture.lower()}_robust_spatial_error.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# Step 22 — Reproducibility metadata
# ------------------------------------------------------------

metadata = {
    "experiment":
        "Reviewer-requested extended robustness analysis",

    "post_hoc":
        True,

    "evaluation_seed_count":
        len(
            EVAL_SEEDS_20
        ),

    "evaluation_seeds":
        EVAL_SEEDS_20,

    "severe_condition": {
        "ap_loss":
            SEVERE_AP_LOSS,

        "sigma_db":
            SEVERE_SIGMA
    },

    "global_ap_outage_levels":
        GLOBAL_OUTAGE_LEVELS,

    "global_outage_noise_levels":
        GLOBAL_NOISE_LEVELS,

    "bootstrap_repetitions":
        BOOTSTRAP_REPS,

    "bootstrap_seed":
        BOOTSTRAP_SEED,

    "bootstrap_procedure":
        (
            "Hierarchical paired bootstrap: resample the 20 corruption "
            "seeds with replacement; within each selected seed resample "
            "the 1111 fingerprints with replacement; compute the paired "
            "mean robust-minus-baseline correctness or baseline-minus-"
            "robust localization error."
        ),

    "model_inputs":
        (
            "Only scaled RSSI features are model inputs. "
            "Observation masks are used only to determine which "
            "AP measurements may be corrupted."
        ),

    "python_version":
        sys.version,

    "numpy_version":
        np.__version__,

    "pandas_version":
        pd.__version__,

    "sklearn_version":
        sklearn.__version__,

    "platform":
        platform.platform(),

    "cpu_count":
        os.cpu_count()
}


with open(
    RESULTS_CSV
    / "13_reproducibility_metadata.json",
    "w"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )


# ------------------------------------------------------------
# Step 23 — Reviewer-ready summary
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("NOTEBOOK 13 — REVIEWER-REQUESTED EXTENDED ANALYSIS")
print("=" * 80)

print("\nA. 20-SEED SEVERE CONDITION")
print(
    severe_summary.round(4).to_string(
        index=False
    )
)

print("\nB. 20-SEED HIERARCHICAL BOOTSTRAP")
print(
    bootstrap_df.round(4).to_string(
        index=False
    )
)

print("\nC. STRUCTURED / GLOBAL AP OUTAGE")
print(
    global_summary.round(4).to_string(
        index=False
    )
)


# ------------------------------------------------------------
# Step 24 — Completion
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("NOTEBOOK 13 COMPLETED SUCCESSFULLY")
print("=" * 80)

print("\nSaved CSV files:")

for path in sorted(
    RESULTS_CSV.glob(
        "13_*"
    )
):

    print(
        "-",
        path.name
    )

print("\nSaved figures:")

for path in sorted(
    RESULTS_FIG.glob(
        "13_*"
    )
):

    print(
        "-",
        path.name
    )

print("\nIMPORTANT:")
print(
    "These analyses were added post hoc in response to reviewer feedback."
)
print(
    "They strengthen the robustness evidence but do not replace the "
    "originally frozen primary evaluation."
)
```

```python
# ============================================================
# Dr.Aans requested:
# Crossed two-way cluster bootstrap
#
# Rows    = corruption seeds
# Columns = the same 1,111 fingerprint identities
#
# This preserves fingerprint identity across corruption seeds.
# ============================================================

import numpy as np
import pandas as pd

N_BOOT = 5000
BOOT_SEED = 23042


def crossed_cluster_bootstrap(
    effect_matrix,
    n_boot=5000,
    random_seed=23042
):
    """
    Two-way crossed bootstrap.

    effect_matrix:
        shape = (n_seeds, n_fingerprints)

    Rows correspond to corruption realizations.
    Columns correspond to fixed fingerprint identities.

    For each replicate:
      - resample seeds with replacement
      - resample fingerprint identities with replacement
      - apply the SAME sampled fingerprint indices across
        all selected seeds
    """

    effects = np.asarray(
        effect_matrix,
        dtype=float
    )

    n_seeds, n_fingerprints = effects.shape

    rng = np.random.default_rng(
        random_seed
    )

    boot = np.empty(
        n_boot,
        dtype=float
    )

    for b in range(n_boot):

        seed_idx = rng.integers(
            0,
            n_seeds,
            size=n_seeds
        )

        fingerprint_idx = rng.integers(
            0,
            n_fingerprints,
            size=n_fingerprints
        )

        sampled = effects[
            np.ix_(
                seed_idx,
                fingerprint_idx
            )
        ]

        boot[b] = sampled.mean()

    estimate = effects.mean()

    lower, upper = np.percentile(
        boot,
        [2.5, 97.5]
    )

    return estimate, lower, upper, boot


crossed_rows = []

for architecture in ["MLP", "RF"]:

    # Classification:
    # robust correctness - repeated-clean correctness
    cls_matrix = np.asarray(
        paired_effects[
            architecture
        ]["classification"]
    )

    cls_est, cls_low, cls_high, _ = (
        crossed_cluster_bootstrap(
            cls_matrix,
            n_boot=N_BOOT,
            random_seed=BOOT_SEED
        )
    )

    # Localization:
    # repeated-clean error - robust error
    loc_matrix = np.asarray(
        paired_effects[
            architecture
        ]["localization"]
    )

    loc_est, loc_low, loc_high, _ = (
        crossed_cluster_bootstrap(
            loc_matrix,
            n_boot=N_BOOT,
            random_seed=BOOT_SEED + 1
        )
    )

    crossed_rows.append(
        {
            "Architecture": architecture,
            "Metric": "Accuracy gain",
            "Estimate": cls_est,
            "CI_Lower": cls_low,
            "CI_Upper": cls_high
        }
    )

    crossed_rows.append(
        {
            "Architecture": architecture,
            "Metric": "Localization error reduction",
            "Estimate": loc_est,
            "CI_Lower": loc_low,
            "CI_Upper": loc_high
        }
    )


crossed_bootstrap_df = pd.DataFrame(
    crossed_rows
)

print("=" * 80)
print("CROSSED TWO-WAY CLUSTER BOOTSTRAP")
print("=" * 80)

print(
    crossed_bootstrap_df
    .round(4)
    .to_string(index=False)
)

crossed_bootstrap_df.to_csv(
    RESULTS_CSV
    / "13_crossed_cluster_bootstrap_summary.csv",
    index=False
)
```

```python
# ============================================================
# SECOND-ROUND REVIEWER :
# Descriptive two-way variance decomposition
# ============================================================

import numpy as np
import pandas as pd


def two_way_variance_decomposition(effect_matrix):
    """
    Descriptive decomposition of variability across:
      1. corruption seeds
      2. fingerprint identities
      3. residual / seed-by-fingerprint interaction

    Rows    = corruption seeds
    Columns = fingerprint identities

    This is a descriptive sensitivity analysis, not a formal
    random-effects population model.
    """

    Y = np.asarray(effect_matrix, dtype=float)

    n_seed, n_fp = Y.shape

    grand_mean = Y.mean()
    seed_means = Y.mean(axis=1)
    fingerprint_means = Y.mean(axis=0)

    ss_seed = n_fp * np.sum(
        (seed_means - grand_mean) ** 2
    )

    ss_fingerprint = n_seed * np.sum(
        (fingerprint_means - grand_mean) ** 2
    )

    residual = (
        Y
        - seed_means[:, None]
        - fingerprint_means[None, :]
        + grand_mean
    )

    ss_residual = np.sum(
        residual ** 2
    )

    ss_total = (
        ss_seed
        + ss_fingerprint
        + ss_residual
    )

    return {
        "Seed_SS": ss_seed,
        "Fingerprint_SS": ss_fingerprint,
        "Residual_SS": ss_residual,
        "Seed_Variance_Fraction": ss_seed / ss_total,
        "Fingerprint_Variance_Fraction": ss_fingerprint / ss_total,
        "Residual_Variance_Fraction": ss_residual / ss_total
    }


variance_rows = []

for architecture in ["MLP", "RF"]:

    cls_matrix = np.asarray(
        paired_effects[
            architecture
        ]["classification"]
    )

    cls_result = two_way_variance_decomposition(
        cls_matrix
    )

    variance_rows.append(
        {
            "Architecture": architecture,
            "Metric": "Classification accuracy effect",
            **cls_result
        }
    )

    loc_matrix = np.asarray(
        paired_effects[
            architecture
        ]["localization"]
    )

    loc_result = two_way_variance_decomposition(
        loc_matrix
    )

    variance_rows.append(
        {
            "Architecture": architecture,
            "Metric": "Localization error reduction",
            **loc_result
        }
    )


variance_df = pd.DataFrame(
    variance_rows
)

variance_df.to_csv(
    RESULTS_CSV / "13_variance_decomposition.csv",
    index=False
)

print("=" * 80)
print("DESCRIPTIVE TWO-WAY VARIANCE DECOMPOSITION")
print("=" * 80)

print(
    variance_df[
        [
            "Architecture",
            "Metric",
            "Seed_Variance_Fraction",
            "Fingerprint_Variance_Fraction",
            "Residual_Variance_Fraction"
        ]
    ]
    .round(4)
    .to_string(index=False)
)
```
