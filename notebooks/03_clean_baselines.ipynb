```python
# ============================================================
# DEN702 Research Paper
# Notebook 03 — Clean Baseline Experiments
#
# Research Topic:
# Robust Wi-Fi Fingerprint-Based Indoor Localization
# Under Access-Point Loss and RSSI Noise
#
# Purpose:
# 1. Load the fixed preprocessed UJIIndoorLoc dataset
# 2. Train clean-condition baseline classifiers:
#       - KNN
#       - Random Forest
#       - MLP
# 3. Evaluate Building-Floor classification performance
# 4. Train clean-condition coordinate regressors:
#       - KNN
#       - Random Forest
#       - MLP
# 5. Evaluate coordinate-localization performance
# 6. Save metrics, figures, predictions, and trained models
#
# IMPORTANT:
# The development set is used for baseline model comparison.
# The official UJIIndoorLoc validation dataset remains reserved
# as the final independent test set.
# ============================================================


# ------------------------------------------------------------
# Step 1 — Import Required Libraries
# ------------------------------------------------------------

from pathlib import Path
import time
import warnings
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from IPython.display import display

from sklearn.neighbors import (
    KNeighborsClassifier,
    KNeighborsRegressor
)

from sklearn.ensemble import (
    RandomForestClassifier,
    RandomForestRegressor
)

from sklearn.neural_network import (
    MLPClassifier,
    MLPRegressor
)

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    mean_absolute_error,
    mean_squared_error
)


warnings.filterwarnings("ignore")

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


print("=" * 72)
print("DEN702 — CLEAN BASELINE EXPERIMENTS")
print("=" * 72)
print("Random state:", RANDOM_STATE)


# ------------------------------------------------------------
# Step 2 — Detect Project Root and Define Paths
# ------------------------------------------------------------

if Path("../data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path("..").resolve()

elif Path("data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path(".").resolve()

else:

    raise FileNotFoundError(
        "Could not locate data/processed/uji_preprocessed.npz.\n"
        "Run Notebook 02 before running Notebook 03."
    )


PROCESSED_DATA_DIR = (
    PROJECT_ROOT / "data" / "processed"
)

RESULTS_CSV_DIR = (
    PROJECT_ROOT / "results" / "csv"
)

RESULTS_FIGURES_DIR = (
    PROJECT_ROOT / "results" / "figures"
)

MODEL_DIR = (
    PROJECT_ROOT / "models" / "03_clean_baselines"
)


RESULTS_CSV_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)

MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True
)


DATA_FILE = (
    PROCESSED_DATA_DIR
    / "uji_preprocessed.npz"
)


print("\nProject root:")
print(PROJECT_ROOT)

print("\nModel directory:")
print(MODEL_DIR)


# ------------------------------------------------------------
# Step 3 — Load the Fixed Experimental Dataset
#
# These are exactly the train/dev/test splits generated in
# Notebook 02. They must remain unchanged for all experiments.
# ------------------------------------------------------------

data = np.load(
    DATA_FILE,
    allow_pickle=True
)


X_train = data["X_train"]
X_dev = data["X_dev"]
X_test = data["X_test"]

y_class_train = data["y_class_train"]
y_class_dev = data["y_class_dev"]
y_class_test = data["y_class_test"]

y_coord_train = data["y_coord_train"]
y_coord_dev = data["y_coord_dev"]
y_coord_test = data["y_coord_test"]


print("\n" + "=" * 72)
print("STEP 3 — DATASET LOADED")
print("=" * 72)

print("Training shape:", X_train.shape)
print("Development shape:", X_dev.shape)
print("Final test shape:", X_test.shape)

print(
    "Classification classes:",
    len(np.unique(y_class_train))
)

print(
    "Coordinate target shape:",
    y_coord_train.shape
)


# ------------------------------------------------------------
# Step 4 — Verify Experimental Protocol
#
# Models are trained on training data.
# Models are compared on development data.
# Final test data remains untouched.
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("STEP 4 — EXPERIMENTAL PROTOCOL")
print("=" * 72)

print(
    "Training set:",
    len(X_train),
    "samples"
)

print(
    "Development set:",
    len(X_dev),
    "samples"
)

print(
    "Reserved final test set:",
    len(X_test),
    "samples"
)

print(
    "\nNotebook 03 does NOT use the final test set "
    "for model selection."
)


# ------------------------------------------------------------
# Step 5 — Helper Function for Clean Result Display
#
# Pandas 3.x does not allow Series.round() when the Series
# contains both strings and numbers.
#
# Results are therefore displayed as one-row DataFrames.
# ------------------------------------------------------------

def display_result(result_dictionary):

    result_df = pd.DataFrame(
        [result_dictionary]
    )

    display(
        result_df.round(4)
    )


# ============================================================
# PART A — BUILDING-FLOOR CLASSIFICATION
# ============================================================


# ------------------------------------------------------------
# Step 6 — Define Classification Evaluation Function
#
# Metrics:
# - Accuracy
# - Macro Precision
# - Macro Recall
# - Macro F1
#
# Macro averaging gives equal importance to each of the
# 13 Building-Floor classes.
# ------------------------------------------------------------

def evaluate_classifier(
    model_name,
    model,
    X_eval,
    y_eval
):

    start_time = time.time()

    predictions = model.predict(
        X_eval
    )

    inference_time = (
        time.time()
        - start_time
    )

    result = {

        "Model":
            model_name,

        "Accuracy":
            accuracy_score(
                y_eval,
                predictions
            ),

        "Precision_Macro":
            precision_score(
                y_eval,
                predictions,
                average="macro",
                zero_division=0
            ),

        "Recall_Macro":
            recall_score(
                y_eval,
                predictions,
                average="macro",
                zero_division=0
            ),

        "F1_Macro":
            f1_score(
                y_eval,
                predictions,
                average="macro",
                zero_division=0
            ),

        "Inference_Time_s":
            inference_time
    }

    return result, predictions


# ------------------------------------------------------------
# Step 7 — Train KNN Classifier
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("STEP 7 — KNN CLASSIFIER")
print("=" * 72)


knn_clf = KNeighborsClassifier(
    n_neighbors=5,
    weights="distance",
    n_jobs=-1
)


start_time = time.time()

knn_clf.fit(
    X_train,
    y_class_train
)

knn_training_time = (
    time.time()
    - start_time
)


(
    knn_class_results,
    knn_class_pred
) = evaluate_classifier(

    "KNN",
    knn_clf,
    X_dev,
    y_class_dev
)


knn_class_results[
    "Training_Time_s"
] = knn_training_time


display_result(
    knn_class_results
)


# ------------------------------------------------------------
# Step 8 — Train Random Forest Classifier
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("STEP 8 — RANDOM FOREST CLASSIFIER")
print("=" * 72)


rf_clf = RandomForestClassifier(
    n_estimators=300,
    random_state=RANDOM_STATE,
    n_jobs=-1
)


start_time = time.time()

rf_clf.fit(
    X_train,
    y_class_train
)

rf_training_time = (
    time.time()
    - start_time
)


(
    rf_class_results,
    rf_class_pred
) = evaluate_classifier(

    "Random Forest",
    rf_clf,
    X_dev,
    y_class_dev
)


rf_class_results[
    "Training_Time_s"
] = rf_training_time


display_result(
    rf_class_results
)


# ------------------------------------------------------------
# Step 9 — Train MLP Classifier
#
# Early stopping is enabled to reduce overfitting.
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("STEP 9 — MLP CLASSIFIER")
print("=" * 72)


mlp_clf = MLPClassifier(

    hidden_layer_sizes=(
        128,
        64
    ),

    activation="relu",

    solver="adam",

    max_iter=300,

    early_stopping=True,

    validation_fraction=0.10,

    n_iter_no_change=15,

    random_state=RANDOM_STATE
)


start_time = time.time()

mlp_clf.fit(
    X_train,
    y_class_train
)

mlp_training_time = (
    time.time()
    - start_time
)


(
    mlp_class_results,
    mlp_class_pred
) = evaluate_classifier(

    "MLP",
    mlp_clf,
    X_dev,
    y_class_dev
)


mlp_class_results[
    "Training_Time_s"
] = mlp_training_time


display_result(
    mlp_class_results
)


print(
    "MLP iterations:",
    mlp_clf.n_iter_
)


# ------------------------------------------------------------
# Step 10 — Create Classification Results Table
# ------------------------------------------------------------

classification_results = pd.DataFrame(
    [
        knn_class_results,
        rf_class_results,
        mlp_class_results
    ]
)


classification_results = classification_results[
    [
        "Model",
        "Accuracy",
        "Precision_Macro",
        "Recall_Macro",
        "F1_Macro",
        "Training_Time_s",
        "Inference_Time_s"
    ]
]


print("\n" + "=" * 72)
print("STEP 10 — CLEAN CLASSIFICATION RESULTS")
print("=" * 72)


display(
    classification_results.round(4)
)


CLASS_RESULTS_FILE = (
    RESULTS_CSV_DIR
    / "03_clean_classification_baselines.csv"
)


classification_results.to_csv(
    CLASS_RESULTS_FILE,
    index=False
)


print(
    "\nClassification results saved:"
)

print(
    CLASS_RESULTS_FILE
)


# ------------------------------------------------------------
# Step 11 — Identify Best Clean Classifier
#
# Macro F1 is used as the primary model-selection metric.
# ------------------------------------------------------------

best_classifier_row = (
    classification_results.loc[
        classification_results[
            "F1_Macro"
        ].idxmax()
    ]
)


best_classifier_name = (
    best_classifier_row[
        "Model"
    ]
)


print("\n" + "=" * 72)
print("STEP 11 — BEST CLEAN CLASSIFIER")
print("=" * 72)


print(
    "Model:",
    best_classifier_name
)

print(
    "Accuracy:",
    round(
        float(
            best_classifier_row[
                "Accuracy"
            ]
        ),
        4
    )
)

print(
    "Macro F1:",
    round(
        float(
            best_classifier_row[
                "F1_Macro"
            ]
        ),
        4
    )
)


# ------------------------------------------------------------
# Step 12 — Plot Classification Baselines
# ------------------------------------------------------------

classification_plot_data = (
    classification_results
    .set_index("Model")[
        [
            "Accuracy",
            "F1_Macro"
        ]
    ]
)


classification_plot_data.plot(
    kind="bar",
    figsize=(9, 5)
)


plt.ylabel(
    "Score"
)

plt.xlabel(
    "Model"
)

plt.title(
    "Clean-Condition Building-Floor Classification"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y",
    alpha=0.2
)

plt.tight_layout()


CLASS_FIGURE_FILE = (
    RESULTS_FIGURES_DIR
    / "03_clean_classification_baselines.png"
)


plt.savefig(
    CLASS_FIGURE_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 13 — Confusion Matrix for Best Classifier
# ------------------------------------------------------------

classification_prediction_map = {

    "KNN":
        knn_class_pred,

    "Random Forest":
        rf_class_pred,

    "MLP":
        mlp_class_pred
}


best_class_predictions = (
    classification_prediction_map[
        best_classifier_name
    ]
)


class_labels = sorted(
    np.unique(
        y_class_dev
    )
)


cm = confusion_matrix(
    y_class_dev,
    best_class_predictions,
    labels=class_labels
)


fig, ax = plt.subplots(
    figsize=(10, 8)
)


disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_labels
)


disp.plot(
    ax=ax,
    cmap="Blues",
    xticks_rotation=45,
    colorbar=False
)


plt.title(
    "Development Confusion Matrix — "
    f"{best_classifier_name}"
)

plt.tight_layout()


CONFUSION_FILE = (
    RESULTS_FIGURES_DIR
    / "03_best_classifier_confusion_matrix.png"
)


plt.savefig(
    CONFUSION_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PART B — COORDINATE LOCALIZATION / REGRESSION
# ============================================================


# ------------------------------------------------------------
# Step 14 — Define Regression Evaluation Function
#
# Metrics:
# - Coordinate MAE
# - Coordinate RMSE
# - Mean Euclidean localization error
# - Median Euclidean localization error
# - 90th percentile localization error
#
# Distances are currently reported in the coordinate units of
# the UJIIndoorLoc dataset.
# ------------------------------------------------------------

def evaluate_regressor(
    model_name,
    model,
    X_eval,
    y_eval,
    prediction_transform=None
):

    start_time = time.time()

    predictions = model.predict(
        X_eval
    )

    inference_time = (
        time.time()
        - start_time
    )


    if prediction_transform is not None:

        predictions = (
            prediction_transform(
                predictions
            )
        )


    coordinate_mae = (
        mean_absolute_error(
            y_eval,
            predictions
        )
    )


    coordinate_rmse = np.sqrt(
        mean_squared_error(
            y_eval,
            predictions
        )
    )


    euclidean_errors = np.sqrt(
        np.sum(
            (
                predictions
                - y_eval
            ) ** 2,
            axis=1
        )
    )


    result = {

        "Model":
            model_name,

        "Coordinate_MAE":
            coordinate_mae,

        "Coordinate_RMSE":
            coordinate_rmse,

        "Mean_Localization_Error":
            np.mean(
                euclidean_errors
            ),

        "Median_Localization_Error":
            np.median(
                euclidean_errors
            ),

        "P90_Localization_Error":
            np.percentile(
                euclidean_errors,
                90
            ),

        "Inference_Time_s":
            inference_time
    }


    return (
        result,
        predictions,
        euclidean_errors
    )


# ------------------------------------------------------------
# Step 15 — Train KNN Coordinate Regressor
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("STEP 15 — KNN REGRESSOR")
print("=" * 72)


knn_reg = KNeighborsRegressor(
    n_neighbors=5,
    weights="distance",
    n_jobs=-1
)


start_time = time.time()

knn_reg.fit(
    X_train,
    y_coord_train
)

knn_reg_training_time = (
    time.time()
    - start_time
)


(
    knn_reg_results,
    knn_reg_pred,
    knn_errors
) = evaluate_regressor(

    "KNN",
    knn_reg,
    X_dev,
    y_coord_dev
)


knn_reg_results[
    "Training_Time_s"
] = knn_reg_training_time


display_result(
    knn_reg_results
)


# ------------------------------------------------------------
# Step 16 — Train Random Forest Coordinate Regressor
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("STEP 16 — RANDOM FOREST REGRESSOR")
print("=" * 72)


rf_reg = RandomForestRegressor(
    n_estimators=300,
    random_state=RANDOM_STATE,
    n_jobs=-1
)


start_time = time.time()

rf_reg.fit(
    X_train,
    y_coord_train
)

rf_reg_training_time = (
    time.time()
    - start_time
)


(
    rf_reg_results,
    rf_reg_pred,
    rf_errors
) = evaluate_regressor(

    "Random Forest",
    rf_reg,
    X_dev,
    y_coord_dev
)


rf_reg_results[
    "Training_Time_s"
] = rf_reg_training_time


display_result(
    rf_reg_results
)


# ------------------------------------------------------------
# Step 17 — Standardize Coordinate Targets for MLP
#
# Neural-network regression benefits from standardized target
# coordinates.
#
# The target scaler is fitted ONLY on training targets.
# ------------------------------------------------------------

coord_scaler = StandardScaler()


y_coord_train_scaled = (
    coord_scaler.fit_transform(
        y_coord_train
    )
)


print("\n" + "=" * 72)
print("STEP 17 — MLP TARGET NORMALIZATION")
print("=" * 72)


print(
    "Scaled target mean:",
    np.round(
        y_coord_train_scaled.mean(
            axis=0
        ),
        4
    )
)


print(
    "Scaled target standard deviation:",
    np.round(
        y_coord_train_scaled.std(
            axis=0
        ),
        4
    )
)


# ------------------------------------------------------------
# Step 18 — Train MLP Coordinate Regressor
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("STEP 18 — MLP REGRESSOR")
print("=" * 72)


mlp_reg = MLPRegressor(

    hidden_layer_sizes=(
        128,
        64
    ),

    activation="relu",

    solver="adam",

    max_iter=300,

    early_stopping=True,

    validation_fraction=0.10,

    n_iter_no_change=15,

    random_state=RANDOM_STATE
)


start_time = time.time()

mlp_reg.fit(
    X_train,
    y_coord_train_scaled
)

mlp_reg_training_time = (
    time.time()
    - start_time
)


(
    mlp_reg_results,
    mlp_reg_pred,
    mlp_errors
) = evaluate_regressor(

    "MLP",
    mlp_reg,
    X_dev,
    y_coord_dev,

    prediction_transform=(
        coord_scaler.inverse_transform
    )
)


mlp_reg_results[
    "Training_Time_s"
] = mlp_reg_training_time


display_result(
    mlp_reg_results
)


print(
    "MLP regression iterations:",
    mlp_reg.n_iter_
)


# ------------------------------------------------------------
# Step 19 — Create Regression Results Table
# ------------------------------------------------------------

regression_results = pd.DataFrame(
    [
        knn_reg_results,
        rf_reg_results,
        mlp_reg_results
    ]
)


regression_results = regression_results[
    [
        "Model",
        "Coordinate_MAE",
        "Coordinate_RMSE",
        "Mean_Localization_Error",
        "Median_Localization_Error",
        "P90_Localization_Error",
        "Training_Time_s",
        "Inference_Time_s"
    ]
]


print("\n" + "=" * 72)
print("STEP 19 — CLEAN REGRESSION RESULTS")
print("=" * 72)


display(
    regression_results.round(4)
)


REG_RESULTS_FILE = (
    RESULTS_CSV_DIR
    / "03_clean_regression_baselines.csv"
)


regression_results.to_csv(
    REG_RESULTS_FILE,
    index=False
)


print(
    "\nRegression results saved:"
)

print(
    REG_RESULTS_FILE
)


# ------------------------------------------------------------
# Step 20 — Identify Best Coordinate Regressor
#
# Mean Euclidean localization error is used as the primary
# comparison metric.
# ------------------------------------------------------------

best_regressor_row = (
    regression_results.loc[
        regression_results[
            "Mean_Localization_Error"
        ].idxmin()
    ]
)


best_regressor_name = (
    best_regressor_row[
        "Model"
    ]
)


print("\n" + "=" * 72)
print("STEP 20 — BEST CLEAN COORDINATE REGRESSOR")
print("=" * 72)


print(
    "Model:",
    best_regressor_name
)

print(
    "Mean localization error:",
    round(
        float(
            best_regressor_row[
                "Mean_Localization_Error"
            ]
        ),
        4
    )
)

print(
    "Median localization error:",
    round(
        float(
            best_regressor_row[
                "Median_Localization_Error"
            ]
        ),
        4
    )
)

print(
    "P90 localization error:",
    round(
        float(
            best_regressor_row[
                "P90_Localization_Error"
            ]
        ),
        4
    )
)


# ------------------------------------------------------------
# Step 21 — Plot Coordinate Localization Baselines
# ------------------------------------------------------------

regression_plot_data = (
    regression_results
    .set_index("Model")[
        [
            "Mean_Localization_Error",
            "Median_Localization_Error"
        ]
    ]
)


regression_plot_data.plot(
    kind="bar",
    figsize=(9, 5)
)


plt.ylabel(
    "Localization Error "
    "(dataset coordinate units)"
)

plt.xlabel(
    "Model"
)

plt.title(
    "Clean-Condition Coordinate Localization Error"
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y",
    alpha=0.2
)

plt.tight_layout()


REG_FIGURE_FILE = (
    RESULTS_FIGURES_DIR
    / "03_clean_regression_baselines.png"
)


plt.savefig(
    REG_FIGURE_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 22 — Create Localization Error CDF
#
# CDF = Cumulative Distribution Function.
#
# This shows the percentage of development samples whose
# localization error falls below a given threshold.
# ------------------------------------------------------------

regression_error_map = {

    "KNN":
        knn_errors,

    "Random Forest":
        rf_errors,

    "MLP":
        mlp_errors
}


best_errors = (
    regression_error_map[
        best_regressor_name
    ]
)


sorted_errors = np.sort(
    best_errors
)


cdf = (
    np.arange(
        1,
        len(sorted_errors) + 1
    )
    / len(sorted_errors)
)


plt.figure(
    figsize=(8, 5)
)


plt.plot(
    sorted_errors,
    cdf
)


plt.xlabel(
    "Localization Error "
    "(dataset coordinate units)"
)

plt.ylabel(
    "Cumulative Probability"
)

plt.title(
    "Localization Error CDF — "
    f"{best_regressor_name}"
)

plt.grid(
    alpha=0.2
)

plt.tight_layout()


CDF_FILE = (
    RESULTS_FIGURES_DIR
    / "03_best_regressor_error_cdf.png"
)


plt.savefig(
    CDF_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 23 — Save Development Predictions
#
# Saving predictions allows later statistical analysis without
# needing to rerun Notebook 03.
# ------------------------------------------------------------

PREDICTIONS_FILE = (
    RESULTS_CSV_DIR
    / "03_development_predictions.npz"
)


np.savez_compressed(

    PREDICTIONS_FILE,

    y_class_true=y_class_dev,

    knn_class_pred=knn_class_pred,
    rf_class_pred=rf_class_pred,
    mlp_class_pred=mlp_class_pred,

    y_coord_true=y_coord_dev,

    knn_coord_pred=knn_reg_pred,
    rf_coord_pred=rf_reg_pred,
    mlp_coord_pred=mlp_reg_pred,

    knn_errors=knn_errors,
    rf_errors=rf_errors,
    mlp_errors=mlp_errors
)


print(
    "\nDevelopment predictions saved:"
)

print(
    PREDICTIONS_FILE
)


# ------------------------------------------------------------
# Step 24 — Save Trained Classification Models
# ------------------------------------------------------------

KNN_CLF_FILE = (
    MODEL_DIR
    / "knn_classifier.joblib"
)

RF_CLF_FILE = (
    MODEL_DIR
    / "random_forest_classifier.joblib"
)

MLP_CLF_FILE = (
    MODEL_DIR
    / "mlp_classifier.joblib"
)


joblib.dump(
    knn_clf,
    KNN_CLF_FILE
)

joblib.dump(
    rf_clf,
    RF_CLF_FILE
)

joblib.dump(
    mlp_clf,
    MLP_CLF_FILE
)


# ------------------------------------------------------------
# Step 25 — Save Trained Regression Models
# ------------------------------------------------------------

KNN_REG_FILE = (
    MODEL_DIR
    / "knn_regressor.joblib"
)

RF_REG_FILE = (
    MODEL_DIR
    / "random_forest_regressor.joblib"
)

MLP_REG_FILE = (
    MODEL_DIR
    / "mlp_regressor.joblib"
)

COORD_SCALER_FILE = (
    MODEL_DIR
    / "coordinate_scaler.joblib"
)


joblib.dump(
    knn_reg,
    KNN_REG_FILE
)

joblib.dump(
    rf_reg,
    RF_REG_FILE
)

joblib.dump(
    mlp_reg,
    MLP_REG_FILE
)

joblib.dump(
    coord_scaler,
    COORD_SCALER_FILE
)


print("\nBaseline models saved successfully.")


# ------------------------------------------------------------
# Step 26 — Create Combined Baseline Summary
# ------------------------------------------------------------

baseline_summary = pd.DataFrame(
    [
        {

            "Task":
                "Building-Floor Classification",

            "Best_Model":
                best_classifier_name,

            "Primary_Metric":
                "Macro F1",

            "Primary_Value":
                float(
                    best_classifier_row[
                        "F1_Macro"
                    ]
                ),

            "Secondary_Metric":
                "Accuracy",

            "Secondary_Value":
                float(
                    best_classifier_row[
                        "Accuracy"
                    ]
                )
        },

        {

            "Task":
                "Coordinate Localization",

            "Best_Model":
                best_regressor_name,

            "Primary_Metric":
                "Mean Localization Error",

            "Primary_Value":
                float(
                    best_regressor_row[
                        "Mean_Localization_Error"
                    ]
                ),

            "Secondary_Metric":
                "Median Localization Error",

            "Secondary_Value":
                float(
                    best_regressor_row[
                        "Median_Localization_Error"
                    ]
                )
        }
    ]
)


SUMMARY_FILE = (
    RESULTS_CSV_DIR
    / "03_clean_baseline_summary.csv"
)


baseline_summary.to_csv(
    SUMMARY_FILE,
    index=False
)


print("\n" + "=" * 72)
print("STEP 26 — CLEAN BASELINE SUMMARY TABLE")
print("=" * 72)


display(
    baseline_summary.round(4)
)


# ------------------------------------------------------------
# Step 27 — Final Verification
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("NOTEBOOK 03 COMPLETED SUCCESSFULLY")
print("=" * 72)


print("\nBEST CLEAN CLASSIFIER")

print(
    "Model:",
    best_classifier_name
)

print(
    "Accuracy:",
    round(
        float(
            best_classifier_row[
                "Accuracy"
            ]
        ),
        4
    )
)

print(
    "Macro Precision:",
    round(
        float(
            best_classifier_row[
                "Precision_Macro"
            ]
        ),
        4
    )
)

print(
    "Macro Recall:",
    round(
        float(
            best_classifier_row[
                "Recall_Macro"
            ]
        ),
        4
    )
)

print(
    "Macro F1:",
    round(
        float(
            best_classifier_row[
                "F1_Macro"
            ]
        ),
        4
    )
)


print("\nBEST CLEAN COORDINATE REGRESSOR")

print(
    "Model:",
    best_regressor_name
)

print(
    "Coordinate MAE:",
    round(
        float(
            best_regressor_row[
                "Coordinate_MAE"
            ]
        ),
        4
    )
)

print(
    "Coordinate RMSE:",
    round(
        float(
            best_regressor_row[
                "Coordinate_RMSE"
            ]
        ),
        4
    )
)

print(
    "Mean localization error:",
    round(
        float(
            best_regressor_row[
                "Mean_Localization_Error"
            ]
        ),
        4
    )
)

print(
    "Median localization error:",
    round(
        float(
            best_regressor_row[
                "Median_Localization_Error"
            ]
        ),
        4
    )
)

print(
    "P90 localization error:",
    round(
        float(
            best_regressor_row[
                "P90_Localization_Error"
            ]
        ),
        4
    )
)


print("\nSaved CSV result files:")

print(
    "1.",
    CLASS_RESULTS_FILE
)

print(
    "2.",
    REG_RESULTS_FILE
)

print(
    "3.",
    SUMMARY_FILE
)

print(
    "4.",
    PREDICTIONS_FILE
)


print("\nSaved figures:")

print(
    "1.",
    CLASS_FIGURE_FILE
)

print(
    "2.",
    CONFUSION_FILE
)

print(
    "3.",
    REG_FIGURE_FILE
)

print(
    "4.",
    CDF_FILE
)


print("\nSaved baseline models:")

print(
    "1.",
    KNN_CLF_FILE
)

print(
    "2.",
    RF_CLF_FILE
)

print(
    "3.",
    MLP_CLF_FILE
)

print(
    "4.",
    KNN_REG_FILE
)

print(
    "5.",
    RF_REG_FILE
)

print(
    "6.",
    MLP_REG_FILE
)

print(
    "7.",
    COORD_SCALER_FILE
)


print("\nNotebook 03 is complete.")

print(
    "Next notebook: "
    "04_ap_loss_experiments.ipynb"
)
```
