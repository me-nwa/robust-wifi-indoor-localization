```python
# ============================================================
# DEN702 Research Paper
# Notebook 04 — Access-Point Loss Robustness Experiments
#
# Research Topic:
# Robust Wi-Fi Fingerprint-Based Indoor Localization
# Under Access-Point Loss and RSSI Noise
#
# Research Question:
# How does increasing Wi-Fi access-point loss affect
# indoor localization performance?
#
# Purpose:
# 1. Load clean baseline models from Notebook 03
# 2. Load RAW development RSSI fingerprints
# 3. Simulate controlled loss of DETECTED APs only
# 4. Evaluate classification robustness
# 5. Evaluate coordinate-localization robustness
# 6. Repeat stochastic corruptions using multiple random seeds
# 7. Report mean ± standard deviation
# 8. Save detailed and summarized experimental results
# 9. Generate robustness degradation figures
#
# IMPORTANT:
# - Models are NOT retrained in this notebook.
# - Only the development dataset is corrupted.
# - The official final test set remains untouched.
# - AP dropout is applied only to APs that were originally
#   detected in each fingerprint.
# ============================================================


# ------------------------------------------------------------
# Step 1 — Import Required Libraries
# ------------------------------------------------------------

from pathlib import Path
import random
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from IPython.display import display

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    mean_absolute_error,
    mean_squared_error
)


# ------------------------------------------------------------
# Step 2 — Reproducibility Configuration
# ------------------------------------------------------------

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


# AP-loss conditions to evaluate

AP_LOSS_LEVELS = [
    0.00,
    0.10,
    0.20,
    0.30,
    0.40,
    0.50
]


# Five independent corruption runs for each non-zero level

EXPERIMENT_SEEDS = [
    42,
    43,
    44,
    45,
    46
]


MISSING_RSSI = -110


print("=" * 76)
print("DEN702 — ACCESS-POINT LOSS ROBUSTNESS EXPERIMENT")
print("=" * 76)

print("AP-loss levels:", AP_LOSS_LEVELS)
print("Experiment seeds:", EXPERIMENT_SEEDS)
print("Missing RSSI representation:", MISSING_RSSI, "dBm")


# ------------------------------------------------------------
# Step 3 — Detect Project Root
# ------------------------------------------------------------

if Path("../data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path("..").resolve()

elif Path("data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path(".").resolve()

else:

    raise FileNotFoundError(
        "Could not locate uji_preprocessed.npz.\n"
        "Run Notebook 02 first."
    )


PROCESSED_DATA_DIR = (
    PROJECT_ROOT / "data" / "processed"
)

MODEL_DIR = (
    PROJECT_ROOT / "models" / "03_clean_baselines"
)

RESULTS_CSV_DIR = (
    PROJECT_ROOT / "results" / "csv"
)

RESULTS_FIGURES_DIR = (
    PROJECT_ROOT / "results" / "figures"
)


RESULTS_CSV_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("\nProject root:")
print(PROJECT_ROOT)


# ------------------------------------------------------------
# Step 4 — Load Preprocessed Dataset
#
# We specifically need:
#
# X_dev_raw
#     Raw development RSSI values in dBm
#
# observed_mask_dev
#     True where AP was originally detected
#
# y_class_dev
#     Building-floor labels
#
# y_coord_dev
#     Coordinate targets
# ------------------------------------------------------------

DATA_FILE = (
    PROCESSED_DATA_DIR
    / "uji_preprocessed.npz"
)


data = np.load(
    DATA_FILE,
    allow_pickle=True
)


X_dev_clean_scaled = (
    data["X_dev"]
)

X_dev_raw = (
    data["X_dev_raw"]
)

observed_mask_dev = (
    data["observed_mask_dev"]
)

y_class_dev = (
    data["y_class_dev"]
)

y_coord_dev = (
    data["y_coord_dev"]
)


print("\n" + "=" * 76)
print("STEP 4 — DEVELOPMENT DATA LOADED")
print("=" * 76)

print(
    "Raw development shape:",
    X_dev_raw.shape
)

print(
    "Observed AP mask shape:",
    observed_mask_dev.shape
)

print(
    "Classification target:",
    y_class_dev.shape
)

print(
    "Coordinate target:",
    y_coord_dev.shape
)


# ------------------------------------------------------------
# Step 5 — Verify Observed-AP Statistics
# ------------------------------------------------------------

detected_aps_per_sample = (
    observed_mask_dev.sum(axis=1)
)


print("\n" + "=" * 76)
print("STEP 5 — DEVELOPMENT AP STATISTICS")
print("=" * 76)


print(
    "Mean detected APs:",
    round(
        float(
            detected_aps_per_sample.mean()
        ),
        2
    )
)

print(
    "Median detected APs:",
    round(
        float(
            np.median(
                detected_aps_per_sample
            )
        ),
        2
    )
)

print(
    "Minimum detected APs:",
    int(
        detected_aps_per_sample.min()
    )
)

print(
    "Maximum detected APs:",
    int(
        detected_aps_per_sample.max()
    )
)


# ------------------------------------------------------------
# Step 6 — Load RSSI MinMax Scaler
#
# The same scaler fitted in Notebook 02 must be reused.
#
# Never fit another scaler on corrupted data.
# ------------------------------------------------------------

RSSI_SCALER_FILE = (
    PROCESSED_DATA_DIR
    / "uji_minmax_scaler.joblib"
)


rssi_scaler = joblib.load(
    RSSI_SCALER_FILE
)


print("\nRSSI scaler loaded:")
print(RSSI_SCALER_FILE)


# ------------------------------------------------------------
# Step 7 — Load Clean Classification Models
# ------------------------------------------------------------

knn_clf = joblib.load(
    MODEL_DIR
    / "knn_classifier.joblib"
)

rf_clf = joblib.load(
    MODEL_DIR
    / "random_forest_classifier.joblib"
)

mlp_clf = joblib.load(
    MODEL_DIR
    / "mlp_classifier.joblib"
)


classification_models = {
    "KNN": knn_clf,
    "Random Forest": rf_clf,
    "MLP": mlp_clf
}


print(
    "\nClassification models loaded:",
    list(
        classification_models.keys()
    )
)


# ------------------------------------------------------------
# Step 8 — Load Clean Regression Models
# ------------------------------------------------------------

knn_reg = joblib.load(
    MODEL_DIR
    / "knn_regressor.joblib"
)

rf_reg = joblib.load(
    MODEL_DIR
    / "random_forest_regressor.joblib"
)

mlp_reg = joblib.load(
    MODEL_DIR
    / "mlp_regressor.joblib"
)


coord_scaler = joblib.load(
    MODEL_DIR
    / "coordinate_scaler.joblib"
)


regression_models = {
    "KNN": knn_reg,
    "Random Forest": rf_reg,
    "MLP": mlp_reg
}


print(
    "Regression models loaded:",
    list(
        regression_models.keys()
    )
)


# ------------------------------------------------------------
# Step 9 — Define AP-Loss Simulation Function
#
# Critical methodological point:
#
# We remove a percentage of CURRENTLY DETECTED APs in each
# fingerprint.
#
# Example:
#
# If a fingerprint contains 20 detected APs and AP loss = 0.20,
# approximately 4 detected APs are replaced with -110 dBm.
#
# APs that were already missing remain missing.
# ------------------------------------------------------------

def apply_ap_loss(
    X_raw,
    observed_mask,
    loss_fraction,
    seed
):

    if not 0 <= loss_fraction <= 1:

        raise ValueError(
            "loss_fraction must be between 0 and 1."
        )


    # No corruption for clean condition

    if loss_fraction == 0:

        return X_raw.copy()


    rng = np.random.default_rng(
        seed
    )


    X_corrupted = (
        X_raw.copy()
    )


    for row_index in range(
        X_corrupted.shape[0]
    ):

        detected_indices = np.flatnonzero(
            observed_mask[
                row_index
            ]
        )


        n_detected = (
            len(
                detected_indices
            )
        )


        if n_detected == 0:

            continue


        n_to_drop = int(
            np.floor(
                n_detected
                * loss_fraction
            )
        )


        # For non-zero corruption, drop at least one AP
        # whenever the fingerprint contains a detected AP.

        if (
            loss_fraction > 0
            and n_to_drop == 0
        ):

            n_to_drop = 1


        n_to_drop = min(
            n_to_drop,
            n_detected
        )


        drop_indices = rng.choice(
            detected_indices,
            size=n_to_drop,
            replace=False
        )


        X_corrupted[
            row_index,
            drop_indices
        ] = MISSING_RSSI


    return X_corrupted


# ------------------------------------------------------------
# Step 10 — Validate AP-Loss Function
#
# Check one fingerprint to confirm that the number of detected
# APs decreases correctly.
# ------------------------------------------------------------

example_index = 0

example_original_count = (
    observed_mask_dev[
        example_index
    ].sum()
)


example_corrupted = apply_ap_loss(
    X_dev_raw[
        example_index:
        example_index + 1
    ],
    observed_mask_dev[
        example_index:
        example_index + 1
    ],
    loss_fraction=0.50,
    seed=RANDOM_STATE
)


example_corrupted_count = (
    example_corrupted[
        0
    ] != MISSING_RSSI
).sum()


print("\n" + "=" * 76)
print("STEP 10 — AP-LOSS FUNCTION CHECK")
print("=" * 76)


print(
    "Original detected APs:",
    int(
        example_original_count
    )
)

print(
    "Detected after 50% loss:",
    int(
        example_corrupted_count
    )
)


# ------------------------------------------------------------
# Step 11 — Classification Evaluation Function
# ------------------------------------------------------------

def evaluate_classification(
    model,
    X,
    y_true
):

    predictions = model.predict(
        X
    )


    return {

        "Accuracy":
            accuracy_score(
                y_true,
                predictions
            ),

        "Precision_Macro":
            precision_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            ),

        "Recall_Macro":
            recall_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            ),

        "F1_Macro":
            f1_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            )
    }


# ------------------------------------------------------------
# Step 12 — Coordinate Evaluation Function
#
# The MLP regression model predicts standardized targets.
#
# Its output therefore must be converted back through the
# coordinate scaler before evaluation.
# ------------------------------------------------------------

def evaluate_regression(
    model_name,
    model,
    X,
    y_true
):

    predictions = model.predict(
        X
    )


    if model_name == "MLP":

        predictions = (
            coord_scaler.inverse_transform(
                predictions
            )
        )


    coordinate_mae = (
        mean_absolute_error(
            y_true,
            predictions
        )
    )


    coordinate_rmse = np.sqrt(
        mean_squared_error(
            y_true,
            predictions
        )
    )


    euclidean_errors = np.sqrt(
        np.sum(
            (
                predictions
                - y_true
            ) ** 2,
            axis=1
        )
    )


    return {

        "Coordinate_MAE":
            coordinate_mae,

        "Coordinate_RMSE":
            coordinate_rmse,

        "Mean_Localization_Error":
            np.mean(
                euclidean_errors
            ),

        "Median_Localization_Error":
            np.median(
                euclidean_errors
            ),

        "P90_Localization_Error":
            np.percentile(
                euclidean_errors,
                90
            )
    }


# ------------------------------------------------------------
# Step 13 — Run AP-Loss Robustness Experiments
#
# For every:
#
# - AP-loss level
# - random seed
# - classification model
# - regression model
#
# performance is measured.
#
# The clean condition is evaluated once per seed for consistent
# table structure.
# ------------------------------------------------------------

classification_records = []
regression_records = []


print("\n" + "=" * 76)
print("STEP 13 — RUNNING AP-LOSS EXPERIMENTS")
print("=" * 76)


experiment_start_time = (
    time.time()
)


for loss_fraction in AP_LOSS_LEVELS:

    print(
        f"\nAP loss: "
        f"{int(loss_fraction * 100)}%"
    )


    for seed in EXPERIMENT_SEEDS:


        # --------------------------------------------
        # Create corrupted raw RSSI fingerprints
        # --------------------------------------------

        X_corrupted_raw = apply_ap_loss(

            X_dev_raw,

            observed_mask_dev,

            loss_fraction=loss_fraction,

            seed=seed
        )


        # --------------------------------------------
        # Transform using ORIGINAL training scaler
        # --------------------------------------------

        X_corrupted_scaled = (
            rssi_scaler.transform(
                X_corrupted_raw
            )
        )


        # --------------------------------------------
        # Classification models
        # --------------------------------------------

        for (
            model_name,
            model
        ) in classification_models.items():


            metrics = evaluate_classification(

                model,

                X_corrupted_scaled,

                y_class_dev
            )


            classification_records.append(
                {

                    "AP_Loss_Fraction":
                        loss_fraction,

                    "AP_Loss_Percent":
                        int(
                            loss_fraction
                            * 100
                        ),

                    "Seed":
                        seed,

                    "Model":
                        model_name,

                    **metrics
                }
            )


        # --------------------------------------------
        # Regression models
        # --------------------------------------------

        for (
            model_name,
            model
        ) in regression_models.items():


            metrics = evaluate_regression(

                model_name,

                model,

                X_corrupted_scaled,

                y_coord_dev
            )


            regression_records.append(
                {

                    "AP_Loss_Fraction":
                        loss_fraction,

                    "AP_Loss_Percent":
                        int(
                            loss_fraction
                            * 100
                        ),

                    "Seed":
                        seed,

                    "Model":
                        model_name,

                    **metrics
                }
            )


experiment_time = (
    time.time()
    - experiment_start_time
)


print(
    "\nExperiments completed in",
    round(
        experiment_time,
        2
    ),
    "seconds"
)


# ------------------------------------------------------------
# Step 14 — Create Detailed Result Tables
# ------------------------------------------------------------

classification_detailed = (
    pd.DataFrame(
        classification_records
    )
)


regression_detailed = (
    pd.DataFrame(
        regression_records
    )
)


print("\n" + "=" * 76)
print("STEP 14 — DETAILED CLASSIFICATION RESULTS")
print("=" * 76)


display(
    classification_detailed.head(
        12
    ).round(4)
)


print("\nDetailed regression results:")

display(
    regression_detailed.head(
        12
    ).round(4)
)


# ------------------------------------------------------------
# Step 15 — Save Detailed Results
# ------------------------------------------------------------

CLASS_DETAILED_FILE = (
    RESULTS_CSV_DIR
    / "04_ap_loss_classification_detailed.csv"
)


REG_DETAILED_FILE = (
    RESULTS_CSV_DIR
    / "04_ap_loss_regression_detailed.csv"
)


classification_detailed.to_csv(
    CLASS_DETAILED_FILE,
    index=False
)


regression_detailed.to_csv(
    REG_DETAILED_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 16 — Calculate Mean and Standard Deviation
#
# Repeated corruption runs allow us to report:
#
#       Mean ± Standard Deviation
#
# rather than one potentially lucky random corruption.
# ------------------------------------------------------------

classification_summary = (

    classification_detailed

    .groupby(
        [
            "Model",
            "AP_Loss_Percent"
        ],
        as_index=False
    )

    .agg(
        Accuracy_Mean=(
            "Accuracy",
            "mean"
        ),

        Accuracy_SD=(
            "Accuracy",
            "std"
        ),

        F1_Mean=(
            "F1_Macro",
            "mean"
        ),

        F1_SD=(
            "F1_Macro",
            "std"
        ),

        Precision_Mean=(
            "Precision_Macro",
            "mean"
        ),

        Recall_Mean=(
            "Recall_Macro",
            "mean"
        )
    )
)


regression_summary = (

    regression_detailed

    .groupby(
        [
            "Model",
            "AP_Loss_Percent"
        ],
        as_index=False
    )

    .agg(
        Coordinate_MAE_Mean=(
            "Coordinate_MAE",
            "mean"
        ),

        Coordinate_MAE_SD=(
            "Coordinate_MAE",
            "std"
        ),

        Coordinate_RMSE_Mean=(
            "Coordinate_RMSE",
            "mean"
        ),

        Mean_Localization_Error_Mean=(
            "Mean_Localization_Error",
            "mean"
        ),

        Mean_Localization_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),

        Median_Localization_Error_Mean=(
            "Median_Localization_Error",
            "mean"
        ),

        P90_Localization_Error_Mean=(
            "P90_Localization_Error",
            "mean"
        )
    )
)


# Clean condition is deterministic, therefore SD = 0

classification_summary = (
    classification_summary.fillna(0)
)

regression_summary = (
    regression_summary.fillna(0)
)


print("\n" + "=" * 76)
print("STEP 16 — CLASSIFICATION SUMMARY")
print("=" * 76)


display(
    classification_summary.round(4)
)


print("\nRegression summary:")

display(
    regression_summary.round(4)
)


# ------------------------------------------------------------
# Step 17 — Save Summary Tables
# ------------------------------------------------------------

CLASS_SUMMARY_FILE = (
    RESULTS_CSV_DIR
    / "04_ap_loss_classification_summary.csv"
)


REG_SUMMARY_FILE = (
    RESULTS_CSV_DIR
    / "04_ap_loss_regression_summary.csv"
)


classification_summary.to_csv(
    CLASS_SUMMARY_FILE,
    index=False
)


regression_summary.to_csv(
    REG_SUMMARY_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 18 — Calculate Classification Degradation
#
# Δ Accuracy =
#
# clean accuracy - corrupted accuracy
#
# Positive values represent performance degradation.
# ------------------------------------------------------------

classification_with_delta = (
    classification_summary.copy()
)


classification_with_delta[
    "Accuracy_Drop"
] = 0.0


classification_with_delta[
    "F1_Drop"
] = 0.0


for model_name in (
    classification_with_delta[
        "Model"
    ].unique()
):


    clean_row = (
        classification_with_delta[
            (
                classification_with_delta[
                    "Model"
                ]
                == model_name
            )
            &
            (
                classification_with_delta[
                    "AP_Loss_Percent"
                ]
                == 0
            )
        ]
    )


    clean_accuracy = (
        clean_row[
            "Accuracy_Mean"
        ].iloc[0]
    )


    clean_f1 = (
        clean_row[
            "F1_Mean"
        ].iloc[0]
    )


    model_mask = (
        classification_with_delta[
            "Model"
        ]
        == model_name
    )


    classification_with_delta.loc[
        model_mask,
        "Accuracy_Drop"
    ] = (

        clean_accuracy
        -
        classification_with_delta.loc[
            model_mask,
            "Accuracy_Mean"
        ]
    )


    classification_with_delta.loc[
        model_mask,
        "F1_Drop"
    ] = (

        clean_f1
        -
        classification_with_delta.loc[
            model_mask,
            "F1_Mean"
        ]
    )


# ------------------------------------------------------------
# Step 19 — Calculate Regression Degradation
#
# Δ Localization Error =
#
# corrupted error - clean error
#
# Positive values represent worse localization.
# ------------------------------------------------------------

regression_with_delta = (
    regression_summary.copy()
)


regression_with_delta[
    "Localization_Error_Increase"
] = 0.0


for model_name in (
    regression_with_delta[
        "Model"
    ].unique()
):


    clean_row = (
        regression_with_delta[
            (
                regression_with_delta[
                    "Model"
                ]
                == model_name
            )
            &
            (
                regression_with_delta[
                    "AP_Loss_Percent"
                ]
                == 0
            )
        ]
    )


    clean_error = (
        clean_row[
            "Mean_Localization_Error_Mean"
        ].iloc[0]
    )


    model_mask = (
        regression_with_delta[
            "Model"
        ]
        == model_name
    )


    regression_with_delta.loc[
        model_mask,
        "Localization_Error_Increase"
    ] = (

        regression_with_delta.loc[
            model_mask,
            "Mean_Localization_Error_Mean"
        ]
        -
        clean_error
    )


# Save degradation results

CLASS_DELTA_FILE = (
    RESULTS_CSV_DIR
    / "04_ap_loss_classification_degradation.csv"
)


REG_DELTA_FILE = (
    RESULTS_CSV_DIR
    / "04_ap_loss_regression_degradation.csv"
)


classification_with_delta.to_csv(
    CLASS_DELTA_FILE,
    index=False
)


regression_with_delta.to_csv(
    REG_DELTA_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 20 — Plot Accuracy vs AP Loss
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 6)
)


for model_name in (
    classification_summary[
        "Model"
    ].unique()
):


    model_data = (
        classification_summary[
            classification_summary[
                "Model"
            ]
            == model_name
        ]
        .sort_values(
            "AP_Loss_Percent"
        )
    )


    plt.errorbar(

        model_data[
            "AP_Loss_Percent"
        ],

        model_data[
            "Accuracy_Mean"
        ],

        yerr=model_data[
            "Accuracy_SD"
        ],

        marker="o",

        capsize=3,

        label=model_name
    )


plt.xlabel(
    "Detected Access Points Removed (%)"
)

plt.ylabel(
    "Building-Floor Classification Accuracy"
)

plt.title(
    "Classification Robustness Under Access-Point Loss"
)

plt.ylim(
    0,
    1.05
)

plt.grid(
    alpha=0.2
)

plt.legend()

plt.tight_layout()


ACCURACY_FIGURE_FILE = (
    RESULTS_FIGURES_DIR
    / "04_accuracy_vs_ap_loss.png"
)


plt.savefig(
    ACCURACY_FIGURE_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 21 — Plot Macro F1 vs AP Loss
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 6)
)


for model_name in (
    classification_summary[
        "Model"
    ].unique()
):


    model_data = (
        classification_summary[
            classification_summary[
                "Model"
            ]
            == model_name
        ]
        .sort_values(
            "AP_Loss_Percent"
        )
    )


    plt.errorbar(

        model_data[
            "AP_Loss_Percent"
        ],

        model_data[
            "F1_Mean"
        ],

        yerr=model_data[
            "F1_SD"
        ],

        marker="o",

        capsize=3,

        label=model_name
    )


plt.xlabel(
    "Detected Access Points Removed (%)"
)

plt.ylabel(
    "Macro F1 Score"
)

plt.title(
    "Macro F1 Robustness Under Access-Point Loss"
)

plt.ylim(
    0,
    1.05
)

plt.grid(
    alpha=0.2
)

plt.legend()

plt.tight_layout()


F1_FIGURE_FILE = (
    RESULTS_FIGURES_DIR
    / "04_f1_vs_ap_loss.png"
)


plt.savefig(
    F1_FIGURE_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 22 — Plot Mean Localization Error vs AP Loss
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 6)
)


for model_name in (
    regression_summary[
        "Model"
    ].unique()
):


    model_data = (
        regression_summary[
            regression_summary[
                "Model"
            ]
            == model_name
        ]
        .sort_values(
            "AP_Loss_Percent"
        )
    )


    plt.errorbar(

        model_data[
            "AP_Loss_Percent"
        ],

        model_data[
            "Mean_Localization_Error_Mean"
        ],

        yerr=model_data[
            "Mean_Localization_Error_SD"
        ],

        marker="o",

        capsize=3,

        label=model_name
    )


plt.xlabel(
    "Detected Access Points Removed (%)"
)

plt.ylabel(
    "Mean Localization Error "
    "(dataset coordinate units)"
)

plt.title(
    "Coordinate Localization Robustness "
    "Under Access-Point Loss"
)

plt.grid(
    alpha=0.2
)

plt.legend()

plt.tight_layout()


LOCALIZATION_FIGURE_FILE = (
    RESULTS_FIGURES_DIR
    / "04_localization_error_vs_ap_loss.png"
)


plt.savefig(
    LOCALIZATION_FIGURE_FILE,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# Step 23 — Identify Model With Lowest Classification
# Degradation at 50% AP Loss
# ------------------------------------------------------------

classification_50 = (
    classification_with_delta[
        classification_with_delta[
            "AP_Loss_Percent"
        ]
        == 50
    ]
    .copy()
)


best_robust_classifier_row = (
    classification_50.loc[
        classification_50[
            "Accuracy_Drop"
        ].idxmin()
    ]
)


# ------------------------------------------------------------
# Step 24 — Identify Coordinate Model With Lowest Error
# at 50% AP Loss
# ------------------------------------------------------------

regression_50 = (
    regression_with_delta[
        regression_with_delta[
            "AP_Loss_Percent"
        ]
        == 50
    ]
    .copy()
)


best_robust_regressor_row = (
    regression_50.loc[
        regression_50[
            "Mean_Localization_Error_Mean"
        ].idxmin()
    ]
)


# ------------------------------------------------------------
# Step 25 — Create Key 50% AP-Loss Result Table
# ------------------------------------------------------------

key_50_classification = (
    classification_50[
        [
            "Model",
            "Accuracy_Mean",
            "Accuracy_SD",
            "F1_Mean",
            "F1_SD",
            "Accuracy_Drop",
            "F1_Drop"
        ]
    ]
)


key_50_regression = (
    regression_50[
        [
            "Model",
            "Mean_Localization_Error_Mean",
            "Mean_Localization_Error_SD",
            "Median_Localization_Error_Mean",
            "P90_Localization_Error_Mean",
            "Localization_Error_Increase"
        ]
    ]
)


print("\n" + "=" * 76)
print("STEP 25 — RESULTS AT 50% DETECTED AP LOSS")
print("=" * 76)


print("\nClassification:")

display(
    key_50_classification.round(4)
)


print("\nCoordinate localization:")

display(
    key_50_regression.round(4)
)


# ------------------------------------------------------------
# Step 26 — Final Experimental Summary
# ------------------------------------------------------------

print("\n" + "=" * 76)
print("NOTEBOOK 04 COMPLETED SUCCESSFULLY")
print("=" * 76)


print("\nExperiment design:")

print(
    "AP-loss levels:",
    [
        f"{int(x * 100)}%"
        for x in AP_LOSS_LEVELS
    ]
)

print(
    "Repeated corruption seeds:",
    len(EXPERIMENT_SEEDS)
)

print(
    "AP loss applied only to originally "
    "detected access points."
)


print("\nMOST ROBUST CLASSIFIER AT 50% AP LOSS")

print(
    "Model:",
    best_robust_classifier_row[
        "Model"
    ]
)

print(
    "Accuracy:",
    round(
        float(
            best_robust_classifier_row[
                "Accuracy_Mean"
            ]
        ),
        4
    )
)

print(
    "Accuracy SD:",
    round(
        float(
            best_robust_classifier_row[
                "Accuracy_SD"
            ]
        ),
        4
    )
)

print(
    "Accuracy drop from clean:",
    round(
        float(
            best_robust_classifier_row[
                "Accuracy_Drop"
            ]
        ),
        4
    )
)

print(
    "Macro F1:",
    round(
        float(
            best_robust_classifier_row[
                "F1_Mean"
            ]
        ),
        4
    )
)


print("\nLOWEST LOCALIZATION ERROR AT 50% AP LOSS")

print(
    "Model:",
    best_robust_regressor_row[
        "Model"
    ]
)

print(
    "Mean localization error:",
    round(
        float(
            best_robust_regressor_row[
                "Mean_Localization_Error_Mean"
            ]
        ),
        4
    )
)

print(
    "Localization error SD:",
    round(
        float(
            best_robust_regressor_row[
                "Mean_Localization_Error_SD"
            ]
        ),
        4
    )
)

print(
    "Increase from clean:",
    round(
        float(
            best_robust_regressor_row[
                "Localization_Error_Increase"
            ]
        ),
        4
    )
)


print("\nSaved CSV files:")

print(
    "1.",
    CLASS_DETAILED_FILE
)

print(
    "2.",
    REG_DETAILED_FILE
)

print(
    "3.",
    CLASS_SUMMARY_FILE
)

print(
    "4.",
    REG_SUMMARY_FILE
)

print(
    "5.",
    CLASS_DELTA_FILE
)

print(
    "6.",
    REG_DELTA_FILE
)


print("\nSaved figures:")

print(
    "1.",
    ACCURACY_FIGURE_FILE
)

print(
    "2.",
    F1_FIGURE_FILE
)

print(
    "3.",
    LOCALIZATION_FIGURE_FILE
)


print("\nNotebook 04 is complete.")

print(
    "Next notebook: "
    "05_rssi_noise_experiments.ipynb"
)
```
