```python
# ============================================================
# DEN702 Research Paper
# Notebook 06 — Combined AP-Loss + RSSI-Noise Experiments
#
# Research Topic:
# Robust Wi-Fi Fingerprint-Based Indoor Localization
# Under Access-Point Loss and RSSI Noise
#
# Purpose:
# 1. Evaluate simultaneous AP loss and RSSI noise
# 2. Test all baseline models under combined degradation
# 3. Repeat stochastic experiments across multiple seeds
# 4. Measure classification and coordinate-localization changes
# 5. Generate 2D robustness heatmaps
# 6. Save detailed and summarized results
#
# IMPORTANT:
# - Models are NOT retrained.
# - Only development data is corrupted.
# - AP loss affects originally detected APs only.
# - RSSI noise affects only APs still visible after AP loss.
# - Missing/dropped APs remain exactly -110 dBm.
# - Final test set remains untouched.
# ============================================================


# ------------------------------------------------------------
# Step 1 — Import Required Libraries
# ------------------------------------------------------------

from pathlib import Path
import time
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from IPython.display import display

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    mean_absolute_error,
    mean_squared_error
)


# ------------------------------------------------------------
# Step 2 — Experiment Configuration
# ------------------------------------------------------------

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


AP_LOSS_LEVELS = [
    0.00,
    0.10,
    0.20,
    0.30,
    0.40,
    0.50
]


RSSI_NOISE_LEVELS = [
    0,
    2,
    5,
    10
]


EXPERIMENT_SEEDS = [
    42,
    43,
    44,
    45,
    46
]


MISSING_RSSI = -110

OBSERVED_MIN_RSSI = -104
OBSERVED_MAX_RSSI = 0


print("=" * 78)
print("DEN702 — COMBINED DEGRADATION EXPERIMENT")
print("=" * 78)

print(
    "AP-loss levels:",
    [
        f"{int(x * 100)}%"
        for x in AP_LOSS_LEVELS
    ]
)

print(
    "RSSI-noise levels:",
    RSSI_NOISE_LEVELS
)

print(
    "Seeds:",
    EXPERIMENT_SEEDS
)


# ------------------------------------------------------------
# Step 3 — Detect Project Root
# ------------------------------------------------------------

if Path("../data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path("..").resolve()

elif Path("data/processed/uji_preprocessed.npz").exists():

    PROJECT_ROOT = Path(".").resolve()

else:

    raise FileNotFoundError(
        "Could not locate uji_preprocessed.npz."
    )


PROCESSED_DATA_DIR = (
    PROJECT_ROOT / "data" / "processed"
)

MODEL_DIR = (
    PROJECT_ROOT / "models" / "03_clean_baselines"
)

RESULTS_CSV_DIR = (
    PROJECT_ROOT / "results" / "csv"
)

RESULTS_FIGURES_DIR = (
    PROJECT_ROOT / "results" / "figures"
)


RESULTS_CSV_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("\nProject root:")
print(PROJECT_ROOT)


# ------------------------------------------------------------
# Step 4 — Load Development Dataset
# ------------------------------------------------------------

data = np.load(
    PROCESSED_DATA_DIR
    / "uji_preprocessed.npz",
    allow_pickle=True
)


X_dev_raw = (
    data["X_dev_raw"]
)

observed_mask_dev = (
    data["observed_mask_dev"]
)

y_class_dev = (
    data["y_class_dev"]
)

y_coord_dev = (
    data["y_coord_dev"]
)


print("\n" + "=" * 78)
print("STEP 4 — DEVELOPMENT DATA")
print("=" * 78)

print(
    "RSSI shape:",
    X_dev_raw.shape
)

print(
    "Observed mask:",
    observed_mask_dev.shape
)

print(
    "Classification target:",
    y_class_dev.shape
)

print(
    "Coordinate target:",
    y_coord_dev.shape
)


# ------------------------------------------------------------
# Step 5 — Load RSSI Scaler
#
# Always reuse the scaler fitted using training data only.
# ------------------------------------------------------------

rssi_scaler = joblib.load(
    PROCESSED_DATA_DIR
    / "uji_minmax_scaler.joblib"
)


# ------------------------------------------------------------
# Step 6 — Load Classification Models
# ------------------------------------------------------------

classification_models = {

    "KNN":

        joblib.load(
            MODEL_DIR
            / "knn_classifier.joblib"
        ),

    "Random Forest":

        joblib.load(
            MODEL_DIR
            / "random_forest_classifier.joblib"
        ),

    "MLP":

        joblib.load(
            MODEL_DIR
            / "mlp_classifier.joblib"
        )
}


print(
    "\nClassification models:",
    list(
        classification_models.keys()
    )
)


# ------------------------------------------------------------
# Step 7 — Load Regression Models
# ------------------------------------------------------------

regression_models = {

    "KNN":

        joblib.load(
            MODEL_DIR
            / "knn_regressor.joblib"
        ),

    "Random Forest":

        joblib.load(
            MODEL_DIR
            / "random_forest_regressor.joblib"
        ),

    "MLP":

        joblib.load(
            MODEL_DIR
            / "mlp_regressor.joblib"
        )
}


coord_scaler = joblib.load(
    MODEL_DIR
    / "coordinate_scaler.joblib"
)


print(
    "Regression models:",
    list(
        regression_models.keys()
    )
)


# ------------------------------------------------------------
# Step 8 — AP-Loss Function
#
# AP loss is applied independently to each fingerprint.
#
# Only originally observed APs can be removed.
# ------------------------------------------------------------

def apply_ap_loss(
    X_raw,
    observed_mask,
    loss_fraction,
    seed
):

    if not 0 <= loss_fraction <= 1:

        raise ValueError(
            "loss_fraction must be between 0 and 1."
        )


    if loss_fraction == 0:

        return X_raw.copy()


    rng = np.random.default_rng(
        seed
    )


    X_corrupted = (
        X_raw.copy()
    )


    for row_index in range(
        X_corrupted.shape[0]
    ):


        detected_indices = np.flatnonzero(
            observed_mask[
                row_index
            ]
        )


        n_detected = len(
            detected_indices
        )


        if n_detected == 0:

            continue


        n_to_drop = int(
            np.floor(
                n_detected
                * loss_fraction
            )
        )


        # Ensure non-zero corruption has an actual effect

        if (
            loss_fraction > 0
            and n_to_drop == 0
        ):

            n_to_drop = 1


        n_to_drop = min(
            n_to_drop,
            n_detected
        )


        drop_indices = rng.choice(

            detected_indices,

            size=n_to_drop,

            replace=False
        )


        X_corrupted[
            row_index,
            drop_indices
        ] = MISSING_RSSI


    return X_corrupted


# ------------------------------------------------------------
# Step 9 — RSSI Noise Function
#
# Gaussian noise is applied only to access points that remain
# visible AFTER AP-loss corruption.
# ------------------------------------------------------------

def apply_rssi_noise_after_loss(
    X_after_loss,
    original_observed_mask,
    sigma,
    seed
):

    if sigma < 0:

        raise ValueError(
            "sigma must be non-negative."
        )


    if sigma == 0:

        return X_after_loss.copy()


    rng = np.random.default_rng(
        seed
    )


    X_noisy = (
        X_after_loss
        .copy()
        .astype(float)
    )


    # AP must:
    # 1. Have originally been observed
    # 2. Still be visible after AP dropout

    surviving_mask = (
        original_observed_mask
        &
        (
            X_after_loss
            != MISSING_RSSI
        )
    )


    noise = rng.normal(

        loc=0.0,

        scale=sigma,

        size=X_noisy.shape
    )


    X_noisy[
        surviving_mask
    ] = (

        X_noisy[
            surviving_mask
        ]

        +

        noise[
            surviving_mask
        ]
    )


    X_noisy[
        surviving_mask
    ] = np.clip(

        X_noisy[
            surviving_mask
        ],

        OBSERVED_MIN_RSSI,

        OBSERVED_MAX_RSSI
    )


    # Guarantee missing/dropped APs stay missing

    X_noisy[
        ~surviving_mask
    ] = MISSING_RSSI


    return X_noisy


# ------------------------------------------------------------
# Step 10 — Combined Corruption Function
# ------------------------------------------------------------

def apply_combined_corruption(
    X_raw,
    observed_mask,
    ap_loss_fraction,
    noise_sigma,
    seed
):

    X_after_loss = apply_ap_loss(

        X_raw,

        observed_mask,

        loss_fraction=ap_loss_fraction,

        seed=seed
    )


    # Different deterministic stream for noise

    noise_seed = (
        seed + 10000
    )


    X_combined = (
        apply_rssi_noise_after_loss(

            X_after_loss,

            observed_mask,

            sigma=noise_sigma,

            seed=noise_seed
        )
    )


    return X_combined


# ------------------------------------------------------------
# Step 11 — Validate Combined Corruption
# ------------------------------------------------------------

example_index = 0


example_combined = (
    apply_combined_corruption(

        X_dev_raw[
            example_index:
            example_index + 1
        ],

        observed_mask_dev[
            example_index:
            example_index + 1
        ],

        ap_loss_fraction=0.50,

        noise_sigma=10,

        seed=RANDOM_STATE
    )
)


original_count = int(
    observed_mask_dev[
        example_index
    ].sum()
)


remaining_count = int(
    (
        example_combined[0]
        != MISSING_RSSI
    ).sum()
)


print("\n" + "=" * 78)
print("STEP 11 — COMBINED CORRUPTION CHECK")
print("=" * 78)

print(
    "Original detected APs:",
    original_count
)

print(
    "Remaining after 50% AP loss:",
    remaining_count
)

print(
    "All dropped/missing APs remain -110:",
    bool(
        np.all(
            example_combined[
                example_combined
                == MISSING_RSSI
            ]
            == MISSING_RSSI
        )
    )
)


# ------------------------------------------------------------
# Step 12 — Classification Evaluation Function
# ------------------------------------------------------------

def evaluate_classification(
    model,
    X,
    y_true
):

    predictions = model.predict(
        X
    )


    return {

        "Accuracy":

            accuracy_score(
                y_true,
                predictions
            ),

        "Precision_Macro":

            precision_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            ),

        "Recall_Macro":

            recall_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            ),

        "F1_Macro":

            f1_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            )
    }


# ------------------------------------------------------------
# Step 13 — Regression Evaluation Function
# ------------------------------------------------------------

def evaluate_regression(
    model_name,
    model,
    X,
    y_true
):

    predictions = model.predict(
        X
    )


    if model_name == "MLP":

        predictions = (
            coord_scaler.inverse_transform(
                predictions
            )
        )


    coordinate_mae = (
        mean_absolute_error(
            y_true,
            predictions
        )
    )


    coordinate_rmse = np.sqrt(
        mean_squared_error(
            y_true,
            predictions
        )
    )


    euclidean_errors = np.sqrt(
        np.sum(
            (
                predictions
                - y_true
            ) ** 2,
            axis=1
        )
    )


    return {

        "Coordinate_MAE":

            coordinate_mae,

        "Coordinate_RMSE":

            coordinate_rmse,

        "Mean_Localization_Error":

            np.mean(
                euclidean_errors
            ),

        "Median_Localization_Error":

            np.median(
                euclidean_errors
            ),

        "P90_Localization_Error":

            np.percentile(
                euclidean_errors,
                90
            )
    }


# ------------------------------------------------------------
# Step 14 — Run Full Combined Degradation Grid
#
# Total corruption conditions:
#
# 6 AP levels × 4 noise levels = 24 conditions
#
# Each condition is repeated using 5 seeds.
# ------------------------------------------------------------

classification_records = []
regression_records = []


total_conditions = (
    len(AP_LOSS_LEVELS)
    *
    len(RSSI_NOISE_LEVELS)
)


print("\n" + "=" * 78)
print("STEP 14 — RUNNING COMBINED EXPERIMENTS")
print("=" * 78)

print(
    "Corruption conditions:",
    total_conditions
)

print(
    "Repeated seeds:",
    len(EXPERIMENT_SEEDS)
)


experiment_start = time.time()


condition_number = 0


for ap_loss in AP_LOSS_LEVELS:


    for sigma in RSSI_NOISE_LEVELS:


        condition_number += 1


        print(

            f"\nCondition "
            f"{condition_number}/{total_conditions}"

        )


        print(

            f"AP loss = "
            f"{int(ap_loss * 100)}%, "
            f"RSSI σ = {sigma} dB"

        )


        for seed in EXPERIMENT_SEEDS:


            X_corrupted_raw = (
                apply_combined_corruption(

                    X_dev_raw,

                    observed_mask_dev,

                    ap_loss_fraction=ap_loss,

                    noise_sigma=sigma,

                    seed=seed
                )
            )


            X_corrupted_scaled = (
                rssi_scaler.transform(
                    X_corrupted_raw
                )
            )


            # ----------------------------------------
            # Classification
            # ----------------------------------------

            for (
                model_name,
                model
            ) in classification_models.items():


                metrics = (
                    evaluate_classification(

                        model,

                        X_corrupted_scaled,

                        y_class_dev
                    )
                )


                classification_records.append(
                    {

                        "AP_Loss_Fraction":
                            ap_loss,

                        "AP_Loss_Percent":
                            int(
                                ap_loss
                                * 100
                            ),

                        "RSSI_Noise_Sigma_dB":
                            sigma,

                        "Seed":
                            seed,

                        "Model":
                            model_name,

                        **metrics
                    }
                )


            # ----------------------------------------
            # Regression
            # ----------------------------------------

            for (
                model_name,
                model
            ) in regression_models.items():


                metrics = (
                    evaluate_regression(

                        model_name,

                        model,

                        X_corrupted_scaled,

                        y_coord_dev
                    )
                )


                regression_records.append(
                    {

                        "AP_Loss_Fraction":
                            ap_loss,

                        "AP_Loss_Percent":
                            int(
                                ap_loss
                                * 100
                            ),

                        "RSSI_Noise_Sigma_dB":
                            sigma,

                        "Seed":
                            seed,

                        "Model":
                            model_name,

                        **metrics
                    }
                )


experiment_time = (
    time.time()
    - experiment_start
)


print(
    "\nCombined experiment completed in",
    round(
        experiment_time,
        2
    ),
    "seconds"
)


# ------------------------------------------------------------
# Step 15 — Create Detailed Results
# ------------------------------------------------------------

classification_detailed = (
    pd.DataFrame(
        classification_records
    )
)


regression_detailed = (
    pd.DataFrame(
        regression_records
    )
)


print("\n" + "=" * 78)
print("STEP 15 — DETAILED RESULTS")
print("=" * 78)


print("\nClassification preview:")

display(
    classification_detailed.head(
        12
    ).round(4)
)


print("\nRegression preview:")

display(
    regression_detailed.head(
        12
    ).round(4)
)


# ------------------------------------------------------------
# Step 16 — Save Detailed Results
# ------------------------------------------------------------

CLASS_DETAILED_FILE = (
    RESULTS_CSV_DIR
    / "06_combined_classification_detailed.csv"
)


REG_DETAILED_FILE = (
    RESULTS_CSV_DIR
    / "06_combined_regression_detailed.csv"
)


classification_detailed.to_csv(
    CLASS_DETAILED_FILE,
    index=False
)


regression_detailed.to_csv(
    REG_DETAILED_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 17 — Summarize Classification Results
# ------------------------------------------------------------

classification_summary = (

    classification_detailed

    .groupby(
        [
            "Model",
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ],
        as_index=False
    )

    .agg(

        Accuracy_Mean=(
            "Accuracy",
            "mean"
        ),

        Accuracy_SD=(
            "Accuracy",
            "std"
        ),

        Precision_Mean=(
            "Precision_Macro",
            "mean"
        ),

        Recall_Mean=(
            "Recall_Macro",
            "mean"
        ),

        F1_Mean=(
            "F1_Macro",
            "mean"
        ),

        F1_SD=(
            "F1_Macro",
            "std"
        )
    )
)


classification_summary = (
    classification_summary.fillna(0)
)


# ------------------------------------------------------------
# Step 18 — Summarize Regression Results
# ------------------------------------------------------------

regression_summary = (

    regression_detailed

    .groupby(
        [
            "Model",
            "AP_Loss_Percent",
            "RSSI_Noise_Sigma_dB"
        ],
        as_index=False
    )

    .agg(

        Coordinate_MAE_Mean=(
            "Coordinate_MAE",
            "mean"
        ),

        Coordinate_MAE_SD=(
            "Coordinate_MAE",
            "std"
        ),

        Coordinate_RMSE_Mean=(
            "Coordinate_RMSE",
            "mean"
        ),

        Mean_Localization_Error_Mean=(
            "Mean_Localization_Error",
            "mean"
        ),

        Mean_Localization_Error_SD=(
            "Mean_Localization_Error",
            "std"
        ),

        Median_Localization_Error_Mean=(
            "Median_Localization_Error",
            "mean"
        ),

        P90_Localization_Error_Mean=(
            "P90_Localization_Error",
            "mean"
        )
    )
)


regression_summary = (
    regression_summary.fillna(0)
)


print("\nClassification summary:")

display(
    classification_summary.round(4)
)


print("\nRegression summary:")

display(
    regression_summary.round(4)
)


# ------------------------------------------------------------
# Step 19 — Save Summary Tables
# ------------------------------------------------------------

CLASS_SUMMARY_FILE = (
    RESULTS_CSV_DIR
    / "06_combined_classification_summary.csv"
)


REG_SUMMARY_FILE = (
    RESULTS_CSV_DIR
    / "06_combined_regression_summary.csv"
)


classification_summary.to_csv(
    CLASS_SUMMARY_FILE,
    index=False
)


regression_summary.to_csv(
    REG_SUMMARY_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 20 — Extract Clean Baselines
#
# Clean condition:
#
# AP loss = 0%
# RSSI noise sigma = 0 dB
# ------------------------------------------------------------

clean_classification = (
    classification_summary[
        (
            classification_summary[
                "AP_Loss_Percent"
            ]
            == 0
        )
        &
        (
            classification_summary[
                "RSSI_Noise_Sigma_dB"
            ]
            == 0
        )
    ]
)


clean_regression = (
    regression_summary[
        (
            regression_summary[
                "AP_Loss_Percent"
            ]
            == 0
        )
        &
        (
            regression_summary[
                "RSSI_Noise_Sigma_dB"
            ]
            == 0
        )
    ]
)


# ------------------------------------------------------------
# Step 21 — Calculate Classification Degradation
# ------------------------------------------------------------

classification_delta = (
    classification_summary.copy()
)


classification_delta[
    "Accuracy_Drop"
] = 0.0


classification_delta[
    "F1_Drop"
] = 0.0


for model_name in (
    classification_delta[
        "Model"
    ].unique()
):


    clean_row = (
        clean_classification[
            clean_classification[
                "Model"
            ]
            == model_name
        ]
    )


    clean_accuracy = float(
        clean_row[
            "Accuracy_Mean"
        ].iloc[0]
    )


    clean_f1 = float(
        clean_row[
            "F1_Mean"
        ].iloc[0]
    )


    mask = (
        classification_delta[
            "Model"
        ]
        == model_name
    )


    classification_delta.loc[
        mask,
        "Accuracy_Drop"
    ] = (

        clean_accuracy

        -

        classification_delta.loc[
            mask,
            "Accuracy_Mean"
        ]
    )


    classification_delta.loc[
        mask,
        "F1_Drop"
    ] = (

        clean_f1

        -

        classification_delta.loc[
            mask,
            "F1_Mean"
        ]
    )


# ------------------------------------------------------------
# Step 22 — Calculate Localization Degradation
# ------------------------------------------------------------

regression_delta = (
    regression_summary.copy()
)


regression_delta[
    "Localization_Error_Increase"
] = 0.0


for model_name in (
    regression_delta[
        "Model"
    ].unique()
):


    clean_row = (
        clean_regression[
            clean_regression[
                "Model"
            ]
            == model_name
        ]
    )


    clean_error = float(
        clean_row[
            "Mean_Localization_Error_Mean"
        ].iloc[0]
    )


    mask = (
        regression_delta[
            "Model"
        ]
        == model_name
    )


    regression_delta.loc[
        mask,
        "Localization_Error_Increase"
    ] = (

        regression_delta.loc[
            mask,
            "Mean_Localization_Error_Mean"
        ]

        -

        clean_error
    )


# ------------------------------------------------------------
# Step 23 — Save Degradation Tables
# ------------------------------------------------------------

CLASS_DELTA_FILE = (
    RESULTS_CSV_DIR
    / "06_combined_classification_degradation.csv"
)


REG_DELTA_FILE = (
    RESULTS_CSV_DIR
    / "06_combined_regression_degradation.csv"
)


classification_delta.to_csv(
    CLASS_DELTA_FILE,
    index=False
)


regression_delta.to_csv(
    REG_DELTA_FILE,
    index=False
)


# ============================================================
# PART C — ROBUSTNESS HEATMAPS
# ============================================================


# ------------------------------------------------------------
# Step 24 — Classification Accuracy Heatmaps
#
# One heatmap is generated for each classifier.
# ------------------------------------------------------------

CLASS_HEATMAP_FILES = []


for model_name in (
    classification_summary[
        "Model"
    ].unique()
):


    model_data = (
        classification_summary[
            classification_summary[
                "Model"
            ]
            == model_name
        ]
    )


    heatmap = (
        model_data.pivot(

            index="RSSI_Noise_Sigma_dB",

            columns="AP_Loss_Percent",

            values="Accuracy_Mean"
        )
        .sort_index()
        .sort_index(axis=1)
    )


    fig, ax = plt.subplots(
        figsize=(9, 5)
    )


    image = ax.imshow(
        heatmap.values,
        aspect="auto",
        origin="lower"
    )


    ax.set_xticks(
        range(
            len(
                heatmap.columns
            )
        )
    )

    ax.set_xticklabels(
        heatmap.columns
    )


    ax.set_yticks(
        range(
            len(
                heatmap.index
            )
        )
    )

    ax.set_yticklabels(
        heatmap.index
    )


    ax.set_xlabel(
        "Detected APs Removed (%)"
    )

    ax.set_ylabel(
        "RSSI Noise σ (dB)"
    )


    ax.set_title(
        f"Classification Accuracy — {model_name}"
    )


    for row in range(
        heatmap.shape[0]
    ):

        for col in range(
            heatmap.shape[1]
        ):

            ax.text(

                col,

                row,

                f"{heatmap.iloc[row, col]:.3f}",

                ha="center",

                va="center"
            )


    fig.colorbar(
        image,
        ax=ax,
        label="Accuracy"
    )


    plt.tight_layout()


    safe_name = (
        model_name
        .lower()
        .replace(
            " ",
            "_"
        )
    )


    figure_file = (
        RESULTS_FIGURES_DIR
        /
        f"06_accuracy_heatmap_{safe_name}.png"
    )


    plt.savefig(
        figure_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


    CLASS_HEATMAP_FILES.append(
        figure_file
    )


# ------------------------------------------------------------
# Step 25 — Localization Error Heatmaps
# ------------------------------------------------------------

REG_HEATMAP_FILES = []


for model_name in (
    regression_summary[
        "Model"
    ].unique()
):


    model_data = (
        regression_summary[
            regression_summary[
                "Model"
            ]
            == model_name
        ]
    )


    heatmap = (
        model_data.pivot(

            index="RSSI_Noise_Sigma_dB",

            columns="AP_Loss_Percent",

            values="Mean_Localization_Error_Mean"
        )
        .sort_index()
        .sort_index(axis=1)
    )


    fig, ax = plt.subplots(
        figsize=(9, 5)
    )


    image = ax.imshow(
        heatmap.values,
        aspect="auto",
        origin="lower"
    )


    ax.set_xticks(
        range(
            len(
                heatmap.columns
            )
        )
    )


    ax.set_xticklabels(
        heatmap.columns
    )


    ax.set_yticks(
        range(
            len(
                heatmap.index
            )
        )
    )


    ax.set_yticklabels(
        heatmap.index
    )


    ax.set_xlabel(
        "Detected APs Removed (%)"
    )

    ax.set_ylabel(
        "RSSI Noise σ (dB)"
    )


    ax.set_title(
        f"Mean Localization Error — {model_name}"
    )


    for row in range(
        heatmap.shape[0]
    ):

        for col in range(
            heatmap.shape[1]
        ):

            ax.text(

                col,

                row,

                f"{heatmap.iloc[row, col]:.2f}",

                ha="center",

                va="center"
            )


    fig.colorbar(
        image,
        ax=ax,
        label="Mean Localization Error"
    )


    plt.tight_layout()


    safe_name = (
        model_name
        .lower()
        .replace(
            " ",
            "_"
        )
    )


    figure_file = (
        RESULTS_FIGURES_DIR
        /
        f"06_localization_heatmap_{safe_name}.png"
    )


    plt.savefig(
        figure_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


    REG_HEATMAP_FILES.append(
        figure_file
    )


# ------------------------------------------------------------
# Step 26 — Extract Most Severe Combined Condition
#
# AP loss = 50%
# RSSI noise sigma = 10 dB
# ------------------------------------------------------------

severe_classification = (
    classification_delta[
        (
            classification_delta[
                "AP_Loss_Percent"
            ]
            == 50
        )
        &
        (
            classification_delta[
                "RSSI_Noise_Sigma_dB"
            ]
            == 10
        )
    ]
    .copy()
)


severe_regression = (
    regression_delta[
        (
            regression_delta[
                "AP_Loss_Percent"
            ]
            == 50
        )
        &
        (
            regression_delta[
                "RSSI_Noise_Sigma_dB"
            ]
            == 10
        )
    ]
    .copy()
)


print("\n" + "=" * 78)
print("SEVERE CONDITION — 50% AP LOSS + σ = 10 dB")
print("=" * 78)


print("\nClassification:")

display(
    severe_classification[
        [
            "Model",
            "Accuracy_Mean",
            "Accuracy_SD",
            "F1_Mean",
            "F1_SD",
            "Accuracy_Drop",
            "F1_Drop"
        ]
    ].round(4)
)


print("\nCoordinate localization:")

display(
    severe_regression[
        [
            "Model",
            "Mean_Localization_Error_Mean",
            "Mean_Localization_Error_SD",
            "Median_Localization_Error_Mean",
            "P90_Localization_Error_Mean",
            "Localization_Error_Increase"
        ]
    ].round(4)
)


# ------------------------------------------------------------
# Step 27 — Identify Strongest Severe-Condition Models
# ------------------------------------------------------------

best_severe_classifier = (
    severe_classification.loc[
        severe_classification[
            "Accuracy_Mean"
        ].idxmax()
    ]
)


best_severe_regressor = (
    severe_regression.loc[
        severe_regression[
            "Mean_Localization_Error_Mean"
        ].idxmin()
    ]
)


# ------------------------------------------------------------
# Step 28 — Compare Single vs Combined Degradation
#
# This comparison helps determine whether RSSI noise adds
# substantial degradation beyond AP loss alone.
# ------------------------------------------------------------

comparison_rows = []


for model_name in (
    classification_summary[
        "Model"
    ].unique()
):


    ap_only = (
        classification_summary[
            (
                classification_summary[
                    "Model"
                ]
                == model_name
            )
            &
            (
                classification_summary[
                    "AP_Loss_Percent"
                ]
                == 50
            )
            &
            (
                classification_summary[
                    "RSSI_Noise_Sigma_dB"
                ]
                == 0
            )
        ][
            "Accuracy_Mean"
        ].iloc[0]
    )


    noise_only = (
        classification_summary[
            (
                classification_summary[
                    "Model"
                ]
                == model_name
            )
            &
            (
                classification_summary[
                    "AP_Loss_Percent"
                ]
                == 0
            )
            &
            (
                classification_summary[
                    "RSSI_Noise_Sigma_dB"
                ]
                == 10
            )
        ][
            "Accuracy_Mean"
        ].iloc[0]
    )


    combined = (
        classification_summary[
            (
                classification_summary[
                    "Model"
                ]
                == model_name
            )
            &
            (
                classification_summary[
                    "AP_Loss_Percent"
                ]
                == 50
            )
            &
            (
                classification_summary[
                    "RSSI_Noise_Sigma_dB"
                ]
                == 10
            )
        ][
            "Accuracy_Mean"
        ].iloc[0]
    )


    comparison_rows.append(
        {

            "Model":
                model_name,

            "50pct_AP_Loss_Only":
                ap_only,

            "Sigma10_Noise_Only":
                noise_only,

            "Combined_50pct_Sigma10":
                combined
        }
    )


combined_comparison = (
    pd.DataFrame(
        comparison_rows
    )
)


print("\n" + "=" * 78)
print("STEP 28 — SINGLE VS COMBINED CLASSIFICATION")
print("=" * 78)


display(
    combined_comparison.round(4)
)


COMPARISON_FILE = (
    RESULTS_CSV_DIR
    / "06_single_vs_combined_classification.csv"
)


combined_comparison.to_csv(
    COMPARISON_FILE,
    index=False
)


# ------------------------------------------------------------
# Step 29 — Final Summary
# ------------------------------------------------------------

print("\n" + "=" * 78)
print("NOTEBOOK 06 COMPLETED SUCCESSFULLY")
print("=" * 78)


print("\nExperiment design:")

print(
    "AP-loss levels:",
    [
        f"{int(x * 100)}%"
        for x in AP_LOSS_LEVELS
    ]
)

print(
    "RSSI-noise levels:",
    RSSI_NOISE_LEVELS
)

print(
    "Repeated seeds:",
    len(EXPERIMENT_SEEDS)
)

print(
    "Total degradation combinations:",
    total_conditions
)


print(
    "\nSTRONGEST CLASSIFIER UNDER "
    "50% AP LOSS + σ = 10 dB"
)

print(
    "Model:",
    best_severe_classifier[
        "Model"
    ]
)

print(
    "Accuracy:",
    round(
        float(
            best_severe_classifier[
                "Accuracy_Mean"
            ]
        ),
        4
    )
)

print(
    "Accuracy SD:",
    round(
        float(
            best_severe_classifier[
                "Accuracy_SD"
            ]
        ),
        4
    )
)

print(
    "Accuracy drop from clean:",
    round(
        float(
            best_severe_classifier[
                "Accuracy_Drop"
            ]
        ),
        4
    )
)

print(
    "Macro F1:",
    round(
        float(
            best_severe_classifier[
                "F1_Mean"
            ]
        ),
        4
    )
)


print(
    "\nLOWEST LOCALIZATION ERROR UNDER "
    "50% AP LOSS + σ = 10 dB"
)

print(
    "Model:",
    best_severe_regressor[
        "Model"
    ]
)

print(
    "Mean localization error:",
    round(
        float(
            best_severe_regressor[
                "Mean_Localization_Error_Mean"
            ]
        ),
        4
    )
)

print(
    "Localization error SD:",
    round(
        float(
            best_severe_regressor[
                "Mean_Localization_Error_SD"
            ]
        ),
        4
    )
)

print(
    "Increase from clean:",
    round(
        float(
            best_severe_regressor[
                "Localization_Error_Increase"
            ]
        ),
        4
    )
)


print("\nSaved result files:")

print(
    "1.",
    CLASS_DETAILED_FILE
)

print(
    "2.",
    REG_DETAILED_FILE
)

print(
    "3.",
    CLASS_SUMMARY_FILE
)

print(
    "4.",
    REG_SUMMARY_FILE
)

print(
    "5.",
    CLASS_DELTA_FILE
)

print(
    "6.",
    REG_DELTA_FILE
)

print(
    "7.",
    COMPARISON_FILE
)


print("\nClassification heatmaps:")

for file in CLASS_HEATMAP_FILES:

    print(
        "-",
        file
    )


print("\nLocalization heatmaps:")

for file in REG_HEATMAP_FILES:

    print(
        "-",
        file
    )


print("\nNotebook 06 is complete.")

print(
    "Next notebook: "
    "07_robust_training.ipynb"
)
```
